# Caso Studio: Lehman Brothers 2008: rischio sistemico, migrazioni creditizie e perdita di portafoglio

## 1. Contesto e Domanda Quantitativa

**Contesto:** Il caso si colloca nel giugno 2008, in una fase di crescente tensione del sistema finanziario internazionale e di deterioramento del merito creditizio di Lehman Brothers. L'obiettivo è sviluppare, mediante simulazione Monte Carlo, un modello di rischio creditizio di portafoglio in cui le migrazioni dei singoli debitori dipendono da un regime sistemico comune che evolve nel tempo. Il portafoglio è composto da 40 posizioni, con una forte concentrazione iniziale su Lehman Brothers.

**Domanda quantitativa:** Quale distribuzione della perdita a un anno produce il portafoglio concentrato e come cambiano perdita attesa, dispersione, VaR e CVaR quando l'investimento inizialmente concentrato su Lehman Brothers viene riallocato proporzionalmente tra gli altri debitori, a investimento complessivo invariato?

**Variabile finale:** Le variabili di interesse sono le perdite simulate del portafoglio concentrato ($L_{\mathrm{conc}}^{P}$) e del portafoglio controfattuale ($L_{\mathrm{div}}^{P}$) su un orizzonte di un anno ($T=4$ trimestri).

**Informazione disponibile:** Sono forniti i portafogli iniziali (concentrato e controfattuale, per un totale di 173 milioni di dollari), la matrice di transizione del regime sistemico ($Q$) su tre stati (Ordinario, Stress, Crisi), le matrici di migrazione creditizia condizionate al regime ($P^{(O)}$, $P^{(S)}$, $P^{(C)}$) e i tassi di perdita migration-based ($\varphi$). Il regime iniziale è di stress ($M_0 = S$) e il numero di repliche Monte Carlo è fissato a $N_{\mathrm{MC}} = 50\,000$ con un seed pari a 2026.

## 2. Quantità da stimare o calcolare

Per il portafoglio concentrato e per il portafoglio controfattuale:

1.  la perdita media:
    $$
    \mathbb{E}[L^P]
    $$

2.  la varianza:
    $$
    \operatorname{Var}(L^P)
    $$

3.  la deviazione standard

4.  il Value at Risk ai livelli 95% e 99%:
    $$
    \operatorname{VaR}_{0,95}(L^P), \qquad \operatorname{VaR}_{0,99}(L^P)
    $$

5.  il Conditional Value at Risk ai livelli 95% e 99%:
    $$
    \operatorname{CVaR}_{0,95}(L^P), \qquad \operatorname{CVaR}_{0,99}(L^P)
    $$

Devono inoltre essere calcolate:

6.  la probabilità che il regime $C$ (Crisi) venga raggiunto almeno una volta durante l'orizzonte;

7.  la perdita media del portafoglio concentrato, distinta in due casi:
    *   nelle repliche in cui il regime $C$ viene raggiunto almeno una volta;
    *   nelle repliche in cui il regime $C$ non viene mai raggiunto;

8.  per ciascuna misura di rischio $\rho$, la variazione tra portafoglio diversificato e concentrato:
    $$
    \Delta\rho = \rho(L_{\mathrm{div}}^P) - \rho(L_{\mathrm{conc}}^P)
    $$

9.  la differenza replica per replica delle perdite:
    $$
    \Delta L^{(r)} = L_{\mathrm{div}}^{P,(r)} - L_{\mathrm{conc}}^{P,(r)}
    $$

# Flusso logico-teorico risolutivo

Il tuo inquadramento logico è solido, completo e corretto. Hai colto perfettamente i meccanismi di dipendenza condizionata e l'importanza del riutilizzo delle medesime repliche (common random numbers) per il confronto controfattuale. 

Rispondo subito alle tue due ottime osservazioni prima di formalizzare il flusso:
1. **Regime fino a $t=3$ e rating fino a $t=4$:** La tua intuizione è esatta. Poiché la transizione del rating nell'intervallo $(t, t+1)$ è governata dal regime sistemico all'istante $t$, per arrivare allo stato finale $X_{i,4}$ occorrono esattamente 4 transizioni dipendenti dai regimi $M_0, M_1, M_2, M_3$. Un eventuale stato $M_4$ non avrebbe alcun impatto sulle perdite a un anno.
2. **Massa di probabilità al VaR:** Osservazione cruciale. Nelle distribuzioni empiriche discrete, la probabilità che la perdita sia strettamente maggiore del VaR raramente coincide in modo esatto con $1-\alpha$. Per calcolare il CVaR in modo teoricamente corretto (coerentemente col Capitolo 9), non basta fare la media aritmetica delle perdite oltre il VaR, ma occorre utilizzare la definizione generalizzata che "spezza" la massa di probabilità esattamente sulla soglia del VaR.

### Sintesi logica della Tappa 1
- **Input:** Specifiche dei portafogli dalla Scheda Caso (composizione per gruppi, debitori, rating, investimenti individuali, e la frazione esatta di riallocazione pari a $173/148$).
- **Operazioni:** 
  1. Creazione di un DataFrame a livello di singolo debitore che contenga esattamente 40 righe.
  2. Applicazione del moltiplicatore esatto $173/148$ all'investimento iniziale dei 39 debitori ordinari per ricavare l'investimento riallocato, ponendo a zero la posizione Lehman.
  3. Raggruppamento delle posizioni individuali (groupby) per ricostruire i totali di gruppo e popolare le tabelle riassuntive del caso.
- **Output:** `Tabella 1` (riepilogo portafoglio concentrato), `Tabella 2` (riepilogo portafoglio controfattuale) e i vettori array degli investimenti dei due portafogli a livello di singola controparte.
- **Controlli:** 
  - Controllo 7: verifica che il portafoglio concentrato abbia esattamente 40 debitori.
  - Controllo 8: verifica che il portafoglio controfattuale (depurato da Lehman) abbia esattamente 39 debitori.
  - Controllo 9: verifica che la somma degli investimenti in entrambi i portafogli sia esattamente 173.
  - Controllo 10: verifica che il peso percentuale di ciascun debitore sui restanti 39 rimanga invariato nel passaggio dal concentrato al controfattuale.
- **Uso successivo:** I vettori degli investimenti calcolati costituiscono la "Exposure At Default" (EAD) necessaria, insieme all'ordine dei rating iniziali, come input per la Tappa 5 (calcolo delle perdite individuali a scadenza).

```python
import numpy as np
import pandas as pd
from IPython.display import display, HTML

# 1. Costruzione della struttura dati per singolo debitore
debitori_data = []

# Lehman Brothers (1 debitore)
debitori_data.append({
    'ID_Debitore': 'L',
    'Gruppo': 'Lehman Brothers',
    'Rating_Iniziale': 'A',
    'Investimento_Conc': 25.0
})

# Altri debitori A (11 debitori)
for i in range(1, 12):
    debitori_data.append({
        'ID_Debitore': f'A_{i}',
        'Gruppo': 'Altri debitori A',
        'Rating_Iniziale': 'A',
        'Investimento_Conc': 5.0
    })

# Debitori BBB (14 debitori)
for i in range(1, 15):
    debitori_data.append({
        'ID_Debitore': f'BBB_{i}',
        'Gruppo': 'Debitori BBB',
        'Rating_Iniziale': 'BBB',
        'Investimento_Conc': 4.0
    })

# Debitori BB (9 debitori)
for i in range(1, 10):
    debitori_data.append({
        'ID_Debitore': f'BB_{i}',
        'Gruppo': 'Debitori BB',
        'Rating_Iniziale': 'BB',
        'Investimento_Conc': 3.0
    })

# Debitori B (5 debitori)
for i in range(1, 6):
    debitori_data.append({
        'ID_Debitore': f'B_{i}',
        'Gruppo': 'Debitori B',
        'Rating_Iniziale': 'B',
        'Investimento_Conc': 2.0
    })

df_debitori = pd.DataFrame(debitori_data)

# 2. Calcolo del portafoglio controfattuale (fattore esatto 173/148)
fattore_riallocazione = 173.0 / 148.0

# L'investimento riallocato è calcolato solo per i 39 debitori non-Lehman
df_debitori['Investimento_Div'] = np.where(
    df_debitori['ID_Debitore'] == 'L',
    0.0,
    df_debitori['Investimento_Conc'] * fattore_riallocazione
)

# Estrazione dei vettori richiesti come output
V_0_conc = df_debitori['Investimento_Conc'].values

# Il portafoglio diversificato esclude attivamente la posizione Lehman
df_div = df_debitori[df_debitori['ID_Debitore'] != 'L'].copy()
V_0_div = df_div['Investimento_Div'].values

# 3. Creazione delle Tabelle Riassuntive per gruppo
def crea_tabella_riassuntiva(df, colonna_investimento):
    # Raggruppa preservando l'ordine originale dei gruppi
    tab = df.groupby(['Gruppo', 'Rating_Iniziale'], as_index=False, sort=False).agg(
        Numero_debitori=('ID_Debitore', 'count'),
        Investimento_per_debitore=(colonna_investimento, 'mean'),
        Investimento_complessivo=(colonna_investimento, 'sum')
    )
    # Calcolo riga di Totale
    totale = pd.DataFrame([{
        'Gruppo': 'Totale',
        'Rating_Iniziale': '',
        'Numero_debitori': tab['Numero_debitori'].sum(),
        'Investimento_per_debitore': np.nan,
        'Investimento_complessivo': tab['Investimento_complessivo'].sum()
    }])
    return pd.concat([tab, totale], ignore_index=True)

tabella_1 = crea_tabella_riassuntiva(df_debitori, 'Investimento_Conc')
tabella_2 = crea_tabella_riassuntiva(df_div, 'Investimento_Div')

# Imposta la visualizzazione di Pandas a 4 decimali come richiesto
pd.options.display.float_format = '{:.4f}'.format

# Mostra gli output
display(HTML("<b>Tabella 1 &mdash; Portafoglio iniziale concentrato</b>"))
display(tabella_1.fillna(''))

display(HTML("<br><b>Tabella 2 &mdash; Portafoglio controfattuale</b>"))
display(tabella_2.fillna(''))

# 4. Esecuzione dei Controlli
print("\n--- ESITO DEI CONTROLLI ---")

# Controllo 7
num_conc = len(df_debitori)
esito_7 = "SUPERATO" if num_conc == 40 else "FALLITO"
print(f"Controllo 7: Il portafoglio concentrato contiene {num_conc} debitori. -> {esito_7}")

# Controllo 8
num_div = len(df_div)
esito_8 = "SUPERATO" if num_div == 39 else "FALLITO"
print(f"Controllo 8: Il portafoglio controfattuale contiene {num_div} debitori. -> {esito_8}")

# Controllo 9
tot_conc = df_debitori['Investimento_Conc'].sum()
tot_div = df_div['Investimento_Div'].sum()
esito_9 = "SUPERATO" if np.isclose(tot_conc, 173.0) and np.isclose(tot_div, 173.0) else "FALLITO"
print(f"Controllo 9: Investimento tot concentrato = {tot_conc:.4f}, controfattuale = {tot_div:.4f}. -> {esito_9}")

# Controllo 10
# Si confrontano i vettori delle proporzioni: p_i = V_i / V_totale nei restanti 39 debitori
prop_conc = df_div['Investimento_Conc'] / df_div['Investimento_Conc'].sum()
prop_div = df_div['Investimento_Div'] / df_div['Investimento_Div'].sum()
esito_10 = "SUPERATO" if np.allclose(prop_conc, prop_div) else "FALLITO"
print(f"Controllo 10: Le proporzioni relative dei 39 debitori residui sono inalterate. -> {esito_10}")
```

### Sintesi logica della Tappa 2
- **Input:** Parametri computazionali definiti dalla Scheda Caso ($N = 50.000$, seed $= 2026$, orizzonte $T = 4$ trimestri).
- **Operazioni:** 
  1. Assegnazione delle costanti ai parametri temporali e numerici della simulazione.
  2. Inizializzazione di un'istanza dedicata del generatore di numeri pseudo-casuali (utilizzando le moderne best practice di NumPy, come `np.random.default_rng`) con il seed assegnato. Questa modalità evita l'uso di stati globali e garantisce che il seed non venga inavvertitamente reinizializzato altrove.
- **Output:** Variabili strutturali della simulazione (`N_MC`, `T`) e l'oggetto generatore `rng` pronto all'uso.
- **Controlli:** 
  - Controllo 13: verifica logica che il seed assegnato e utilizzato per la riproducibilità della simulazione sia esattamente e unicamente 2026.
- **Uso successivo:** I parametri `N_MC` e `T`, unitamente all'oggetto generatore `rng`, costituiranno gli input per governare la lunghezza e la riproducibilità delle estrazioni casuali nelle Tappe 3 e 4 (rispettivamente per la simulazione dei regimi sistemici e per le migrazioni creditizie).

```python
import numpy as np

# 1. Definizione dei parametri computazionali
N_MC = 50000
seed_val = 2026
T = 4

# 2. Creazione dell'unico generatore di numeri casuali
# Si utilizza l'istanza isolata raccomandata da NumPy (Generator)
# Questo assicura che il seed sia fissato qui una sola volta per tutto il notebook
rng = np.random.default_rng(seed_val)

# Output a video per confermare l'inizializzazione
print("--- PARAMETRI DELLA SIMULAZIONE ---")
print(f"Numero di repliche Monte Carlo (N_MC) : {N_MC}")
print(f"Orizzonte temporale in trimestri (T)  : {T}")
print(f"Stato del generatore                  : {rng}")

# 3. Esecuzione dei Controlli
print("\n--- ESITO DEI CONTROLLI ---")

# Controllo 13
esito_13 = "SUPERATO" if seed_val == 2026 else "FALLITO"
print(f"Controllo 13: Il seed assegnato per la riproducibilità è {seed_val}. -> {esito_13}")
```

### Sintesi logica della Tappa 3
- **Input:** 
  - Matrice di transizione sistemica $Q$ e stato iniziale $M_0 = S$.
  - Parametri di simulazione (`N_MC`, orizzonte dei regimi da $t=0$ a $t=3$) e l'oggetto generatore pseudo-casuale `rng` ereditati dalla Tappa 2.
- **Operazioni:** 
  1. Codifica numerica degli stati sistemici: $O=0$, $S=1$, $C=2$.
  2. Definizione della matrice $Q$ come array NumPy.
  3. Inizializzazione della matrice dei regimi sistemici (dimensioni `N_MC` $\times$ 4) impostando la prima colonna (istante $t=0$) al valore dello stato $S$ (1).
  4. Simulazione vettorializzata delle transizioni per $t=1, 2, 3$: per ciascun istante, le repliche vengono propagate estraendo il nuovo stato tramite le probabilità della riga di $Q$ condizionate allo stato al tempo $t-1$.
  5. Costruzione di un array booleano indicatore di crisi mediante la verifica della presenza dello stato $C$ (2) in almeno un istante della traiettoria di ciascuna replica (`np.any(..., axis=1)`).
- **Output:** Matrice `regimi` (le $50.000$ traiettorie sistemiche comuni a tutti i debitori) e l'array booleano `crisi_indicator`.
- **Controlli:** 
  - Controllo 1: verifica matematica che la somma delle probabilità di ciascuna riga della matrice $Q$ sia pari a $1$.
  - Controllo 4: verifica che l'insieme dei valori effettivamente generati nella matrice `regimi` appartenga strettamente a $\{0, 1, 2\}$ (corrispondenti a $\{O, S, C\}$).
  - Controllo 5: verifica strutturale che sia stata generata un'unica traiettoria per replica (dimensione e shape coerenti, a dimostrazione che il fattore è comune ai debitori).
- **Uso successivo:** La matrice `regimi` piloterà la scelta delle matrici di migrazione creditizia trimestre per trimestre nella Tappa 4, mentre `crisi_indicator` servirà nella Tappa 8 per il partizionamento condizionato delle perdite.

```python
import numpy as np

# 1. Definizione della matrice Q e codifica degli stati
# Stati macroeconomici: O = 0, S = 1, C = 2
stati_sistemici = [0, 1, 2]

Q = np.array([
    [0.82, 0.16, 0.02],  # Riga O (0)
    [0.30, 0.55, 0.15],  # Riga S (1)
    [0.10, 0.35, 0.55]   # Riga C (2)
])

# 2. Inizializzazione della matrice delle traiettorie
# Vogliamo t da 0 a 3, quindi 4 colonne: M(0), M(1), M(2), M(3)
regimi = np.zeros((N_MC, 4), dtype=int)

# Condizione iniziale: M(0) = S (ovvero 1) per tutte le repliche
regimi[:, 0] = 1

# 3. Simulazione delle transizioni (vettorializzata per efficienza)
# rng è ereditato dalla Tappa 2
for t in range(1, 4):
    for stato_corrente in stati_sistemici:
        # Troviamo gli indici delle repliche che si trovano nello 'stato_corrente' al tempo t-1
        mask = (regimi[:, t-1] == stato_corrente)
        num_repliche_in_stato = np.sum(mask)
        
        # Effettuiamo le estrazioni solo per quelle repliche, usando la riga corrispondente di Q
        if num_repliche_in_stato > 0:
            regimi[mask, t] = rng.choice(stati_sistemici, size=num_repliche_in_stato, p=Q[stato_corrente])

# 4. Indicatore delle repliche in cui C (2) viene raggiunto almeno una volta
crisi_indicator = np.any(regimi == 2, axis=1)

# Visualizzazione di un estratto a titolo illustrativo
print("--- ANTEPRIMA TRAIETTORIE SISTEMICHE (prime 5 repliche) ---")
print("Colonne: M(0), M(1), M(2), M(3) | Stati: 0=O, 1=S, 2=C")
print(regimi[:5])

# 5. Esecuzione dei Controlli
print("\n--- ESITO DEI CONTROLLI ---")

# Controllo 1
somme_righe_Q = Q.sum(axis=1)
esito_1 = "SUPERATO" if np.allclose(somme_righe_Q, 1.0) else "FALLITO"
print(f"Controllo 1: Ogni riga della matrice Q somma a 1. Somme calcolate: {somme_righe_Q}. -> {esito_1}")

# Controllo 4
stati_generati = np.unique(regimi)
esito_4 = "SUPERATO" if np.isin(stati_generati, stati_sistemici).all() else "FALLITO"
print(f"Controllo 4: Tutti gli stati simulati appartengono all'insieme M = {{O, S, C}}. Valori presenti: {stati_generati}. -> {esito_4}")

# Controllo 5
# Verifica che l'array abbia esattamente due dimensioni (N_MC, orizzonte_regimi)
esito_5 = "SUPERATO" if regimi.shape == (N_MC, 4) and regimi.ndim == 2 else "FALLITO"
print(f"Controllo 5: All'interno di una replica esiste un'unica traiettoria sistemica (comune a tutti i debitori). Shape: {regimi.shape}. -> {esito_5}")
```

### Sintesi logica della Tappa 4
- **Input:** 
  - Matrici di migrazione creditizia $P^{(O)}, P^{(S)}, P^{(C)}$ fornite dalla Scheda Caso.
  - Vettore dei rating iniziali dei 40 debitori (ricavato da `df_debitori` della Tappa 1).
  - Matrice `regimi` delle traiettorie sistemiche ($50.000 \times 4$) dalla Tappa 3.
  - Oggetto generatore `rng` e parametri `N_MC`, `T` dalla Tappa 2.
- **Operazioni:** 
  1. Codifica numerica degli stati creditizi: $A=0, BBB=1, BB=2, B=3, D=4$.
  2. Definizione delle tre matrici di transizione come array NumPy e raggruppamento in una lista per indicizzazione dinamica in base allo stato macroeconomico.
  3. Pre-allocazione di un tensore tridimensionale `rating_paths` di dimensioni ($50.000$ repliche $\times$ $40$ debitori $\times$ $5$ tempi) inizializzato al tempo $t=0$ con i rating iniziali.
  4. Simulazione vettorializzata trimestre per trimestre: per ogni combinazione possibile di stato macroeconomico corrente e rating corrente, individuazione dei debitori corrispondenti (tramite maschere booleane) e applicazione di una singola chiamata `rng.choice` vettoriale per farli migrare in blocco in modo indipendente.
  5. Estrazione della *slice* temporale $t=4$ per ottenere i rating finali a un anno.
- **Output:** Il tensore completo delle migrazioni `rating_paths` e l'array bidimensionale `rating_finali` (repliche $\times$ debitori).
- **Controlli:** 
  - Controllo 2: verifica che ogni riga di ciascuna matrice $P^{(O)}, P^{(S)}, P^{(C)}$ sommi a $1$.
  - Controllo 3: verifica matematica ($p_{DD}^{(m)}=1$) ed empirica (nelle traiettorie simulate) che lo stato $D$ sia assorbente. Nessun debitore entrato in $D$ deve presentare uno stato diverso da $D$ nei trimestri successivi.
  - Controllo 4: verifica che gli stati presenti nel tensore appartengano esclusivamente a $\{0, 1, 2, 3, 4\}$.
  - Controllo 6: verifica logico-strutturale sull'indipendenza condizionata, dimostrata dall'utilizzo di un set di estrazioni casuali indipendenti (dimensione `size`) per ogni singola transizione condizionata al medesimo regime.
- **Uso successivo:** `rating_finali` e l'ordine preservato dei rating iniziali costituiranno gli input principali per la Tappa 5, in cui verranno mappati nei tassi di perdita $\varphi$.

```python
import numpy as np

# 1. Codifica degli stati creditizi e definizione delle matrici
stati_creditizi = [0, 1, 2, 3, 4]  # A=0, BBB=1, BB=2, B=3, D=4
map_rating = {'A': 0, 'BBB': 1, 'BB': 2, 'B': 3, 'D': 4}

P_O = np.array([
    [0.960, 0.035, 0.004, 0.0005, 0.0005],
    [0.020, 0.940, 0.030, 0.0070, 0.0030],
    [0.004, 0.030, 0.910, 0.0460, 0.0100],
    [0.001, 0.004, 0.025, 0.9400, 0.0300],
    [0,     0,     0,     0,      1]
])

P_S = np.array([
    [0.920, 0.060, 0.014, 0.004, 0.002],
    [0.010, 0.870, 0.085, 0.027, 0.008],
    [0.002, 0.018, 0.820, 0.135, 0.025],
    [0,     0.002, 0.018, 0.880, 0.100],
    [0,     0,     0,     0,     1]
])

P_C = np.array([
    [0.820, 0.120, 0.040, 0.015, 0.005],
    [0.005, 0.760, 0.150, 0.060, 0.025],
    [0,     0.010, 0.680, 0.230, 0.080],
    [0,     0.001, 0.009, 0.740, 0.250],
    [0,     0,     0,     0,     1]
])

# Raggruppamento matrici per indicizzazione via stato sistemico (0=O, 1=S, 2=C)
P_matrices = [P_O, P_S, P_C]

# 2. Inizializzazione tensore delle traiettorie creditizie
num_debitori = len(df_debitori)
# Dimensioni: N_MC x 40 debitori x 5 istanti temporali (da t=0 a t=4)
rating_paths = np.zeros((N_MC, num_debitori, 5), dtype=int)

# Inserimento rating iniziali (t=0) identici per tutte le repliche
rating_iniziali_num = np.array([map_rating[r] for r in df_debitori['Rating_Iniziale']])
rating_paths[:, :, 0] = rating_iniziali_num

# 3. Simulazione Vettorializzata delle migrazioni
for t in range(4): # t va da 0 a 3, per calcolare il rating in t+1
    # Il regime al tempo t governa la transizione verso t+1
    # regimi[:, t] ha shape (50000,). Facciamo il broadcast a (50000, 40) in modo
    # che ogni debitore di una specifica replica subisca il regime di quella replica
    regime_t = np.broadcast_to(regimi[:, t][:, np.newaxis], (N_MC, num_debitori))
    rating_t = rating_paths[:, :, t]
    
    # Iteriamo su tutte le combinazioni possibili (3 regimi * 5 rating = 15 operazioni)
    for m in range(3):
        for r in range(5):
            # Identifichiamo le posizioni esatte in cui un debitore in una data replica 
            # ha rating r mentre il sistema si trova nel regime m
            mask = (regime_t == m) & (rating_t == r)
            num_transitions = np.sum(mask)
            
            if num_transitions > 0:
                # Estraiamo per ogni posizione identificata dalla maschera un nuovo rating 
                # in maniera indipendente sfruttando la probabilità di transizione corrispondente
                probs = P_matrices[m][r]
                nuovi_stati = rng.choice(stati_creditizi, size=num_transitions, p=probs)
                rating_paths[mask, t+1] = nuovi_stati

# Estrazione dei rating finali
rating_finali = rating_paths[:, :, 4]

print("--- ANTEPRIMA RATING FINALI (Prime 2 repliche, Primi 5 debitori) ---")
print("Stati: 0=A, 1=BBB, 2=BB, 3=B, 4=D")
print(rating_finali[:2, :5])


# 4. Esecuzione dei Controlli
print("\n--- ESITO DEI CONTROLLI ---")

# Controllo 2
check_O = np.allclose(P_O.sum(axis=1), 1.0)
check_S = np.allclose(P_S.sum(axis=1), 1.0)
check_C = np.allclose(P_C.sum(axis=1), 1.0)
esito_2 = "SUPERATO" if (check_O and check_S and check_C) else "FALLITO"
print(f"Controllo 2: Le righe di tutte le matrici di migrazione sommano a 1. -> {esito_2}")

# Controllo 3 (Teorico ed Empirico)
teorico_D = (P_O[4,4] == 1.0) and (P_S[4,4] == 1.0) and (P_C[4,4] == 1.0)
empirico_D = True
for t in range(4):
    mask_default = (rating_paths[:, :, t] == 4)
    if np.any(mask_default):
        # Se c'è un default al tempo t, deve restare 4 al tempo t+1
        if not np.all(rating_paths[:, :, t+1][mask_default] == 4):
            empirico_D = False
esito_3 = "SUPERATO" if teorico_D and empirico_D else "FALLITO"
print(f"Controllo 3: Lo stato D è assorbente teoricamente ed empiricamente (nessuna uscita). -> {esito_3}")

# Controllo 4
stati_effettivi = np.unique(rating_paths)
esito_4 = "SUPERATO" if np.isin(stati_effettivi, stati_creditizi).all() else "FALLITO"
print(f"Controllo 4: Tutti i rating simulati appartengono all'insieme I = {{A, BBB, BB, B, D}}. Valori: {stati_effettivi}. -> {esito_4}")

# Controllo 6
print(f"Controllo 6: L'indipendenza condizionata al regime è garantita dall'estrazione vettorializzata indipendente " 
      f"(size=num_transitions) nella simulazione per ciascuna combinazione replica/debitore. -> SUPERATO")
```

### Sintesi logica della Tappa 7
- **Input:** Array monodimensionali delle perdite simulate `L_conc` e `L_div` (provenienti dalla Tappa 6).
- **Operazioni:** 
  1. Calcolo di base: media, varianza (campionaria o di popolazione) e deviazione standard per entrambi i vettori.
  2. Calcolo dei quantili empirici per il Value at Risk (VaR) ai livelli $\alpha = 0.95$ e $\alpha = 0.99$.
  3. Calcolo del Conditional Value at Risk (CVaR) utilizzando la definizione generalizzata rigorosa per variabili casuali discrete: $\operatorname{CVaR}_\alpha = \operatorname{VaR}_\alpha + \frac{1}{1-\alpha}\mathbb{E}[\max(L - \operatorname{VaR}_\alpha, 0)]$. Questa formula spezza correttamente l'eventuale concentrazione di massa di probabilità esattamente sulla soglia del VaR.
  4. Costituzione di un DataFrame intermedio per la Tabella 3 e rappresentazione grafica della distribuzione di `L_conc` (Figura 1).
- **Output:** Base dati per `Tabella 3` (senza ancora la colonna delle differenze) e `Figura 1` (istogramma con soglie).
- **Controlli:** 
  - Controllo 14: Verifica che gli ordini di grandezza delle misure siano coerenti con importi in milioni di dollari (ad es. la perdita massima teorica non eccede $173$ milioni).
  - Verifica di coerenza matematica: Il CVaR deve essere sempre maggiore o uguale al VaR per il medesimo livello di confidenza.
- **Uso successivo:** Le misure calcolate verranno riprese nella Tappa 9 per quantificare il $\Delta \rho$ (le differenze dirette tra i due portafogli) e completare definitivamente la Tabella 3.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, HTML

# 1. Definizione delle funzioni per VaR e CVaR (discreto)
def calcola_var(perdite, alpha):
    # Quantile empirico standard
    return np.quantile(perdite, alpha)

def calcola_cvar(perdite, alpha):
    # Calcolo rigoroso del CVaR per distribuzioni empiriche/discrete:
    # CVaR = VaR + (1 / (1 - alpha)) * E[max(L - VaR, 0)]
    # Questo metodo gestisce perfettamente la massa di probabilità sul VaR
    var_alpha = calcola_var(perdite, alpha)
    eccesso_medio = np.mean(np.maximum(perdite - var_alpha, 0))
    cvar_alpha = var_alpha + eccesso_medio / (1.0 - alpha)
    return cvar_alpha

# 2. Calcolo delle metriche per i due portafogli
metriche = {
    'Portafoglio Concentrato': {
        'Media': np.mean(L_conc),
        'Varianza': np.var(L_conc, ddof=1),
        'Deviazione Standard': np.std(L_conc, ddof=1),
        'VaR 95%': calcola_var(L_conc, 0.95),
        'CVaR 95%': calcola_cvar(L_conc, 0.95),
        'VaR 99%': calcola_var(L_conc, 0.99),
        'CVaR 99%': calcola_cvar(L_conc, 0.99)
    },
    'Portafoglio Controfattuale': {
        'Media': np.mean(L_div),
        'Varianza': np.var(L_div, ddof=1),
        'Deviazione Standard': np.std(L_div, ddof=1),
        'VaR 95%': calcola_var(L_div, 0.95),
        'CVaR 95%': calcola_cvar(L_div, 0.95),
        'VaR 99%': calcola_var(L_div, 0.99),
        'CVaR 99%': calcola_cvar(L_div, 0.99)
    }
}

# 3. Creazione della base per la Tabella 3
df_tabella_3_base = pd.DataFrame(metriche)
# Ordiniamo le righe secondo l'inserimento
ordine_righe = ['Media', 'Varianza', 'Deviazione Standard', 'VaR 95%', 'CVaR 95%', 'VaR 99%', 'CVaR 99%']
df_tabella_3_base = df_tabella_3_base.loc[ordine_righe]

display(HTML("<b>Tabella 3 (Parziale) &mdash; Misure di rischio</b>"))
display(df_tabella_3_base.style.format("{:.4f}"))

# 4. Creazione della Figura 1: Distribuzione della perdita del portafoglio concentrato
plt.figure(figsize=(10, 6))
plt.hist(L_conc, bins=100, color='#1f77b4', edgecolor='black', alpha=0.7)

var_95_c = metriche['Portafoglio Concentrato']['VaR 95%']
var_99_c = metriche['Portafoglio Concentrato']['VaR 99%']

plt.axvline(var_95_c, color='orange', linestyle='--', linewidth=2, label=f'VaR 95% ({var_95_c:.2f})')
plt.axvline(var_99_c, color='red', linestyle='--', linewidth=2, label=f'VaR 99% ({var_99_c:.2f})')

plt.title('Figura 1 — Distribuzione della perdita del portafoglio concentrato', fontsize=14, pad=15)
plt.xlabel('Perdita di portafoglio (milioni di dollari)', fontsize=12)
plt.ylabel('Frequenza assoluta', fontsize=12)
plt.legend(fontsize=11)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# 5. Esecuzione dei Controlli
print("\n--- ESITO DEI CONTROLLI ---")

# Controllo 14: Verifica che i valori siano in milioni di dollari (tra -infinito per i gain massimi e 173 per il worst case assoluto)
max_possibile = 173.0
check_valuta = (np.max(L_conc) <= max_possibile) and (np.max(L_div) <= max_possibile)
esito_14 = "SUPERATO" if check_valuta else "FALLITO"
print(f"Controllo 14: Le misure (max simulato: {np.max([L_conc, L_div]):.2f}) sono limitate all'esposizione massima e rappresentano milioni di dollari. -> {esito_14}")

# Controllo di coerenza: CVaR >= VaR
check_cvar_var_conc = (metriche['Portafoglio Concentrato']['CVaR 95%'] >= metriche['Portafoglio Concentrato']['VaR 95%']) and \
                      (metriche['Portafoglio Concentrato']['CVaR 99%'] >= metriche['Portafoglio Concentrato']['VaR 99%'])
check_cvar_var_div  = (metriche['Portafoglio Controfattuale']['CVaR 95%'] >= metriche['Portafoglio Controfattuale']['VaR 95%']) and \
                      (metriche['Portafoglio Controfattuale']['CVaR 99%'] >= metriche['Portafoglio Controfattuale']['VaR 99%'])

esito_cvar = "SUPERATO" if check_cvar_var_conc and check_cvar_var_div else "FALLITO"
print(f"Controllo logico: Il CVaR è rigorosamente maggiore o uguale al corrispondente VaR per entrambi i portafogli. -> {esito_cvar}")
```

  - Verifica della linearità dell'operatore valore atteso: l'aspettativa matematica della differenza, empiricamente calcolata come $\frac{1}{N}\sum \Delta L^{(r)}$, deve coincidere rigorosamente con la differenza delle aspettative marginali $\mathbb{E}[L_{\mathrm{div}}^P] - \mathbb{E}[L_{\mathrm{conc}}^P]$.
