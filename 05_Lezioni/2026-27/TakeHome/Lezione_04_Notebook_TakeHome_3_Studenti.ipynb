# Caso Studio: Shortfall di margine condizionato a regimi del costo dell'energia

**Contesto**  
Il caso esamina un'impresa manifatturiera energivora per la quale il prezzo dell'energia rappresenta una rilevante fonte di rischio operativo. Un aumento del prezzo unitario dell'energia riduce il margine operativo effettivo dell'impresa. Se tale margine scende sotto una soglia minima desiderata ($m^\star$), l'impresa registra una carenza economica definita **shortfall di margine**.

**Domanda quantitativa**  
Qual è la distribuzione dello shortfall di margine e come cambia lo shortfall atteso condizionando sui diversi regimi del costo dell'energia?

**Variabile finale e Informazione disponibile**  
- **Variabile finale:** Shortfall di margine, definito come $S = (m^\star - M)^+$, dove $M$ è il margine operativo.  
- **Informazione disponibile:** L'incertezza sul prezzo unitario dell'energia ($P_E$) è modellata tramite una partizione in tre regimi informativi (eventi $A_1$, $A_2$, $A_3$). La sigma-algebra informativa è $\mathcal{G} = \sigma(A_1, A_2, A_3)$. Condizionatamente a ciascun evento $A_g$, il prezzo dell'energia si distribuisce come una normale $P_E \mid A_g \sim \mathcal{N}(\mu_g, \sigma_g^2)$.

---

## 1. Quantità da stimare o calcolare

Il calcolo computazionale richiede di stimare le seguenti quantità non condizionate:

$$
\mathbb{E}[S], \qquad \mathbb{P}(S>0), \qquad \mathbb{P}(S>s^\star), \qquad q_\alpha(S)
$$

E le seguenti quantità condizionate per ciascun regime ($g = 1, 2, 3$):

$$
\mathbb{E}[S \mid A_g]
$$

Deve inoltre essere definito il valore atteso condizionato rispetto all'informazione di regime (variabile casuale):

$$
\mathbb{E}[S \mid \mathcal{G}] = \sum_{g=1}^{3} \mathbb{E}[S \mid A_g] \mathbf{1}_{A_g}
$$

Infine, deve essere verificata esplicitamente la formula del valore atteso totale:

$$
\mathbb{E}[S] = \sum_{g=1}^{3} \mathbb{P}(A_g) \mathbb{E}[S \mid A_g]
$$

## 2. Output richiesti

### Tabelle

1. **Tabella dei parametri**: comprendente dati economico-finanziari, probabilità dei regimi, parametri condizionati e soglie implicite.
2. **Tabella di controllo della simulazione**: con frequenze empiriche dei regimi e media/deviazione standard empirica di $P_E \mid A_g$.
3. **Tabella descrittiva delle grandezze principali**: comprendente almeno $P_E$, $M$ e $S$, con media, deviazione standard e principali statistiche descrittive.
4. **Tabella delle quantità di rischio**: comprendente $\widehat{\mathbb{P}}(S>0)$, $\widehat{\mathbb{P}}(S>s^\star)$, quantili principali e le tre medie condizionate $\widehat{\mathbb{E}}[S \mid A_g]$.
5. **Tabella di verifica del valore atteso totale**: confrontando la media globale dello shortfall con la media pesata delle medie condizionate.

### Grafici

1. Istogramma o densità empirica del prezzo dell'energia $P_E$.
2. Istogramma o densità empirica del margine operativo $M$.
3. Distribuzione empirica dello shortfall $S$, con evidenza della massa in zero quando osservabile e della soglia $s^\star$.
4. Grafico delle medie condizionate $\widehat{\mathbb{E}}[S \mid A_g]$.
5. Confronto tra le distribuzioni condizionate dello shortfall.

## 3. Controlli richiesti

3. Sia rilevata l'eventuale presenza di valori negativi di $P_E$.
4. Siano rispettate le relazioni $C_E=Q_E P_E$, $M=G_0-Q_E P_E$ e $S=(m^\star-M)^+$.
5. Lo shortfall sia sempre non negativo.
6. $\widehat{\mathbb{P}}(S>0)$ sia coerente con la soglia $P_E>120$.
7. $\widehat{\mathbb{P}}(S>s^\star)$ sia coerente con la soglia $P_E>140$.
8. La media globale di $S$ coincida, salvo arrotondamenti, con la media pesata delle medie condizionate.
9. I quantili rispettino l'ordinamento atteso e la media cada tra minimo e massimo campionario.

```python
import matplotlib.pyplot as plt

# CONTROLLO PREVENTIVO: La somma delle probabilità dei regimi deve essere 1
assert np.isclose(sum(prob_A), 1.0), "Errore: le probabilità dei regimi non sommano a 1."

# 1. Simulazione dell'appartenenza ai regimi (A1=1, A2=2, A3=3)
# Utilizziamo np.random.choice con le probabilità assegnate
regimi = np.random.choice([1, 2, 3], size=n_sim, p=prob_A)

# 2. Generazione di mu e sigma array condizionati al regime estratto per la simulazione vettoriale
mu_array = np.choose(regimi - 1, mu_g)
sigma_array = np.choose(regimi - 1, sigma_g)

# 3. Simulazione del prezzo dell'energia P_E condizionato
P_E = np.random.normal(loc=mu_array, scale=sigma_array)

# 4. Calcolo delle statistiche empiriche e costruzione della Tabella 2
# Creiamo un DataFrame temporaneo per calcolare le statistiche raggruppate
df_sim = pd.DataFrame({'Regime': regimi, 'P_E': P_E})
stat_empiriche = df_sim.groupby('Regime')['P_E'].agg(
    Numerosita='count',
    Media_Empirica='mean',
    Std_Empirica='std'
)

# Aggiungiamo i valori teorici per il confronto
stat_empiriche['Prob_Teorica'] = prob_A
stat_empiriche['Freq_Empirica'] = stat_empiriche['Numerosita'] / n_sim
stat_empiriche['Media_Teorica'] = mu_g
stat_empiriche['Std_Teorica'] = sigma_g

# Riordino le colonne per una migliore leggibilità
tabella_2 = stat_empiriche[['Prob_Teorica', 'Freq_Empirica', 'Numerosita', 
                            'Media_Teorica', 'Media_Empirica', 
                            'Std_Teorica', 'Std_Empirica']].copy()
tabella_2.index = ['A1 (Normalizzazione)', 'A2 (Tensione)', 'A3 (Shock)']

# 5. Rilevamento valori negativi di P_E (Controllo esplicito richiesto)
num_prezzi_negativi = np.sum(P_E < 0)

# ================== OUTPUT VISIVI ==================

print("Tabella 2 - Controllo della simulazione (Confronto Teorico vs Empirico)\n")
display(tabella_2.round(4))

print(f"\n--- CONTROLLO PREZZI NEGATIVI ---")
print(f"Valori di P_E negativi rilevati: {num_prezzi_negativi} su {n_sim} simulazioni ({(num_prezzi_negativi/n_sim)*100:.4f}%).")
if num_prezzi_negativi > 0:
    print("Nota: Come specificato dalla Scheda Caso, i valori negativi sono mantenuti e trattati come limite della specificazione normale.")

# Generazione Grafico 1: Istogramma/Densità di P_E
plt.figure(figsize=(10, 6))
plt.hist(P_E, bins=100, density=True, alpha=0.7, color='steelblue', edgecolor='black')
plt.title("Grafico 1 - Distribuzione Empirica del Prezzo dell'Energia ($P_E$)", fontsize=14)
plt.xlabel("Prezzo dell'Energia (euro/MWh)", fontsize=12)
plt.ylabel("Densità", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

**Sintesi logica della Tappa 5:**
- **Input:** Array globale dello shortfall $S$ (generato nella Tappa 3); array dei regimi estratti (generato nella Tappa 2); medie empiriche condizionate $\widehat{\mathbb{E}}[S \mid A_g]$ (calcolate nella Tappa 4).
- **Operazioni:** 
  1. Calcolo diretto della media empirica globale sull'intero array $S$.
  2. Calcolo delle frequenze empiriche dei tre regimi informativi.
  3. Calcolo della somma pesata delle medie condizionate, utilizzando le frequenze empiriche dei regimi come pesi.
  4. Confronto matematico tra le due grandezze per verificare la legge del valore atteso totale.
- **Output:**
  1. **Tabella 5**: Tabella di verifica del valore atteso totale (affiancamento del risultato globale indipendente e del risultato della somma pesata empirica).
- **Controlli:**
  1. Verifica logico-numerica (tramite funzione di uguaglianza approssimata per float) che la media globale empirica coincida, salvo arrotondamenti di macchina, con la media pesata empirica.
- **Uso successivo:** Gli output e le conclusioni verificate in questa tappa chiudono la fase computazionale e costituiranno la base per l'Interpretazione Critica finale (Fase 8).

```python
import numpy as np
import pandas as pd

# ================== OPERAZIONI E CALCOLI ==================

# 1. Calcolo diretto di E[S] globale sull'array S (calcolo indipendente)
E_S_globale = np.mean(S)

# 2. Calcolo delle frequenze empiriche dei regimi
freq_A1 = np.mean(regimi == 1)
freq_A2 = np.mean(regimi == 2)
freq_A3 = np.mean(regimi == 3)

# 3. Calcolo della media pesata delle medie condizionate (Teorema del Valore Atteso Totale)
# Le medie condizionate (media_S_A1, media_S_A2, media_S_A3) provengono dalla Tappa 4
E_S_pesata = (freq_A1 * media_S_A1) + (freq_A2 * media_S_A2) + (freq_A3 * media_S_A3)

# ================== CONTROLLI RICHIESTI ==================

# 4. Controllo di uguaglianza (salvo minimi arrotondamenti di macchina)
check_valore_atteso_totale = np.isclose(E_S_globale, E_S_pesata)

print("--- CONTROLLO TEOREMA DEL VALORE ATTESO TOTALE ---")
print(f"Uguaglianza tra media globale e media pesata empirica rispettata: {check_valore_atteso_totale}")
print(f"Differenza assoluta tra i due metodi di calcolo: {abs(E_S_globale - E_S_pesata):.10f} euro\n")

# ================== OUTPUT TABELLARE ==================

# 5. Costruzione Tabella 5: Verifica del Valore Atteso Totale
dati_tabella_5 = {
    "Metrica di Verifica": [
        "Media globale di S (Calcolo diretto e indipendente)",
        "Somma pesata delle medie condizionate di S (Teorema V.A.T.)",
        "Differenza di calcolo (Arrotondamento)"
    ],
    "Valore (euro)": [
        E_S_globale,
        E_S_pesata,
        abs(E_S_globale - E_S_pesata)
    ]
}

tabella_5 = pd.DataFrame(dati_tabella_5).set_index("Metrica di Verifica")

print("Tabella 5 - Verifica del Valore Atteso Totale\n")
display(tabella_5.round(4))
```
