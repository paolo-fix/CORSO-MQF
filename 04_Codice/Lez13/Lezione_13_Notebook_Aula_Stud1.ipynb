## 1. Titolo, contesto e domanda quantitativa

- **Titolo del caso:** Silicon Valley Bank 2023: piano ALM, sensitività e scenari di stress
- **Contesto:** Il caso fa riferimento alla crisi di Silicon Valley Bank (SVB) del marzo 2023 come quadro stilizzato per analizzare, tramite programmazione lineare deterministica multiperiodale, l'interazione tra composizione dell'attivo iniziale, cash flow generati, fabbisogni di liquidità temporali, capacità di funding esterno e relativo costo del funding.
- **Domanda quantitativa:** Come cambia il piano ALM ottimo di SVB al variare separatamente e congiuntamente del fabbisogno di liquidità ($\theta$), della capacità massima di funding ($\phi$) e del costo del funding ($\psi$), e fino a quale livello di aumento dei fabbisogni iniziali il problema rimane ammissibile?
- **Variabile finale:** Massimizzazione della funzione obiettivo del valore della banca $\Pi(x,g,u;\psi)$ determinata dai rendimenti delle attività, dai proventi delle giacenze di liquidità e dai costi del funding esterno.
- **Informazione disponibile:** 
  - Risorse iniziali disponibili $A_0 = 100$.
  - Vettore redditività attivi $c' = (0.010, 0.025, 0.040, 0.050)$.
  - Vettore perdite unitarie sotto stress $\ell' = (0, 0.02, 0.15, 0.08)$ con limite prudenziale $K = 7$.
  - Limite di concentrazione sull'attivo a lunga scadenza $x_3 \le 30$.
  - Matrice dei cash flow delle attività $A^{\mathrm{CF}}$ ($4 \times 4$).
  - Vettore dei fabbisogni benchmark $d' = (30, 30, 22, 15)$.
  - Vettore rendimenti giacenze ${r^g}' = (0.005, 0.006, 0.007)$.
  - Vettore costi benchmark del funding ${r^u}' = (0.015, 0.020, 0.025)$.
  - Vettore capacità massima benchmark di funding $\bar{u}' = (10, 8, 6)$.
  - Griglie parametrate per le analisi di sensitività univariata ($\theta$, $\phi$, $\psi$) e configurazioni per la scenario analysis.

## Tappa 1: Parametri del Modello, Struttura Matriciale Canonica e Routine Parametrica

In questa tappa si definisce la struttura formale del problema di Programmazione Lineare (LP) per il modello ALM stilizzato di Silicon Valley Bank. 

L'ordinamento esplicito delle 11 variabili decisionale nel vettore $z \in \mathbb{R}^{11}_+$ è fissato come segue:
$$z = (x_1, x_2, x_3, x_4, g_1, g_2, g_3, g_4, u_1, u_2, u_3)'$$
dove $x_j$ rappresentano gli attivi iniziali, $g_t$ le giacenze di liquidità alla fine della data $t$ e $u_t$ il funding esterno raccolto alla data $t$.

Il problema di massimizzazione del valore della banca:
$$\max \Pi(x,g,u;\psi) = c'x + \sum_{t=1}^3 r_t^g g_t - \sum_{t=1}^3 \psi r_t^u u_t$$
viene convertito nella forma standard richiesta dal solver `scipy.optimize.linprog` mediante minimizzazione del negativo: $\min -f'z$, dove $f = -c_{\Pi}$.

| **Capacità Benchmark Funding** | $\bar{u}'$ | $(10.0, 8.0, 6.0)$ | Limite massimo esogeno di funding raccoglibile |
| **Matrice Cash Flow** | $A^{\mathrm{CF}}$ | $\begin{pmatrix} 1.00 & 0.65 & 0.05 & 0.05 \\ 0.00 & 0.35 & 0.10 & 0.15 \\ 0.00 & 0.00 & 0.25 & 0.30 \\ 0.00 & 0.00 & 0.60 & 0.50 \end{pmatrix}$ | Flussi prodotti da una unità investita in ciascun attivo alle date $t=1,2,3,4$ |

```python
import numpy as np
import pandas as pd
from scipy.optimize import linprog

# ==============================================================================
# TAPPA 1: Dati, Struttura Matriciale Canonica e Funzione Parametrica solve_alm
# ==============================================================================

# 1. Definizione dell'ordinamento delle variabili decisionali (vettore z in R^11)
# z = [x1, x2, x3, x4, g1, g2, g3, g4, u1, u2, u3]'
VAR_NAMES = [
    "x1 (Cassa/Riserve)", "x2 (Titoli Breve)", "x3 (Titoli Lunga)", "x4 (Prestiti)",
    "g1 (Giacenza t=1)", "g2 (Giacenza t=2)", "g3 (Giacenza t=3)", "g4 (Giacenza t=4)",
    "u1 (Funding t=1)", "u2 (Funding t=2)", "u3 (Funding t=3)"
]

# 2. Parametri di base del modello (Benchmark)
A0 = 100.0  # Risorse iniziali
c = np.array([0.010, 0.025, 0.040, 0.050])  # Redditività attivi
ell = np.array([0.0, 0.02, 0.15, 0.08])     # Perdite unitarie sotto stress
K = 7.0                                     # Limite prudenziale
x3_max = 30.0                               # Limite di concentrazione su x3

# Matrice dei Cash Flow A_CF (4 x 4)
# Righe: date t=1..4; Colonne: attivi x1..x4
A_CF = np.array([
    [1.00, 0.65, 0.05, 0.05],
    [0.00, 0.35, 0.10, 0.15],
    [0.00, 0.00, 0.25, 0.30],
    [0.00, 0.00, 0.60, 0.50]
])

# Fabbisogni benchmark
d_bench = np.array([30.0, 30.0, 22.0, 15.0])

# Rendimenti giacenze e costi/capacità benchmark del funding
r_g = np.array([0.005, 0.006, 0.007])
r_u_bench = np.array([0.015, 0.020, 0.025])
u_bar_bench = np.array([10.0, 8.0, 6.0])

# 3. Costruzione della funzione parametrica solve_alm(theta, phi, psi)
def solve_alm(theta=1.0, phi=1.0, psi=1.0):
    """
    Risolve il problema di programmazione lineare ALM per dati valori di:
    - theta: moltiplicatore fabbisogni iniziali (t=1, t=2)
    - phi: moltiplicatore capacità massima di funding
    - psi: moltiplicatore costo del funding
    """
    # Vector c_pi per la funzione obiettivo di massimizzazione:
    # Max Pi = c'x + sum(r_g * g) - sum(psi * r_u * u)
    # linprog minimizza f'z, quindi impostiamo f = -c_pi
    f = np.zeros(11)
    f[0:4] = -c
    f[4:7] = -r_g
    f[8:11] = psi * r_u_bench

    # --- VINCOLI DI UGUAGLIANZA: A_eq @ z = b_eq ---
    # 5 equazioni: 1 risorse iniziali + 4 bilanci temporali
    A_eq = np.zeros((5, 11))
    b_eq = np.zeros(5)

    # Equazione 0: Risorse iniziali sum(x_j) = A0
    A_eq[0, 0:4] = 1.0
    b_eq[0] = A0

    # Fabbisogni parametrizzati d(theta)
    d_theta = np.array([30.0 * theta, 30.0 * theta, 22.0, 15.0])

    # Equazione 1 (t=1): a1'x + u1 - g1 = d1(theta)
    A_eq[1, 0:4] = A_CF[0, :]
    A_eq[1, 4] = -1.0  # -g1
    A_eq[1, 8] = 1.0   # +u1
    b_eq[1] = d_theta[0]

    # Equazione 2 (t=2): a2'x + (1+r1^g)g1 + u2 - (1+psi*r1^u)u1 - g2 = d2(theta)
    A_eq[2, 0:4] = A_CF[1, :]
    A_eq[2, 4] = 1.0 + r_g[0]            # +(1+r1^g)g1
    A_eq[2, 5] = -1.0                    # -g2
    A_eq[2, 8] = -(1.0 + psi * r_u_bench[0]) # -(1+psi*r1^u)u1
    A_eq[2, 9] = 1.0                     # +u2
    b_eq[2] = d_theta[1]

    # Equazione 3 (t=3): a3'x + (1+r2^g)g2 + u3 - (1+psi*r2^u)u2 - g3 = d3
    A_eq[3, 0:4] = A_CF[2, :]
    A_eq[3, 5] = 1.0 + r_g[1]            # +(1+r2^g)g2
    A_eq[3, 6] = -1.0                    # -g3
    A_eq[3, 9] = -(1.0 + psi * r_u_bench[1]) # -(1+psi*r2^u)u2
    A_eq[3, 10] = 1.0                    # +u3
    b_eq[3] = d_theta[2]

    # Equazione 4 (t=4): a4'x + (1+r3^g)g3 - (1+psi*r3^u)u3 - g4 = d4
    A_eq[4, 0:4] = A_CF[3, :]
    A_eq[4, 6] = 1.0 + r_g[2]            # +(1+r3^g)g3
    A_eq[4, 7] = -1.0                    # -g4
    A_eq[4, 10] = -(1.0 + psi * r_u_bench[2]) # -(1+psi*r3^u)u3
    b_eq[4] = d_theta[3]

    # --- VINCOLI DI DISUGUAGLIANZA: A_ub @ z <= b_ub ---
    # 2 vincoli: prudenziale e concentrazione x3
    A_ub = np.zeros((2, 11))
    b_ub = np.zeros(2)

    # Vincolo 0: ell'x <= K
    A_ub[0, 0:4] = ell
    b_ub[0] = K

    # Vincolo 1: x3 <= 30
    A_ub[1, 2] = 1.0
    b_ub[1] = x3_max

    # --- BOUNDS PER LE VARIABILI ---
    # x_j >= 0, g_t >= 0, 0 <= u_t <= phi * u_bar_t
    bounds = []
    # x1..x4
    for _ in range(4):
        bounds.append((0.0, None))
    # g1..g4
    for _ in range(4):
        bounds.append((0.0, None))
    # u1..u3
    u_cap = phi * u_bar_bench
    for t in range(3):
        bounds.append((0.0, u_cap[t]))

    # Risoluzione mediante scipy.optimize.linprog (metodo Highs)
    res = linprog(c=f, A_ub=A_ub, b_ub=b_ub, A_eq=A_eq, b_eq=b_eq, bounds=bounds, method='highs')

    return res, f, A_eq, b_eq, A_ub, b_ub, bounds

# Controllo preliminare di consistenza dimensionale e compilazione
res_test, f_test, A_eq_test, b_eq_test, A_ub_test, b_ub_test, bounds_test = solve_alm(1.0, 1.0, 1.0)
print("Inizializzazione completata con successo.")
print(f"Dimensioni A_eq: {A_eq_test.shape} (attese 5x11)")
print(f"Dimensioni A_ub: {A_ub_test.shape} (attese 2x11)")
print(f"Stato test preliminare solver: success={res_test.success}")
```

```python
# ==============================================================================
# TAPPA 3: Controlli Indipendenti, Residuali, Slack e Valori Marginali
# ==============================================================================

TOL = 1e-5

# ------------------------------------------------------------------------------
# 1. TABELLA 3: VERIFICA DEI BILANCI TEMPORALI E RESIDUALI
# ------------------------------------------------------------------------------
# A_eq @ z* - b_eq
residuali_eq = np.abs(A_eq_bench @ z_bench - b_eq_bench)

# Dettaglio uscite/entrate per ciascuna data t=1..4
# t=1: Disponibile = a1'x + u1, Fabbisogno = d1 + g1
disp_t1 = np.dot(A_CF[0, :], x_bench) + u_bench[0]
fabb_t1 = d_bench[0] + g_bench[0]

# t=2: Disponibile = a2'x + (1+r1^g)g1 + u2, Fabbisogno = d2 + (1+r1^u)u1 + g2
disp_t2 = np.dot(A_CF[1, :], x_bench) + (1.0 + r_g[0]) * g_bench[0] + u_bench[1]
fabb_t2 = d_bench[1] + (1.0 + r_u_bench[0]) * u_bench[0] + g_bench[1]

# t=3: Disponibile = a3'x + (1+r2^g)g2 + u3, Fabbisogno = d3 + (1+r2^u)u2 + g3
disp_t3 = np.dot(A_CF[2, :], x_bench) + (1.0 + r_g[1]) * g_bench[1] + u_bench[2]
fabb_t3 = d_bench[2] + (1.0 + r_u_bench[1]) * u_bench[1] + g_bench[2]

# t=4: Disponibile = a4'x + (1+r3^g)g3, Fabbisogno = d4 + (1+r3^u)u3 + g4
disp_t4 = np.dot(A_CF[3, :], x_bench) + (1.0 + r_g[2]) * g_bench[2]
fabb_t4 = d_bench[3] + (1.0 + r_u_bench[2]) * u_bench[2] + g_bench[3]

tabella_3_data = {
    "Data / Equazione": ["t=0 (Risorse)", "t=1 (Bilancio L1)", "t=2 (Bilancio L2)", "t=3 (Bilancio L3)", "t=4 (Bilancio L4)"],
    "Disponibile": [np.sum(x_bench), disp_t1, disp_t2, disp_t3, disp_t4],
    "Impiego / Fabbisogno": [A0, fabb_t1, fabb_t2, fabb_t3, fabb_t4],
    "Giacenza Finale": ["—", g_bench[0], g_bench[1], g_bench[2], g_bench[3]],
    "Residuale Absolute": residuali_eq
}

df_tabella_3 = pd.DataFrame(tabella_3_data)
df_tabella_3["Disponibile"] = df_tabella_3["Disponibile"].round(4)
df_tabella_3["Impiego / Fabbisogno"] = df_tabella_3["Impiego / Fabbisogno"].round(4)
df_tabella_3["Residuale Absolute"] = df_tabella_3["Residuale Absolute"].apply(lambda val: f"{val:.2e}")

print("=== TABELLA 3: VERIFICA BILANCI TEMPORALI BENCHMARK ===")
print(df_tabella_3.to_string(index=False))

# ------------------------------------------------------------------------------
# 2. TABELLA 4: ANALISI DEGLI SLACK E VINCOLI ATTIVI
# ------------------------------------------------------------------------------
# Vincoli A_ub @ z <= b_ub
assunto_ub = A_ub_bench @ z_bench
slack_ub = res_bench.slack[0:2] if hasattr(res_bench, 'slack') and len(res_bench.slack) >= 2 else (b_ub_bench - assunto_ub)

# Bounds funding: u_t <= u_bar_t
assunto_u = u_bench
slack_u = u_bar_bench - assunto_u

nami_vincoli = [
    "Prudenziale (ell'x <= K)",
    "Concentrazione (x3 <= 30)",
    "Capacita Funding u1 (u1 <= 10)",
    "Capacita Funding u2 (u2 <= 8)",
    "Capacita Funding u3 (u3 <= 6)"
]

valori_assunti = list(assunto_ub) + list(assunto_u)
limiti_ub = list(b_ub_bench) + list(u_bar_bench)
slacks_tot = list(slack_ub) + list(slack_u)
stati_vincolo = ["ATTIVO" if s <= TOL else "NON ATTIVO" for s in slacks_tot]

tabella_4_data = {
    "Vincolo / Bound": nami_vincoli,
    "Valore Assunto": valori_assunti,
    "Limite (b_ub)": limiti_ub,
    "Slack": slacks_tot,
    "Stato Vincolo": stati_vincolo
}

df_tabella_4 = pd.DataFrame(tabella_4_data)
df_tabella_4["Valore Assunto"] = df_tabella_4["Valore Assunto"].round(4)
df_tabella_4["Slack"] = df_tabella_4["Slack"].round(4)

print("\n=== TABELLA 4: VINCOLI E SLACK BENCHMARK ===")
print(df_tabella_4.to_string(index=False))

# ------------------------------------------------------------------------------
# 3. TABELLA 5: VALORI MARGINALI E TEST DI PERTURBAZIONE
# ------------------------------------------------------------------------------
# Estrazione prezzi ombra dal solver (res.ineqlin, res.eqlin oppure duali generali)
# linprog (Highs) restituisce res.eqlin.marginals e res.ineqlin.marginals
duali_eq = res_bench.eqlin.marginals if hasattr(res_bench, 'eqlin') else np.zeros(5)
duali_ub = res_bench.ineqlin.marginals if hasattr(res_bench, 'ineqlin') else np.zeros(2)

# Per linprog min -f'z, i marginali indicano d(-Pi)/db = - dPi/db.
# Pertanto, il valore marginale economico e dPi/db = - duale_solver.
val_marg_eq = -duali_eq

# Test di perturbazione per A0 e per il bilancio t=1 (delta_b = +0.1)
# Perturbazione A0
res_pert_A0, _, _, _, _, _, _ = solve_alm(1.0, 1.0, 1.0)
# Per testare esattamente il prezzo ombra di A0, variamo A0 di +0.1 manualmente
A0_pert = A0 + 0.1
# Risoluzione diretta per test A0
f_p = np.zeros(11)
f_p[0:4] = -c; f_p[4:7] = -r_g; f_p[8:11] = r_u_bench
A_eq_p = A_eq_bench.copy(); A_eq_p[0, 0:4] = 1.0
b_eq_p = b_eq_bench.copy(); b_eq_p[0] = A0_pert
res_p_A0 = linprog(c=f_p, A_ub=A_ub_bench, b_ub=b_ub_bench, A_eq=A_eq_p, b_eq=b_eq_p, bounds=bounds_bench, method='highs')
Pi_p_A0 = -res_p_A0.fun
delta_Pi_A0_num = (Pi_p_A0 - Pi_bench) / 0.1

tabella_5_data = {
    "Vincolo / Bilancio": ["Risorse Iniziali (A0)", "Bilancio L1 (t=1)", "Bilancio L2 (t=2)", "Bilancio L3 (t=3)", "Bilancio L4 (t=4)"],
    "Prezzo Ombra Solver": duali_eq,
    "Valore Marginale Economico": val_marg_eq,
    "Test Perturbazione (dPi/db)": [delta_Pi_A0_num, "—", "—", "—", "—"],
    "Esito Perturbazione": ["CONFERMATO" if np.isclose(val_marg_eq[0], delta_Pi_A0_num, atol=1e-4) else "—", "—", "—", "—", "—"]
}

df_tabella_5 = pd.DataFrame(tabella_5_data)
df_tabella_5["Prezzo Ombra Solver"] = df_tabella_5["Prezzo Ombra Solver"].round(6)
df_tabella_5["Valore Marginale Economico"] = df_tabella_5["Valore Marginale Economico"].round(6)
df_tabella_5["Test Perturbazione (dPi/db)"] = df_tabella_5["Test Perturbazione (dPi/db)"].apply(lambda x: f"{x:.6f}" if isinstance(x, float) else x)

print("\n=== TABELLA 5: VALORI MARGINALI BENCHMARK ===")
print(df_tabella_5.to_string(index=False))
```

```python
import matplotlib.pyplot as plt

# ==============================================================================
# TAPPA 4: Cicli Parametrici di Sensitività Univariata e Generazione Grafici (Fig 1-10)
# ==============================================================================

TOL = 1e-5

# Funzione ausiliaria per elaborare una griglia di sensitività univariata completa
def run_sensitivity(param_name, grid, fixed_params):
    results = []
    for val in grid:
        p_dict = fixed_params.copy()
        p_dict[param_name] = val
        res, f, A_eq, b_eq, A_ub, b_ub, bounds = solve_alm(**p_dict)
        
        is_feasible = res.success
        if is_feasible:
            Pi_val = -res.fun
            z_val = res.x
            x_val = z_val[0:4]
            g_val = z_val[4:8]
            u_val = z_val[8:11]
            
            # Capacità massima al denominatore per rho_u
            phi_curr = p_dict.get('phi', 1.0)
            u_cap_curr = phi_curr * u_bar_bench
            rho_u = u_val / u_cap_curr
            
            # Identificazione vincoli e bounds attivi
            slack_ub = b_ub - (A_ub @ z_val)
            slack_u = u_cap_curr - u_val
            attivi = []
            if slack_ub[0] <= TOL: attivi.append("Prudenziale")
            if slack_ub[1] <= TOL: attivi.append("x3<=30")
            for t in range(3):
                if slack_u[t] <= TOL: attivi.append(f"u{t+1}_max")
            str_attivi = ", ".join(attivi) if attivi else "Nessuno"
            
            # Prezzi ombra (marginali)
            duali_eq = res.eqlin.marginals if hasattr(res, 'eqlin') else np.zeros(5)
            duali_ub = res.ineqlin.marginals if hasattr(res, 'ineqlin') else np.zeros(2)
        else:
            Pi_val = np.nan
            z_val = np.full(11, np.nan)
            x_val = np.full(4, np.nan)
            g_val = np.full(4, np.nan)
            u_val = np.full(3, np.nan)
            rho_u = np.full(3, np.nan)
            str_attivi = "NON AMMISSABILE"
            duali_eq = np.full(5, np.nan)
            duali_ub = np.full(2, np.nan)
            
        results.append({
            'param_value': val,
            'success': res.success,
            'message': res.message,
            'Pi': Pi_val,
            'x': x_val,
            'g': g_val,
            'u': u_val,
            'rho_u': rho_u,
            'vincoli_attivi': str_attivi,
            'duali_eq': duali_eq,
            'duali_ub': duali_ub
        })
    return pd.DataFrame(results)

# 1. GRIGLIE DI SENSITIVITÀ ASSEGNATE DALLA SCHEDA CASO
grid_theta = np.round(np.arange(1.00, 1.101, 0.01), 2)
grid_phi = np.round(np.arange(0.50, 1.51, 0.10), 2)
grid_psi = np.array([0.50, 0.75, 1.00, 1.25, 1.50, 1.75, 2.00])

# 2. ESECUZIONE SENSITIVITÀ
df_sens_theta = run_sensitivity('theta', grid_theta, {'phi': 1.0, 'psi': 1.0})
df_sens_phi = run_sensitivity('phi', grid_phi, {'theta': 1.0, 'psi': 1.0})
df_sens_psi = run_sensitivity('psi', grid_psi, {'theta': 1.0, 'phi': 1.0})

# Calcolo delle variazioni rispetto al benchmark e derivata numerica (valore marginale del parametro)
for df_s in [df_sens_theta, df_sens_phi, df_sens_psi]:
    df_s['Delta_Pi'] = df_s['Pi'] - Pi_bench
    df_s['Delta_pct_Pi'] = 100.0 * (df_s['Pi'] / Pi_bench - 1.0)
    # Calcolo valore marginale dPi/dp via differenze finite
    df_s['dPi_dp'] = np.gradient(df_s['Pi'], df_s['param_value'])

# Formattazione completa per stampa Tabelle 6, 7, 8
def format_sens_table(df, param_label):
    out = pd.DataFrame()
    out[param_label] = df['param_value']
    out['Stato'] = df['success'].apply(lambda s: "OK" if s else "NON AMM")
    out['Pi*'] = df['Pi'].round(4)
    out['Delta Pi*'] = df['Delta_Pi'].round(4)
    out['Delta % Pi*'] = df['Delta_pct_Pi'].round(2)
    out['x* (x1,x2,x3,x4)'] = df['x'].apply(lambda v: [round(i, 2) for i in v] if not np.isnan(v[0]) else "N/A")
    out['g* (g1,g2,g3,g4)'] = df['g'].apply(lambda v: [round(i, 2) for i in v] if not np.isnan(v[0]) else "N/A")
    out['u* (u1,u2,u3)'] = df['u'].apply(lambda v: [round(i, 2) for i in v] if not np.isnan(v[0]) else "N/A")
    out['rho_u (t=1,2,3)'] = df['rho_u'].apply(lambda v: [round(i, 2) for i in v] if not np.isnan(v[0]) else "N/A")
    out['Vincoli Attivi'] = df['vincoli_attivi']
    out[f'dPi*/d{param_label}'] = df['dPi_dp'].round(4)
    return out

print("=== TABELLA 6: SENSITIVITÀ A THETA ===")
print(format_sens_table(df_sens_theta, 'theta').to_string(index=False))

print("\n=== TABELLA 7: SENSITIVITÀ A PHI ===")
print(format_sens_table(df_sens_phi, 'phi').to_string(index=False))

print("\n=== TABELLA 8: SENSITIVITÀ A PSI ===")
print(format_sens_table(df_sens_psi, 'psi').to_string(index=False))

# ------------------------------------------------------------------------------
# GENERAZIONE DELLE FIGURE 1 - 10
# ------------------------------------------------------------------------------
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')

# Filtraggio soli punti fattibili per i grafici
df_th_feas = df_sens_theta[df_sens_theta['success']].copy()
df_ph_feas = df_sens_phi[df_sens_phi['success']].copy()
df_ps_feas = df_sens_psi[df_sens_psi['success']].copy()

# Fig 1 & Fig 2: Pi*(theta) e Delta % Pi*(theta)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(df_th_feas['param_value'], df_th_feas['Pi'], 'o-', color='navy', linewidth=2)
axes[0].set_title("Figura 1 — Valore Ottimo Pi*(theta)", fontweight='bold')
axes[0].set_xlabel("theta"); axes[0].set_ylabel("Pi*")

axes[1].plot(df_th_feas['param_value'], df_th_feas['Delta_pct_Pi'], 's--', color='crimson')
axes[1].set_title("Figura 2 — Variazione % Valore Ottimo su theta", fontweight='bold')
axes[1].set_xlabel("theta"); axes[1].set_ylabel("Delta % Pi*")
plt.tight_layout()
plt.show()

# Fig 3 & Fig 4: Composizione x*(theta) e Variazioni Delta x*(theta)
x_mat_th = np.array(df_th_feas['x'].tolist())
x_bench_mat = np.tile(x_bench, (len(df_th_feas), 1))
delta_x_mat = x_mat_th - x_bench_mat

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for j in range(4):
    axes[0].plot(df_th_feas['param_value'], x_mat_th[:, j], 'o-', label=f"x{j+1}")
    axes[1].plot(df_th_feas['param_value'], delta_x_mat[:, j], 's--', label=f"Delta x{j+1}")

axes[0].set_title("Figura 3 — Asset Allocation x*(theta)", fontweight='bold')
axes[0].set_xlabel("theta"); axes[0].legend()
axes[1].set_title("Figura 4 — Variazioni Asset Allocation Delta x*(theta)", fontweight='bold')
axes[1].set_xlabel("theta"); axes[1].legend()
plt.tight_layout()
plt.show()

# Fig 5 & Fig 6: Funding u*(theta) e Giacenze g*(theta)
u_mat_th = np.array(df_th_feas['u'].tolist())
g_mat_th = np.array(df_th_feas['g'].tolist())

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for t in range(3):
    axes[0].plot(df_th_feas['param_value'], u_mat_th[:, t], 'o-', label=f"u{t+1}")
for t in range(4):
    axes[1].plot(df_th_feas['param_value'], g_mat_th[:, t], 's--', label=f"g{t+1}")

axes[0].set_title("Figura 5 — Funding Ottimo u*(theta)", fontweight='bold')
axes[0].set_xlabel("theta"); axes[0].legend()
axes[1].set_title("Figura 6 — Giacenze Ottime g*(theta)", fontweight='bold')
axes[1].set_xlabel("theta"); axes[1].legend()
plt.tight_layout()
plt.show()

# Fig 7 & Fig 8: Sensitività a phi e psi
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(df_ph_feas['param_value'], df_ph_feas['Pi'], 'o-', color='darkgreen')
axes[0].set_title("Figura 7 — Sensitività a Capacità Funding Pi*(phi)", fontweight='bold')
axes[0].set_xlabel("phi"); axes[0].set_ylabel("Pi*")

axes[1].plot(df_ps_feas['param_value'], df_ps_feas['Pi'], 'd-', color='purple')
axes[1].set_title("Figura 8 — Sensitività a Costo Funding Pi*(psi)", fontweight='bold')
axes[1].set_xlabel("psi"); axes[1].set_ylabel("Pi*")
plt.tight_layout()
plt.show()

# Fig 9 & Fig 10: Prezzi Ombra e Mappa Vincoli Attivi su theta
duali_eq_th = -np.array(df_th_feas['duali_eq'].tolist())
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(df_th_feas['param_value'], duali_eq_th[:, 0], 'o-', label="dPi/dA0")
axes[0].plot(df_th_feas['param_value'], duali_eq_th[:, 1], 's--', label="dPi/dd1")
axes[0].plot(df_th_feas['param_value'], duali_eq_th[:, 2], '^:', label="dPi/dd2")
axes[0].set_title("Figura 9 — Valori Marginali lungo theta", fontweight='bold')
axes[0].set_xlabel("theta"); axes[0].legend()

# Mappa vincoli attivi (Utilizzo u_t / u_bar_t)
rho_mat_th = np.array(df_th_feas['rho_u'].tolist())
for t in range(3):
    axes[1].plot(df_th_feas['param_value'], rho_mat_th[:, t], 'o-', label=f"rho_u{t+1}")
axes[1].axhline(1.0, color='red', linestyle='--', label="Limite Attivo")
axes[1].set_title("Figura 10 — Utilizzo Capacità Funding (Mappa Vincoli)", fontweight='bold')
axes[1].set_xlabel("theta"); axes[1].legend()
plt.tight_layout()
plt.show()
```

## Tappa 5: Frontiera di Fattibilità e Determinazione di $\theta_{\mathrm{crit}}$ (Tabella 9)

In questa tappa si esegue la ricerca rigorosa della **frontiera di ammissibilità** del modello ALM stilizzato rispetto allo stress sui fabbisogni di liquidità ($\theta$).

L'analisi si sviluppa in due momenti sequenziali:
1. **Individuazione dell'intervallo critico sulla griglia iniziale:** A partire dalla griglia discreta $\theta \in \{1.00, 1.01, \ldots, 1.10\}$ (con $\phi=1.00$ e $\psi=1.00$), si isolano l'ultimo valore per il quale il problema ammette soluzione ammissibile ($\theta_{\mathrm{amm}}$) e il primo valore per il quale il solver dichiara l'ammissibilità non raggiungibile ($\theta_{\mathrm{non\_amm}}$).
2. **Raffinamento sequenziale su griglia fine:** Nell'intervallo $[\theta_{\mathrm{amm}}, \theta_{\mathrm{non\_amm}}]$ viene generata una griglia ad alta risoluzione con passo $\Delta\theta = 0.0001$ per determinare con elevata precisione numerica l'intervallo ristretto contenente la soglia critica:
   $$\theta_{\mathrm{crit}} \in (\theta_{\mathrm{ultimo\_amm}}, \theta_{\mathrm{primo\_non\_amm}}]$$

In ottemperanza ai vincoli della Scheda Caso, le soluzioni numeriche o i vettori $res.x$ corrispondenti a valori non ammissibili vengono totalmente esclusi dalle tabelle di merito e dalle elaborazioni grafiche.

### Tabella 9 — Raffinamento della Frontiera di Fattibilità

La tabella riporta la scansione ad alta risoluzione nell'intorno del punto di rottura di ammissibilità, evidenziando il preciso cambio di stato del solver (`res.success`).

```python
# ==============================================================================
# TAPPA 5: Frontiera di Fattibilità e Raffinamento per theta_crit (Tabella 9)
# ==============================================================================

# 1. Individuazione del punto di cambio di stato sulla griglia di sensitività iniziale
last_feasible_theta = None
first_infeasible_theta = None

for idx, row in df_sens_theta.iterrows():
    if row['success']:
        last_feasible_theta = row['param_value']
    else:
        first_infeasible_theta = row['param_value']
        break

print("=== SANSIONE INIZIALE FRONTIERA ===")
print(f"Ultimo theta ammissibile sulla griglia iniziale: {last_feasible_theta:.2f}")
print(f"Primo theta NON ammissibile sulla griglia iniziale: {first_infeasible_theta:.2f}")

# 2. Raffinamento ad alta risoluzione nell'intervallo individuato (passo 0.0001)
if last_feasible_theta is not None and first_infeasible_theta is not None:
    fine_grid = np.round(np.arange(last_feasible_theta, first_infeasible_theta + 0.00005, 0.0001), 4)
else:
    # Fallback su intervallo standard se la griglia iniziale fosse interamente ammissibile
    fine_grid = np.round(np.arange(1.00, 1.10, 0.001), 4)

results_fine = []
theta_crit_lower = None
theta_crit_upper = None

for th in fine_grid:
    res, _, _, _, _, _, _ = solve_alm(theta=th, phi=1.0, psi=1.0)
    is_feas = res.success
    
    if is_feas:
        theta_crit_lower = th
    elif theta_crit_upper is None and not is_feas:
        theta_crit_upper = th
        
    results_fine.append({
        'theta': th,
        'd1_d2': 30.0 * th,
        'success': res.success,
        'status_msg': res.message,
        'esito': "AMMISSABILE" if is_feas else "NON AMMISSABILE"
    })

df_fine = pd.DataFrame(results_fine)

# Selezione di un sottoinsieme rappresentativo attorno al punto di rottura per la Tabella 9
idx_break = df_fine[df_fine['success'] == False].index[0] if any(~df_fine['success']) else len(df_fine)-1
start_show = max(0, idx_break - 5)
end_show = min(len(df_fine), idx_break + 5)

df_tabella_9 = df_fine.iloc[start_show:end_show].copy()
df_tabella_9['Note'] = df_tabella_9.apply(
    lambda r: "Frontiera (Ultimo Ammissibile)" if r['theta'] == theta_crit_lower 
    else ("Frontiera (Primo Non Ammissibile)" if r['theta'] == theta_crit_upper else "—"), axis=1
)

print("\n=== TABELLA 9: RAFFINAMENTO DELLA FRONTIERA DI FATTIBILITÀ ===")
print(df_tabella_9.to_string(index=False))

print("\n=== SINTESI DELLA FRONTIERA CRITICA ===")
print(f"Intervallo ristretto contenente theta_crit: ({theta_crit_lower:.4f}, {theta_crit_upper:.4f}]")
print(f"Incremento massimo sostenibile dei fabbisogni iniziali d1, d2: +{(theta_crit_lower - 1.0)*100:.2f}%")
```

## Tappa 6: Scenario Analysis, Confronto tra Risk Driver Congiunti (Tabella 10) e Figure 11–14

In questa tappa finale si valuta il piano ALM di Silicon Valley Bank sotto tre configurazioni deterministiche nelle quali i tre risk driver variano congiuntamente:

1. **Scenario Standard:** $(\theta=1.00, \phi=1.00, \psi=1.00)$ — Coincide con il Benchmark.
2. **Scenario Mediamente Critico:** $(\theta=1.03, \phi=0.90, \psi=1.25)$ — Combina un aumento del $3\%$ dei fabbisogni iniziali, una riduzione del $10\%$ della capacità di funding e un aumento del $25\%$ del costo del funding.
3. **Scenario Critico:** $(\theta=1.06, \phi=0.70, \psi=1.50)$ — Combina un incremento del $6\%$ dei fabbisogni iniziali, un taglio del $30\%$ della capacità di funding e un aumento del $50\%$ del costo del funding.

Per gli scenari *Mediamente Critico* e *Critico* si calcola la **perdita assoluta di valore** rispetto allo Standard:
$$
L_s = \Pi^*_{\mathrm{Standard}} - \Pi^*_s
$$

Gli scenari rappresentano configurazioni assegnate e non eventi con probabilità associate.

```python
# ==============================================================================
# TAPPA 6: Scenario Analysis, Tabella 10 e Figure 11-14
# ==============================================================================

# 1. Definizione delle configurazioni esatte assegnate dalla Sezione 4.13
scenari_config = [
    {"Scenario": "Standard", "theta": 1.00, "phi": 1.00, "psi": 1.00},
    {"Scenario": "Mediamente critico", "theta": 1.03, "phi": 0.90, "psi": 1.25},
    {"Scenario": "Critico", "theta": 1.06, "phi": 0.70, "psi": 1.50}
]

scenari_results = []

for sc in scenari_config:
    res, f, A_eq, b_eq, A_ub, b_ub, bounds = solve_alm(
        theta=sc["theta"], phi=sc["phi"], psi=sc["psi"]
    )
    
    is_feas = res.success
    if is_feas:
        Pi_val = -res.fun
        z_val = res.x
        x_val = z_val[0:4]
        g_val = z_val[4:8]
        u_val = z_val[8:11]
        
        # Denominatore per utilizzo capacità u_bar_s = phi * u_bar_bench
        u_cap_sc = sc["phi"] * u_bar_bench
        rho_u_sc = u_val / u_cap_sc
        
        # Controllo vincoli attivi
        assunto_ub = A_ub @ z_val
        slack_ub = b_ub - assunto_ub
        slack_u = u_cap_sc - u_val
        
        attivi_list = []
        if slack_ub[0] <= 1e-5: attivi_list.append("Prudenziale")
        if slack_ub[1] <= 1e-5: attivi_list.append("x3<=30")
        for t in range(3):
            if slack_u[t] <= 1e-5: attivi_list.append(f"u{t+1}_max")
        str_attivi = ", ".join(attivi_list) if attivi_list else "Nessuno"
    else:
        Pi_val = np.nan
        x_val = np.full(4, np.nan)
        g_val = np.full(4, np.nan)
        u_val = np.full(3, np.nan)
        rho_u_sc = np.full(3, np.nan)
        str_attivi = "NON AMMISSABILE"
        
    scenari_results.append({
        "Scenario": sc["Scenario"],
        "theta": sc["theta"],
        "phi": sc["phi"],
        "psi": sc["psi"],
        "success": is_feas,
        "Pi": Pi_val,
        "x": x_val,
        "g": g_val,
        "u": u_val,
        "rho_u": rho_u_sc,
        "vincoli_attivi": str_attivi
    })

df_scenari = pd.DataFrame(scenari_results)

# Calcolo della perdita di valore L_s rispetto allo Standard
Pi_standard = df_scenari.loc[df_scenari["Scenario"] == "Standard", "Pi"].values[0]
df_scenari["L_s"] = Pi_standard - df_scenari["Pi"]

# 2. Formattazione e Stampa della Tabella 10
tabella_10_print = pd.DataFrame()
tabella_10_print["Scenario"] = df_scenari["Scenario"]
tabella_10_print["theta"] = df_scenari["theta"]
tabella_10_print["phi"] = df_scenari["phi"]
tabella_10_print["psi"] = df_scenari["psi"]
tabella_10_print["Pi*"] = df_scenari["Pi"].round(4)
tabella_10_print["Perdita L_s"] = df_scenari["L_s"].round(4)
tabella_10_print["x* (x1,x2,x3,x4)"] = df_scenari["x"].apply(lambda v: [round(i, 2) for i in v])
tabella_10_print["g* (g1,g2,g3,g4)"] = df_scenari["g"].apply(lambda v: [round(i, 2) for i in v])
tabella_10_print["u* (u1,u2,u3)"] = df_scenari["u"].apply(lambda v: [round(i, 2) for i in v])
tabella_10_print["rho_u (t=1,2,3)"] = df_scenari["rho_u"].apply(lambda v: [round(i, 2) for i in v])
tabella_10_print["Vincoli Attivi"] = df_scenari["vincoli_attivi"]

print("=== TABELLA 10: CONFRONTO DEGLI SCENARI ===")
print(tabella_10_print.to_string(index=False))

# ------------------------------------------------------------------------------
# GENERAZIONE DELLE FIGURE 11 - 14
# ------------------------------------------------------------------------------
names_scen = df_scenari["Scenario"].tolist()
x_pos = np.arange(len(names_scen))

# Figura 11 — Valore Ottimo e Perdita nei Tre Scenari
fig, ax1 = plt.subplots(figsize=(8, 4))
bars = ax1.bar(x_pos, df_scenari["Pi"], color=['navy', 'darkorange', 'crimson'], width=0.4)
ax1.set_xticks(x_pos)
ax1.set_xticklabels(names_scen, fontweight='bold')
ax1.set_ylabel("Valore Ottimo Pi*")
ax1.set_title("Figura 11 — Valore Ottimo nei Tre Scenari", fontweight='bold')

for bar in bars:
    yval = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2.0, yval/2.0, f"{yval:.4f}", ha='center', va='center', color='white', fontweight='bold')

plt.tight_layout()
plt.show()

# Figura 12 — Composizione Ottima dell'Attivo nei Tre Scenari
x_scen_mat = np.array(df_scenari["x"].tolist())
fig, ax = plt.subplots(figsize=(8, 4))
bottom = np.zeros(len(names_scen))

for j in range(4):
    ax.bar(x_pos, x_scen_mat[:, j], bottom=bottom, label=f"x{j+1}", width=0.5)
    bottom += x_scen_mat[:, j]

ax.set_xticks(x_pos)
ax.set_xticklabels(names_scen, fontweight='bold')
ax.set_ylabel("Ammontare Allocato")
ax.set_title("Figura 12 — Composizione Ottima dell'Attivo nei Tre Scenari", fontweight='bold')
ax.legend(loc='upper right')
plt.tight_layout()
plt.show()

# Figura 13 & Figura 14 — Funding e Giacenze nei Tre Scenari
u_scen_mat = np.array(df_scenari["u"].tolist())
g_scen_mat = np.array(df_scenari["g"].tolist())

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Fig 13: Funding u*
for t in range(3):
    axes[0].plot(names_scen, u_scen_mat[:, t], 'o-', label=f"u{t+1}")
axes[0].set_title("Figura 13 — Funding nei Tre Scenari", fontweight='bold')
axes[0].set_ylabel("Ammontare Funding"); axes[0].legend()

# Fig 14: Giacenze g*
for t in range(4):
    axes[1].plot(names_scen, g_scen_mat[:, t], 's--', label=f"g{t+1}")
axes[1].set_title("Figura 14 — Profilo Temporale Giacenze nei Tre Scenari", fontweight='bold')
axes[1].set_ylabel("Ammontare Giacenza"); axes[1].legend()

plt.tight_layout()
plt.show()
```
