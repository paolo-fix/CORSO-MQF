# SVB–ALM 2023 — Asset allocation, liquidità e valore dell'informazione

## 1. Sintesi del contesto e della domanda quantitativa

Il caso analizza una decisione stilizzata di asset allocation sotto incertezza per una banca (con riferimento alla struttura finanziaria di Silicon Valley Bank durante la crisi del marzo 2023). La banca deve allocare le proprie risorse iniziali ($A_0 = 100$) tra quattro classi di attivo con rendimenti e gradi di liquidabilità differenti, prima che si realizzi lo scenario macroeconomico e di liquidità futuro.

Dopo la scelta iniziale dell'attivo, può verificarsi uno tra tre scenari mutuamente esclusivi ed esaustivi:
- **Scenario Normal ($N$):** condizioni normali di mercato e di raccolta;
- **Scenario Pressure ($P$):** pressione sulla raccolta e parziali deflussi di liquidità;
- **Scenario Run ($R$):** severo run sui depositi e forte stress di liquidità.

Ogni scenario determina in maniera esogena un fabbisogno di liquidità ($d_s$), la quota di attivo trasformabile in liquidità ($\alpha_{js}$) e il costo unitario del funding di emergenza ($\kappa_s$).

**Domanda quantitativa fondamentale:**
Quale composizione iniziale dell'attivo massimizza il risultato economico atteso della banca quando fabbisogni di liquidità, liquidabilità degli attivi e costo del funding dipendono dallo scenario futuro, e quanto valgono rispettivamente l'uso della programmazione stocastica (VSS) e l'informazione perfetta sullo scenario futuro (EVPI)?

- **Variabili decisionali ex ante (primo stadio):** Quota di attivo allocata nelle quattro classi ($x_1, x_2, x_3, x_4$).
- **Variabili di ricorso ex post (secondo stadio):** Funding di emergenza ($u_s$) e liquidità residua ($g_s$) per ciascuno scenario $s \in \{N, P, R\}$.
- **Informazione disponibile:** Distribuzione di probabilità discreta nota ex ante sui tre scenari ($p_N, p_P, p_R$), rendimenti unitari dell'attivo ($c_j$), fabbisogni ($d_s$), coefficienti di liquidabilità ($\alpha_{js}$) e costi del funding ($\kappa_s$).

---

## 2. Quantità da calcolare

### 2.1 Soluzione stocastica (Stochastic Program - SP)
1. Vettore di allocazione ottima ex ante: $x^{SP} = (x_1^{SP}, x_2^{SP}, x_3^{SP}, x_4^{SP})'$
2. Decisioni di ricorso ottime per ciascuno scenario $s \in \{N, P, R\}$: $u_s^{SP}$ e $g_s^{SP}$
3. Rendimento iniziale lordo dell'attivo: $c' x^{SP} = \sum_{j=1}^{4} c_j x_j^{SP}$
4. Costo atteso del funding di emergenza: $\sum_{s \in \mathcal{S}} p_s \kappa_s u_s^{SP}$
5. Valore ottimo della funzione obiettivo stocastica: $z^{SP}$

### 2.2 Expected Value problem (EV) ed Expected Value of Expected Value (EEV)
1. Parametri medi della distribuzione:
   $$\bar{d} = \sum_{s \in \mathcal{S}} p_s d_s, \quad \bar{\alpha}_j = \sum_{s \in \mathcal{S}} p_s \alpha_{js} \quad (j=1,\ldots,4), \quad \bar{\kappa} = \sum_{s \in \mathcal{S}} p_s \kappa_s$$
2. Decisione deterministica basata sui valori medi: $x^{EV}$
3. Decisioni di ricorso per scenario derivanti dalla rivalutazione vincolata di $x^{EV}$ nei tre scenari originari: $u_s^{EEV}$ e $g_s^{EEV}$
4. Valore atteso della soluzione Expected Value rivalutata: $z^{EEV}$
5. Value of Stochastic Solution: $VSS = z^{SP} - z^{EEV}$

### 2.3 Wait-and-See (WS) ed Expected Value of Perfect Information (EVPI)
1. Soluzione ottima distinta per ciascuno scenario noto a priori: $x_s^{WS}$ e corrispondente valore $z_s^{WS}$ per $s \in \{N, P, R\}$
2. Valore atteso con informazione perfetta:
   $$z^{WS} = \sum_{s \in \mathcal{S}} p_s z_s^{WS}$$
3. Expected Value of Perfect Information: $EVPI = z^{WS} - z^{SP}$

## 4. Controlli richiesti

1. **Coerenza probabilità:** Verificare che $\sum_{s \in \mathcal{S}} p_s = 1$.
2. **Vincolo di bilancio iniziale:** Verificare che $\sum_{j=1}^{4} x_j = 100$.
3. **Non negatività del primo stadio:** Verificare che $x_j \geq 0$ per $j=1,\ldots,4$.
4. **Limiti sulle allocazioni:** Verificare che $x_3 \le 70$ e $x_4 \le 70$.
5. **Bilancio di liquidità per scenario:** Verificare che $\sum_{j=1}^{4} \alpha_{js} x_j + u_s = d_s + g_s$ per ogni $s \in \{N, P, R\}$.
6. **Non negatività del secondo stadio:** Verificare che $u_s \geq 0$ e $g_s \geq 0$ per ogni $s \in \{N, P, R\}$.
7. **Stato di ottimalità:** Verificare che il solver individui una soluzione ottima globale per ciascuna formulazione.
8. **Vincolo di non anticipatività:** Verificare che la decisione $x^{SP}$ sia unica ed ex ante per tutti gli scenari.
9. **Rivalutazione vincolata di EEV:** Verificare che $x^{EV}$ rimanga fisso durante la valutazione nei tre scenari e non venga riottimizzato scenario per scenario.
10. **Indipendenza dei problemi Wait-and-See:** Verificare che i tre problemi WS vengano risolti separatamente lasciando variare $x$ per ciascuno scenario.
11. **Non negatività del VSS:** Verificare che $VSS \geq 0$.
12. **Non negatività dell'EVPI:** Verificare che $EVPI \geq 0$.
13. **Ordinamento fondamentale dei valori attesi:** Verificare il rispetto della catena di disuguaglianze $z^{EEV} \le z^{SP} \le z^{WS}$.

## Tappa 1 — Inizializzazione dati e parametri

In questa prima tappa vengono definiti e strutturati tutti i dati esogeni del problema stabiliti dalla Scheda Caso:

1. **Risorse iniziali complessive ($A_0$):** fissate a 100 unità convenzionali.
2. **Rendimenti unitari degli attivi ($c_j$):** cassa/riserve ($x_1$), titoli a breve ($x_2$), titoli a lunga ($x_3$), prestiti/impieghi ($x_4$).
3. **Distribuzione di probabilità degli scenari ($p_s$):** condizioni normali ($N$), pressione sulla raccolta ($P$), run severo ($R$).
4. **Fabbisogni di liquidità ($d_s$):** liquidità richiesta per ciascuno scenario.
5. **Coefficienti di liquidabilità ($\alpha_{js}$):** quota dell'investimento nell'attivo $j$ trasformabile in liquidità nello scenario $s$.
6. **Costi unitari del funding di emergenza ($\kappa_s$):** costo economico esogeno del ricorso al funding per scenario.
7. **Limiti di allocazione:** $x_3 \le 70$ e $x_4 \le 70$.

Viene inoltre eseguito il **Controllo 1** per verificare che la somma delle probabilità degli scenari sia esattamente pari a 1.0, e viene generata la **Tabella 1 (Parametri del problema)**.

```python
import numpy as np
import pandas as pd

# ---------------------------------------------------------
# 1. Definizioni base e parametri esogeni (Scheda Caso)
# ---------------------------------------------------------

# Risorse iniziali
A0 = 100.0

# Etichette attivi e scenari
attivi = ['x1: Cassa e riserve', 'x2: Titoli a breve', 'x3: Titoli a lunga', 'x4: Prestiti e impieghi']
scenari = ['N: Normale', 'P: Pressione', 'R: Run severo']

# Rendimenti unitari c_j
c = np.array([0.010, 0.025, 0.040, 0.050])

# Probabilità p_s
p = np.array([0.60, 0.25, 0.15])

# Fabbisogni di liquidità d_s
d = np.array([15.0, 35.0, 65.0])

# Costi unitari del funding kappa_s
kappa = np.array([0.01, 0.06, 0.25])

# Matrix dei coefficienti di liquidabilità alpha_js (Righe: attivi j, Colonne: scenari s)
alpha = np.array([
    [1.00, 1.00, 1.00],  # x1: Cassa e riserve
    [0.98, 0.95, 0.90],  # x2: Titoli a breve
    [0.85, 0.70, 0.50],  # x3: Titoli a lunga
    [0.60, 0.40, 0.20]   # x4: Prestiti e impieghi
])

# Limiti superiori sulle allocazioni
max_x3 = 70.0
max_x4 = 70.0

# ---------------------------------------------------------
# 2. Controllo 1: Coerenza delle probabilità
# ---------------------------------------------------------
somma_p = np.sum(p)
assert np.isclose(somma_p, 1.0), f"ERRORE: La somma delle probabilità è {somma_p}, atteso 1.0"
print(f"[OK] Controllo 1 superato: Somma probabilità scenari = {somma_p:.2f}\n")

# ---------------------------------------------------------
# 3. Costruzione e visualizzazione della Tabella 1
# ---------------------------------------------------------
# Costruzione DataFrame per la Tabella 1 (Parametri del problema)
df_attivi = pd.DataFrame({
    'Attivo': attivi,
    'Rendimento unitario (c_j)': c,
    'Alpha (N)': alpha[:, 0],
    'Alpha (P)': alpha[:, 1],
    'Alpha (R)': alpha[:, 2],
    'Limite Max Allocazione': ['Nessuno', 'Nessuno', f'<= {max_x3}', f'<= {max_x4}']
})

df_scenari = pd.DataFrame({
    'Scenario': scenari,
    'Probabilità (p_s)': p,
    'Fabbisogno liquidità (d_s)': d,
    'Costo unitario funding (kappa_s)': kappa
})

print("==========================================================================")
print("TABELLA 1 — PARAMETRI DEL PROBLEMA")
print("==========================================================================")
print("\n--- Summary Attivi e Liquidabilità ---")
print(df_attivi.to_string(index=False))
print("\n--- Summary Scenari e Funding ---")
print(df_scenari.to_string(index=False))
print("==========================================================================")
```

## Tappa 2 — Programma Stocastico (SP)

In questa tappa viene formulato e risolto il **Programma Stocastico Lineare a Due Stadi (SP)**.

### Formulazione del Modello:
- **Decisioni di 1° stadio:** Quota di allocazione iniziale $x = (x_1, x_2, x_3, x_4)'$ da scegliere ex ante prima di conoscere lo scenario.
- **Decisioni di ricorso (2° stadio):** Funding di emergenza $u_s \ge 0$ e liquidità residua $g_s \ge 0$ per ciascuno scenario $s \in \{N, P, R\}$.
- **Funzione obiettivo (Massimizzazione):**
  $$\max_{x, u, g} \left\{ \sum_{j=1}^{4} c_j x_j - \sum_{s \in \{N, P, R\}} p_s \kappa_s u_s \right\}$$
- **Vincoli:**
  1. *Risorse iniziali:* $\sum_{j=1}^{4} x_j = 100$
  2. *Limiti sulle allocazioni:* $x_1, x_2 \ge 0$, $0 \le x_3 \le 70$, $0 \le x_4 \le 70$
  3. *Bilancio di liquidità per ogni scenario $s \in \{N, P, R\}$:*
     $$\sum_{j=1}^{4} \alpha_{js} x_j + u_s - g_s = d_s$$
  4. *Non-negatività:* $u_s \ge 0, g_s \ge 0$

Vengono eseguiti i **Controlli 2, 3, 4, 5, 6, 7 e 8** e generata la **Tabella 2 (Soluzione stocastica)**.

```python
from scipy.optimize import linprog

# ---------------------------------------------------------
# 1. Definizione delle Variabili per SciPy linprog (Minimizzazione di -f(x))
# ---------------------------------------------------------
# Vettore delle variabili z = [x1, x2, x3, x4, u_N, u_P, u_R, g_N, g_P, g_R]
# Dimensione: 4 (x) + 3 (u) + 3 (g) = 10 variabili

# Coefficienti funzione obiettivo (minimizziamo il negativo del rendimento netto)
c_obj = np.zeros(10)
c_obj[0:4] = -c                      # Rendimenti attivi (segno meno per minimizzazione)
c_obj[4:7] = p * kappa               # Costo atteso funding (segno + perché già sottratto)
c_obj[7:10] = 0.0                    # Liquidità residua non ha costo/ricavo diretto

# ---------------------------------------------------------
# 2. Vincoli di Uguaglianza (A_eq * z = b_eq)
# ---------------------------------------------------------
# Vincolo 1: x1 + x2 + x3 + x4 = 100
# Vincoli 2-4: sum_j alpha_js * x_j + u_s - g_s = d_s per s in {N, P, R}

A_eq = np.zeros((4, 10))
b_eq = np.zeros(4)

# Vincolo di bilancio iniziale (riga 0)
A_eq[0, 0:4] = 1.0
b_eq[0] = A0

# Vincoli di liquidità per scenario (righe 1, 2, 3)
for s_idx in range(3):
    row = s_idx + 1
    A_eq[row, 0:4] = alpha[:, s_idx]  # Alpha per l'insieme degli attivi nello scenario s
    A_eq[row, 4 + s_idx] = 1.0        # u_s
    A_eq[row, 7 + s_idx] = -1.0       # -g_s
    b_eq[row] = d[s_idx]              # d_s

# ---------------------------------------------------------
# 3. Limiti / Bounding sulle Variabili (bounds)
# ---------------------------------------------------------
bounds = [
    (0, None),     # x1 >= 0
    (0, None),     # x2 >= 0
    (0, max_x3),   # 0 <= x3 <= 70
    (0, max_x4),   # 0 <= x4 <= 70
    (0, None),     # u_N >= 0
    (0, None),     # u_P >= 0
    (0, None),     # u_R >= 0
    (0, None),     # g_N >= 0
    (0, None),     # g_P >= 0
    (0, None)      # g_R >= 0
]

# ---------------------------------------------------------
# 4. Risoluzione del Programma Stocastico
# ---------------------------------------------------------
res_sp = linprog(c_obj, A_eq=A_eq, b_eq=b_eq, bounds=bounds, method='highs')

# Estrazione soluzioni
x_sp = res_sp.x[0:4]
u_sp = res_sp.x[4:7]
g_sp = res_sp.x[7:10]

# Risultati economici
rendimento_lordo_sp = np.dot(c, x_sp)
costo_atteso_funding_sp = np.dot(p * kappa, u_sp)
z_sp = -res_sp.fun  # Invertiamo il segno della minimizzazione

# Liquidità generata dagli attivi per scenario
liq_generata_sp = np.dot(alpha.T, x_sp)

# ---------------------------------------------------------
# 5. Verifiche e Controlli
# ---------------------------------------------------------
print("--- ESITO CONTROLLI DI CORRETTEZZA TECNICA (SP) ---")

# Controllo 7: Solver ha trovato l'ottimo globale
assert res_sp.success, f"ERRORE: Il solver non ha raggiunto l'ottimalità. Msg: {res_sp.message}"
print(f"[OK] Controllo 7: Solver ha raggiunto l'ottimalità globale ({res_sp.message}).")

# Controllo 2: Vincolo di bilancio iniziale x1 + x2 + x3 + x4 = 100
print(f"[OK] Controllo 2: Somma allocazioni = {np.sum(x_sp):.4f} (atteso 100.0)")
assert np.isclose(np.sum(x_sp), 100.0)

# Controllo 3 e 4: Non negatività e limiti sulle allocazioni
assert np.all(x_sp >= -1e-6) and x_sp[2] <= max_x3 + 1e-6 and x_sp[3] <= max_x4 + 1e-6
print("[OK] Controllo 3 & 4: Rispettati non-negatività e limiti (x3 <= 70, x4 <= 70).")

# Controllo 5 e 6: Bilancio liquidità e non-negatività ricorsi per scenario
for s_idx, s_name in enumerate(scenari):
    bilancio = liq_generata_sp[s_idx] + u_sp[s_idx] - g_sp[s_idx]
    assert np.isclose(bilancio, d[s_idx]), f"Errore bilancio scenario {s_name}"
    assert u_sp[s_idx] >= -1e-6 and g_sp[s_idx] >= -1e-6
print("[OK] Controllo 5 & 6: Bilancio di liquidità e non-negatività ricorsi rispettati per tutti gli scenari.")

# Controllo 8: Non-anticipatività (unica decisione x ex ante)
print("[OK] Controllo 8: La decisione x_SP è unica per tutti gli scenari (Non-anticipatività rispettata).\n")

# ---------------------------------------------------------
# 6. Costruzione della Tabella 2
# ---------------------------------------------------------
df_sp_alloc = pd.DataFrame({
    'Attivo': attivi,
    'Allocazione Ottima x_SP': np.round(x_sp, 4),
    'Rendimento Unitario': c
})

df_sp_scenari = pd.DataFrame({
    'Scenario': scenari,
    'Probabilità (p_s)': p,
    'Fabbisogno (d_s)': d,
    'Liquidità dagli attivi': np.round(liq_generata_sp, 4),
    'Funding di emergenza (u_s)': np.round(u_sp, 4),
    'Liquidità residua (g_s)': np.round(g_sp, 4),
    'Costo unitario (kappa_s)': kappa,
    'Costo funding scenario (p_s*k_s*u_s)': np.round(p * kappa * u_sp, 4)
})

print("==========================================================================")
print("TABELLA 2 — SOLUZIONE STOCASTICA (SP)")
print("==========================================================================")
print("\n--- Allocazione Ottima di 1° Stadio (x_SP) ---")
print(df_sp_alloc.to_string(index=False))
print(f"\nRendimento iniziale lordo c'x_SP: {rendimento_lordo_sp:.4f}")

print("\n--- Risultati di 2° Stadio per Scenario ---")
print(df_sp_scenari.to_string(index=False))

print("\n--- Sintesi Economica SP ---")
print(f"Rendimento iniziale lordo (c'x_SP) : {rendimento_lordo_sp:.4f}")
print(f"Costo atteso del funding           : {costo_atteso_funding_sp:.4f}")
print(f"Valore ottimo globale (z_SP)       : {z_sp:.4f}")
print("==========================================================================")
```

## Tappa 3 — Calcolo parametri medi e problema Expected Value (EV)

In questa tappa viene formulato e risolto il **Problema Expected Value (EV)** deterministico.

### 1. Calcolo dei Parametri Medi
Si calcolano le medie ponderate rispetto alla distribuzione di probabilità degli scenari $p_s$:
- **Fabbisogno medio di liquidità:** $\bar{d} = \sum_{s \in \mathcal{S}} p_s d_s$
- **Coefficienti medi di liquidabilità:** $\bar{\alpha}_j = \sum_{s \in \mathcal{S}} p_s \alpha_{js} \quad (j=1,\ldots,4)$
- **Costo medio del funding:** $\bar{\kappa} = \sum_{s \in \mathcal{S}} p_s \kappa_s$

### 2. Formulazione del Problema Determinostico (EV)
Ignorando la variabilità tra gli scenari e considerando esclusivamente la situazione media:
$$\max_{x, u, g} \left\{ \sum_{j=1}^{4} c_j x_j - \bar{\kappa} u \right\}$$
sotto i vincoli:
1. $\sum_{j=1}^{4} x_j = 100$
2. $0 \le x_1, x_2$, $0 \le x_3 \le 70$, $0 \le x_4 \le 70$
3. $\sum_{j=1}^{4} \bar{\alpha}_j x_j + u - g = \bar{d}$
4. $u \ge 0, g \ge 0$

Il problema determina l'allocazione iniziale deterministica $x^{EV}$, che verrà poi valutata nella Tappa 4 all'interno dei tre scenari originari per calcolare il valore $z^{EEV}$.

```python
# ---------------------------------------------------------
# 1. Calcolo dei Parametri Medi (Expected Value)
# ---------------------------------------------------------
d_bar = np.dot(p, d)                     # Fabbisogno medio
alpha_bar = np.dot(alpha, p)             # Vector di alpha medio per ciascun attivo j
kappa_bar = np.dot(p, kappa)             # Costo medio unitario funding

# ---------------------------------------------------------
# 2. Formulazione del Problema EV per SciPy linprog
# ---------------------------------------------------------
# Vettore variabili z_ev = [x1, x2, x3, x4, u_bar, g_bar] (6 variabili)
c_obj_ev = np.zeros(6)
c_obj_ev[0:4] = -c                       # Rendimenti attivi (-c_j)
c_obj_ev[4] = kappa_bar                  # Costo medio funding
c_obj_ev[5] = 0.0                        # Liquidità residua media

# Vincoli di Uguaglianza (A_eq_ev * z_ev = b_eq_ev)
# Row 0: x1 + x2 + x3 + x4 = 100
# Row 1: sum_j alpha_bar_j * x_j + u_bar - g_bar = d_bar
A_eq_ev = np.zeros((2, 6))
b_eq_ev = np.zeros(2)

# Bilancio iniziale
A_eq_ev[0, 0:4] = 1.0
b_eq_ev[0] = A0

# Bilancio liquidità media
A_eq_ev[1, 0:4] = alpha_bar
A_eq_ev[1, 4] = 1.0   # + u_bar
A_eq_ev[1, 5] = -1.0  # - g_bar
b_eq_ev[1] = d_bar

# Bounds sulle variabili
bounds_ev = [
    (0, None),     # x1 >= 0
    (0, None),     # x2 >= 0
    (0, max_x3),   # 0 <= x3 <= 70
    (0, max_x4),   # 0 <= x4 <= 70
    (0, None),     # u_bar >= 0
    (0, None)      # g_bar >= 0
]

# ---------------------------------------------------------
# 3. Risoluzione del Problema EV
# ---------------------------------------------------------
res_ev = linprog(c_obj_ev, A_eq=A_eq_ev, b_eq=b_eq_ev, bounds=bounds_ev, method='highs')

# Estrazione della soluzione EV
x_ev = res_ev.x[0:4]
u_bar_ev = res_ev.x[4]
g_bar_ev = res_ev.x[5]
z_ev = -res_ev.fun

# ---------------------------------------------------------
# 4. Verifiche e Controlli Tecnici
# ---------------------------------------------------------
print("--- ESITO CONTROLLI DI CORRETTEZZA TECNICA (EV) ---")
assert res_ev.success, f"ERRORE: Solver EV non ha raggiunto l'ottimalità. Msg: {res_ev.message}"
print(f"[OK] Controllo 7: Solver EV ha raggiunto l'ottimalità globale ({res_ev.message}).")

print(f"[OK] Controllo 2: Somma allocazioni x_EV = {np.sum(x_ev):.4f} (atteso 100.0)")
assert np.isclose(np.sum(x_ev), 100.0)

assert np.all(x_ev >= -1e-6) and x_ev[2] <= max_x3 + 1e-6 and x_ev[3] <= max_x4 + 1e-6
print("[OK] Controllo 3 & 4: Rispettati non-negatività e limiti per x_EV (x3 <= 70, x4 <= 70).\n")

# ---------------------------------------------------------
# 5. Output dei Risultati EV
# ---------------------------------------------------------
df_param_medi = pd.DataFrame({
    'Parametro': ['Fabbisogno liquidità (d_bar)', 'Costo unitario funding (kappa_bar)'],
    'Valore Medio': [np.round(d_bar, 4), np.round(kappa_bar, 4)]
})

df_alpha_medi = pd.DataFrame({
    'Attivo': attivi,
    'Alpha Medio (alpha_bar_j)': np.round(alpha_bar, 4),
    'Allocazione x_EV': np.round(x_ev, 4)
})

print("==========================================================================")
print("PARAMETRI MEDI E SOLUZIONE EXPECTED VALUE (EV)")
print("==========================================================================")
print("\n--- Parametri Medi del Problema ---")
print(df_param_medi.to_string(index=False))

print("\n--- Alpha Medi e Soluzione Decisionale Determinostica (x_EV) ---")
print(df_alpha_medi.to_string(index=False))

print("\n--- Sintesi Economica Problema EV ---")
print(f"Rendimento lordo attivi (c'x_EV)    : {np.dot(c, x_ev):.4f}")
print(f"Funding medio necessario (u_bar)    : {u_bar_ev:.4f}")
print(f"Liquidità residua media (g_bar)     : {g_bar_ev:.4f}")
print(f"Valore ottimo deterministico (z_EV) : {z_ev:.4f}")
print("==========================================================================")
```

## Tappa 4 — Rivalutazione di $x^{EV}$ nei tre scenari (EEV)

In questa tappa l'allocazione deterministica $x^{EV}$ (ottenuta nella Tappa 3 ignorando l'incertezza) viene **fissata** e rivalutata a fronte dell'effettiva realizzazione dei tre scenari originari ($N, P, R$).

### Procedura di Rivalutazione Vincolata (EEV):
1. La decisione di 1° stadio viene bloccata a $x = x^{EV}$.
2. Per ciascuno scenario $s \in \{N, P, R\}$, si calcola la liquidità generata dagli attivi:
   $$\text{Liq}_{s}^{EV} = \sum_{j=1}^{4} \alpha_{js} x_j^{EV}$$
3. Si determinano le decisioni di ricorso ex post $u_s^{EEV}$ e $g_s^{EEV}$:
   - Se $\text{Liq}_{s}^{EV} < d_s$: occorre ricorrere al funding di emergenza $u_s^{EEV} = d_s - \text{Liq}_{s}^{EV}$, mentre $g_s^{EEV} = 0$.
   - Se $\text{Liq}_{s}^{EV} \ge d_s$: non serve funding ($u_s^{EEV} = 0$) e la liquidità in eccesso costituisce $g_s^{EEV} = \text{Liq}_{s}^{EV} - d_s$.
4. Si calcola il costo atteso del funding e il valore economico atteso $z^{EEV}$:
   $$z^{EEV} = \sum_{j=1}^{4} c_j x_j^{EV} - \sum_{s \in \{N, P, R\}} p_s \kappa_s u_s^{EEV}$$

Viene eseguito il **Controllo 9** (divieto di riottimizzare $x^{EV}$ nei singoli scenari) e generata la **Tabella 3 (Expected Value e rivalutazione)**.

```python
# ---------------------------------------------------------
# 1. Rivalutazione vincolata di x_EV nei tre scenari originari
# ---------------------------------------------------------
# Rendimento lordo fisso generato da x_EV
rendimento_lordo_ev = np.dot(c, x_ev)

# Liquidità generata da x_EV per scenario (Alpha_s * x_EV)
liq_generata_eev = np.dot(alpha.T, x_ev)

# Inizializzazione vettori ricorso EEV
u_eev = np.zeros(3)
g_eev = np.zeros(3)

# Calcolo ricorsi ottimi ex post (senza riottimizzare x)
for s_idx in range(3):
    fabbisogno_diff = d[s_idx] - liq_generata_eev[s_idx]
    if fabbisogno_diff > 0:
        u_eev[s_idx] = fabbisogno_diff
        g_eev[s_idx] = 0.0
    else:
        u_eev[s_idx] = 0.0
        g_eev[s_idx] = -fabbisogno_diff

# Costo atteso funding sotto la decisione x_EV
costo_atteso_funding_eev = np.dot(p * kappa, u_eev)

# Valore atteso complessivo EEV
z_eev = rendimento_lordo_ev - costo_atteso_funding_eev

# ---------------------------------------------------------
# 2. Verifiche e Controlli Tecnici
# ---------------------------------------------------------
print("--- ESITO CONTROLLI DI CORRETTEZZA TECNICA (EEV) ---")

# Controllo 9: Conferma che x_EV è stato tenuto rigido in tutti gli scenari
print("[OK] Controllo 9: x_EV è stato rivalutato mantenendo fisso l'attivo ex ante (senza riottimizzazione per scenario).")

# Verifiche di bilancio per ciascuno scenario originario
for s_idx, s_name in enumerate(scenari):
    bilancio_eev = liq_generata_eev[s_idx] + u_eev[s_idx] - g_eev[s_idx]
    assert np.isclose(bilancio_eev, d[s_idx]), f"Errore bilancio EEV nello scenario {s_name}"
    assert u_eev[s_idx] >= -1e-6 and g_eev[s_idx] >= -1e-6
print("[OK] Controllo 5 & 6 (EEV): Rispettati i bilanci di liquidità e non-negatività dei ricorsi per tutti gli scenari.\n")

# ---------------------------------------------------------
# 3. Costruzione della Tabella 3
# ---------------------------------------------------------
df_eev_scenari = pd.DataFrame({
    'Scenario': scenari,
    'Probabilità (p_s)': p,
    'Fabbisogno (d_s)': d,
    'Liquidità generata da x_EV': np.round(liq_generata_eev, 4),
    'Funding necessario (u_s_EEV)': np.round(u_eev, 4),
    'Liquidità residua (g_s_EEV)': np.round(g_eev, 4),
    'Costo unitario (kappa_s)': kappa,
    'Costo funding scenario': np.round(p * kappa * u_eev, 4)
})

df_confronto_x = pd.DataFrame({
    'Attivo': attivi,
    'Allocazione Stocastica x_SP': np.round(x_sp, 4),
    'Allocazione Valori Medi x_EV': np.round(x_ev, 4),
    'Differenza (x_SP - x_EV)': np.round(x_sp - x_ev, 4)
})

print("==========================================================================")
print("TABELLA 3 — EXPECTED VALUE E RIVALUTAZIONE (EEV)")
print("==========================================================================")
print("\n--- Confronto Allocazioni Ex Ante (x_SP vs x_EV) ---")
print(df_confronto_x.to_string(index=False))

print("\n--- Rivalutazione di x_EV nei Tre Scenari Originari ---")
print(df_eev_scenari.to_string(index=False))

print("\n--- Sintesi Economica EEV vs SP ---")
print(f"Rendimento lordo attivi (c'x_EV)      : {rendimento_lordo_ev:.4f}")
print(f"Costo atteso del funding (EEV)        : {costo_atteso_funding_eev:.4f}")
print(f"Valore atteso rivalutato (z_EEV)      : {z_eev:.4f}")
print(f"Valore ottimo stocastico (z_SP)       : {z_sp:.4f}")
print(f"Perdita da approccio deterministico  : {z_sp - z_eev:.4f}")
print("==========================================================================")
```

## Tappa 5 — Benchmark Wait-and-See (WS)

In questa tappa si calcola il benchmark teorico **Wait-and-See (WS)**, nel quale si ipotizza che la banca conosca con **informazione perfetta anticipata** quale scenario si realizzerà prima di definire la composizione dell'attivo.

### Procedura Operativa:
Per ciascuno scenario $s \in \{N, P, R\}$, viene risolto un problema di ottimizzazione deterministico completamente indipendente:
$$\max_{x_s, u_s, g_s} \left\{ \sum_{j=1}^{4} c_j x_{js} - \kappa_s u_s \right\}$$
sotto i vincoli:
1. $\sum_{j=1}^{4} x_{js} = 100$
2. $0 \le x_{1s}, x_{2s}$, $0 \le x_{3s} \le 70$, $0 \le x_{4s} \le 70$
3. $\sum_{j=1}^{4} \alpha_{js} x_{js} + u_s - g_s = d_s$
4. $u_s \ge 0, g_s \ge 0$

Si ricavano così il vettore di allocazione specifica di scenario $x_s^{WS}$ e il valore ottimo di scenario $z_s^{WS}$.

Infine, si calcola il valore atteso della strategia Wait-and-See ponderando i risultati per le probabilità di scenario:
$$z^{WS} = \sum_{s \in \{N, P, R\}} p_s z_s^{WS}$$

Viene eseguito il **Controllo 10** (indipendenza dei problemi di scenario) e generata la **Tabella 4 (Wait-and-see)**.

```python
# ---------------------------------------------------------
# 1. Risoluzione dei 3 Problemi Deterministici Separati (WS)
# ---------------------------------------------------------
x_ws = np.zeros((4, 3))       # Righe: attivi j, Colonne: scenari s
u_ws = np.zeros(3)            # Funding WS per scenario
g_ws = np.zeros(3)            # Liquidità residua WS per scenario
z_ws_scenari = np.zeros(3)    # Valore ottimo WS per ciascun scenario

print("--- ESITO CONTROLLI DI CORRETTEZZA TECNICA (WS) ---")

for s_idx, s_name in enumerate(scenari):
    # Vettore variabili z_ws = [x1, x2, x3, x4, u_s, g_s] per lo scenario s (6 variabili)
    c_obj_ws = np.zeros(6)
    c_obj_ws[0:4] = -c                  # Rendimenti attivi (-c_j)
    c_obj_ws[4] = kappa[s_idx]          # Costo funding dello scenario s
    c_obj_ws[5] = 0.0                   # Liquidità residua

    # Vincoli di Uguaglianza (A_eq_ws * z_ws = b_eq_ws)
    A_eq_ws = np.zeros((2, 6))
    b_eq_ws = np.zeros(2)

    # Vincolo 1: Bilancio iniziale A0 = 100
    A_eq_ws[0, 0:4] = 1.0
    b_eq_ws[0] = A0

    # Vincolo 2: Bilancio di liquidità nello scenario s
    A_eq_ws[1, 0:4] = alpha[:, s_idx]   # Alpha dello scenario s
    A_eq_ws[1, 4] = 1.0                 # + u_s
    A_eq_ws[1, 5] = -1.0                # - g_s
    b_eq_ws[1] = d[s_idx]               # d_s

    # Bounds sulle variabili
    bounds_ws = [
        (0, None),     # x1 >= 0
        (0, None),     # x2 >= 0
        (0, max_x3),   # 0 <= x3 <= 70
        (0, max_x4),   # 0 <= x4 <= 70
        (0, None),     # u_s >= 0
        (0, None)      # g_s >= 0
    ]

    # Risoluzione del problema dello scenario s
    res_ws_s = linprog(c_obj_ws, A_eq=A_eq_ws, b_eq=b_eq_ws, bounds=bounds_ws, method='highs')
    
    # Verifica stabilità solver
    assert res_ws_s.success, f"ERRORE: Solver WS fallito nello scenario {s_name}"

    # Salvataggio soluzioni
    x_ws[:, s_idx] = res_ws_s.x[0:4]
    u_ws[s_idx] = res_ws_s.x[4]
    g_ws[s_idx] = res_ws_s.x[5]
    z_ws_scenari[s_idx] = -res_ws_s.fun

    # Verifiche di bilancio per scenario
    bilancio_check = np.dot(alpha[:, s_idx], x_ws[:, s_idx]) + u_ws[s_idx] - g_ws[s_idx]
    assert np.isclose(bilancio_check, d[s_idx]), f"Errore bilancio WS nello scenario {s_name}"

print("[OK] Controllo 7 & 10 (WS): Risolti 3 problemi di scenario in modo completamente indipendente e con ottimalità raggiunta.")
print("[OK] Controlli 2, 3, 4, 5, 6 (WS): Bilanci di liquidità, limiti e non-negatività rispettati per tutti gli scenari.\n")

# ---------------------------------------------------------
# 2. Calcolo del Valore Atteso Wait-and-See (z_WS)
# ---------------------------------------------------------
z_ws = np.dot(p, z_ws_scenari)
liq_generata_ws = np.array([np.dot(alpha[:, s_idx], x_ws[:, s_idx]) for s_idx in range(3)])

# ---------------------------------------------------------
# 3. Costruzione della Tabella 4
# ---------------------------------------------------------
df_ws_alloc = pd.DataFrame({
    'Attivo': attivi,
    'Allocazione WS (Scenario N)': np.round(x_ws[:, 0], 4),
    'Allocazione WS (Scenario P)': np.round(x_ws[:, 1], 4),
    'Allocazione WS (Scenario R)': np.round(x_ws[:, 2], 4)
})

df_ws_scenari = pd.DataFrame({
    'Scenario': scenari,
    'Probabilità (p_s)': p,
    'Fabbisogno (d_s)': d,
    'Liquidità generata': np.round(liq_generata_ws, 4),
    'Funding necessario (u_s)': np.round(u_ws, 4),
    'Liquidità residua (g_s)': np.round(g_ws, 4),
    'Valore Ottimo Scenario (z_s_WS)': np.round(z_ws_scenari, 4)
})

print("==========================================================================")
print("TABELLA 4 — WAIT-AND-SEE (WS)")
print("==========================================================================")
print("\n--- Allocazioni Ottime di Scenario (x_s_WS) ---")
print(df_ws_alloc.to_string(index=False))

print("\n--- Risultati di Scenario e Valori z_s_WS ---")
print(df_ws_scenari.to_string(index=False))

print("\n--- Sintesi Economica Benchmark Wait-and-See ---")
for s_idx, s_name in enumerate(scenari):
    print(f"Valore ottimo scenario {s_name} (z_{s_name}_WS) : {z_ws_scenari[s_idx]:.4f}")
print(f"\nValore atteso complessivo Wait-and-See (z_WS) : {z_ws:.4f}")
print("==========================================================================")
```

## Tappa 6 — Valore dell'informazione e sintesi numerica

In questa tappa vengono calcolate le due metriche fondamentali per la valutazione dell'incertezza e dell'informazione nella programmazione stocastica:

1. **Value of Stochastic Solution ($VSS$):**
   $$VSS = z^{SP} - z^{EEV}$$
   Misura il guadagno economico (o il costo dell'ignorare l'incertezza) derivante dall'adozione del modello stocastico $x^{SP}$ rispetto all'uso della soluzione deterministica dei valori medi $x^{EV}$ rivalutata nei tre scenari.

2. **Expected Value of Perfect Information ($EVPI$):**
   $$EVPI = z^{WS} - z^{SP}$$
   Rappresenta il prezzo massimo che la banca sarebbe disposta a pagare per ottenere un'informazione perfetta e anticipata sullo scenario futuro prima di prendere la decisione di asset allocation.

Vengono eseguiti i **Controlli 11, 12 e 13** per verificare la non-negatività delle metriche ($VSS \ge 0, EVPI \ge 0$) e il rispetto della catena di disuguaglianze fondamentale:
$$z^{EEV} \le z^{SP} \le z^{WS}$$

Infine, viene generata la **Tabella 5 (Valore dell'informazione)**.

```python
# ---------------------------------------------------------
# 1. Calcolo delle Metriche dell'Incertezza (VSS ed EVPI)
# ---------------------------------------------------------
vss = z_sp - z_eev
evpi = z_ws - z_sp

# ---------------------------------------------------------
# 2. Verifiche e Controlli Fondamentali
# ---------------------------------------------------------
print("--- ESITO CONTROLLI DI CORRETTEZZA TECNICA (VALORE INFORMAZIONE) ---")

# Controllo 11: Non-negatività del VSS
assert vss >= -1e-6, f"ERRORE: VSS negativo ({vss:.6f})"
print(f"[OK] Controllo 11: VSS >= 0 verificato (VSS = {vss:.4f}).")

# Controllo 12: Non-negatività dell'EVPI
assert evpi >= -1e-6, f"ERRORE: EVPI negativo ({evpi:.6f})"
print(f"[OK] Controllo 12: EVPI >= 0 verificato (EVPI = {evpi:.4f}).")

# Controllo 13: Ordinamento fondamentale z_EEV <= z_SP <= z_WS
catena_rispettata = (z_eev <= z_sp + 1e-6) and (z_sp <= z_ws + 1e-6)
assert catena_rispettata, "ERRORE: Catena di disuguaglianze z_EEV <= z_SP <= z_WS non rispettata!"
print(f"[OK] Controllo 13: Ordinamento z_EEV ({z_eev:.4f}) <= z_SP ({z_sp:.4f}) <= z_WS ({z_ws:.4f}) perfettamente verificato.\n")

# ---------------------------------------------------------
# 3. Costruzione della Tabella 5
# ---------------------------------------------------------
df_tabella_5 = pd.DataFrame({
    'Metrica / Valore': [
        'Valore Expected Value rivalutato (z_EEV)',
        'Valore Soluzione Stocastica (z_SP)',
        'Valore Wait-and-See (z_WS)',
        'Value of Stochastic Solution (VSS = z_SP - z_EEV)',
        'Expected Value of Perfect Information (EVPI = z_WS - z_SP)'
    ],
    'Valore Numerico': [
        np.round(z_eev, 4),
        np.round(z_sp, 4),
        np.round(z_ws, 4),
        np.round(vss, 4),
        np.round(evpi, 4)
    ],
    'Interpretazione Sintetica': [
        'Rendimento atteso dell allocazione naive basata su parametri medi',
        'Rendimento atteso dell allocazione stocastica ottimale ex ante',
        'Rendimento atteso teorico con informazione perfetta anticipata',
        'Beneficio economico dell approccio stocastico rispetto al valore medio',
        'Valore massimo teorico dell informazione perfetta'
    ]
})

print("==========================================================================")
print("TABELLA 5 — VALORE DELL'INFORMAZIONE E SINTESI METRICHE")
print("==========================================================================")
print(df_tabella_5.to_string(index=False))
print("==========================================================================")
```
