# SVB–ALM 2023 — Asset allocation, liquidità e valore dell'informazione

## 1. Sintesi del contesto e della domanda quantitativa

Il caso analizza una decisione stilizzata di asset allocation sotto incertezza per una banca (con riferimento alla struttura finanziaria di Silicon Valley Bank durante la crisi del marzo 2023). La banca deve allocare le proprie risorse iniziali ($A_0 = 100$) tra quattro classi di attivo con rendimenti e gradi di liquidabilità differenti, prima che si realizzi lo scenario macroeconomico e di liquidità futuro.

Dopo la scelta iniziale dell'attivo, può verificarsi uno tra tre scenari mutuamente esclusivi ed esaustivi:
- **Scenario Normal ($N$):** condizioni normali di mercato e di raccolta;
- **Scenario Pressure ($P$):** pressione sulla raccolta e parziali deflussi di liquidità;
- **Scenario Run ($R$):** severo run sui depositi e forte stress di liquidità.

Ogni scenario determina in maniera esogena un fabbisogno di liquidità ($d_s$), la quota di attivo trasformabile in liquidità ($\alpha_{js}$) e il costo unitario del funding di emergenza ($\kappa_s$).

**Domanda quantitativa fondamentale:**
Quale composizione iniziale dell'attivo massimizza il risultato economico atteso della banca quando fabbisogni di liquidità, liquidabilità degli attivi e costo del funding dipendono dallo scenario futuro, e quanto valgono rispettivamente l'uso della programmazione stocastica (VSS) e l'informazione perfetta sullo scenario futuro (EVPI)?

- **Variabili decisionali ex ante (primo stadio):** Quota di attivo allocata nelle quattro classi ($x_1, x_2, x_3, x_4$).
- **Variabili di ricorso ex post (secondo stadio):** Funding di emergenza ($u_s$) e liquidità residua ($g_s$) per ciascuno scenario $s \in \{N, P, R\}$.
- **Informazione disponibile:** Distribuzione di probabilità discreta nota ex ante sui tre scenari ($p_N, p_P, p_R$), rendimenti unitari dell'attivo ($c_j$), fabbisogni ($d_s$), coefficienti di liquidabilità ($\alpha_{js}$) e costi del funding ($\kappa_s$).

## 2. Quantità da calcolare

### 2.1 Soluzione stocastica (Stochastic Program - SP)
1. Vettore di allocazione ottima ex ante: $x^{SP} = (x_1^{SP}, x_2^{SP}, x_3^{SP}, x_4^{SP})'$
2. Decisioni di ricorso ottime per ciascuno scenario $s \in \{N, P, R\}$: $u_s^{SP}$ e $g_s^{SP}$
3. Rendimento iniziale lordo dell'attivo: $c' x^{SP} = \sum_{j=1}^{4} c_j x_j^{SP}$
4. Costo atteso del funding di emergenza: $\sum_{s \in \mathcal{S}} p_s \kappa_s u_s^{SP}$
5. Valore ottimo della funzione obiettivo stocastica: $z^{SP}$

### 2.2 Expected Value problem (EV) ed Expected Value of Expected Value (EEV)
1. Parametri medi della distribuzione:
   $$\bar{d} = \sum_{s \in \mathcal{S}} p_s d_s, \quad \bar{\alpha}_j = \sum_{s \in \mathcal{S}} p_s \alpha_{js} \quad (j=1,\ldots,4), \quad \bar{\kappa} = \sum_{s \in \mathcal{S}} p_s \kappa_s$$
2. Decisione deterministica basata sui valori medi: $x^{EV}$
3. Decisioni di ricorso per scenario derivanti dalla rivalutazione vincolata di $x^{EV}$ nei tre scenari originari: $u_s^{EEV}$ e $g_s^{EEV}$
4. Valore atteso della soluzione Expected Value rivalutata: $z^{EEV}$
5. Value of Stochastic Solution: $VSS = z^{SP} - z^{EEV}$

### 2.3 Wait-and-See (WS) ed Expected Value of Perfect Information (EVPI)
1. Soluzione ottima distinta per ciascuno scenario noto a priori: $x_s^{WS}$ e corrispondente valore $z_s^{WS}$ per $s \in \{N, P, R\}$
2. Valore atteso con informazione perfetta:
   $$z^{WS} = \sum_{s \in \mathcal{S}} p_s z_s^{WS}$$
3. Expected Value of Perfect Information: $EVPI = z^{WS} - z^{SP}$

## 4. Controlli richiesti

1. **Coerenza probabilità:** Verificare che $\sum_{s \in \mathcal{S}} p_s = 1$.
2. **Vincolo di bilancio iniziale:** Verificare che $\sum_{j=1}^{4} x_j = 100$.
3. **Non negatività del primo stadio:** Verificare che $x_j \geq 0$ per $j=1,\ldots,4$.
4. **Limiti sulle allocazioni:** Verificare che $x_3 \le 70$ e $x_4 \le 70$.
5. **Bilancio di liquidità per scenario:** Verificare che $\sum_{j=1}^{4} \alpha_{js} x_j + u_s = d_s + g_s$ per ogni $s \in \{N, P, R\}$.
6. **Non negatività del secondo stadio:** Verificare che $u_s \geq 0$ e $g_s \geq 0$ per ogni $s \in \{N, P, R\}$.
7. **Stato di ottimalità:** Verificare che il solver individui una soluzione ottima globale per ciascuna formulazione.
8. **Vincolo di non anticipatività:** Verificare che la decisione $x^{SP}$ sia unica ed ex ante per tutti gli scenari.
9. **Rivalutazione vincolata di EEV:** Verificare che $x^{EV}$ rimanga fisso durante la valutazione nei tre scenari e non venga riottimizzato scenario per scenario.
10. **Indipendenza dei problemi Wait-and-See:** Verificare che i tre problemi WS vengano risolti separatamente lasciando variare $x$ per ciascuno scenario.
11. **Non negatività del VSS:** Verificare che $VSS \geq 0$.
12. **Non negatività dell'EVPI:** Verificare che $EVPI \geq 0$.
13. **Ordinamento fondamentale dei valori attesi:** Verificare il rispetto della catena di disuguaglianze $z^{EEV} \le z^{SP} \le z^{WS}$.
