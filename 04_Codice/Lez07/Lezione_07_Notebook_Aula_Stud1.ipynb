# Tappa 1 — Definizione dei Parametri, Griglia Temporale e Generazione degli Shock Correlati

## 1. Copia fedele della richiesta dello studente

> Mi sembra utile che l'output sia caratterizzato dalla definizione dei parametri del caso, dalla costruzione della griglia temporale e dalla generazione degli shock casuali correlati per inflazione e tasso. Vorrei anche vedere un controllo che la correlazione ottenuta sia abbastanza vicina a quella assegnata.

---

## 2. Sintesi logica degli output previsti
Nella presente tappa vengono definiti e strutturati gli elementi fondamentali per l'intera simulazione Monte Carlo:
1. **Dizionario dei Parametri del Caso:** Inizializzazione di tutti i parametri contrattuali, di processo stocastico e di simulazione assegnati dalla Scheda Caso.
2. **Griglia Temporale:** Costruzione del vettore delle scadenze $t_k \in [0, T]$ con passo mensile $\Delta t = 1/12$ ($K = 60$ sottointervalli) e identificazione delle scadenze di pagamento delle cedole semestrali.
3. **Generazione degli Shock Correlati:** Fissazione del seed casuale (`12345`) per garantire la riproducibilità numerica. Generazione di $M = 50.000$ traiettorie di shock normali standard indipendenti $(Z^{(1)}, Z^{(2)})$ e successiva trasformazione mediante decomposizione di Cholesky per ottenere gli shock correlati $(Z^{(i)}, Z^{(r)})$ caratterizzati da coefficiente di correlazione $\rho = 0{,}25$:

$$Z^{(i)} = Z^{(1)}$$
$$Z^{(r)} = \rho Z^{(1)} + \sqrt{1-\rho^2} Z^{(2)}$$

4. **Controllo di Robustezza (Controllo 1):** Calcolo della correlazione empirica tra gli shock generati e verifica automatica tramite istruzione `assert` rispetto alla tolleranza fissata ($10^{-2}$).

```python
# ==============================================================================
# TAPPA 1: Parametri, Griglia Temporale e Shock Correlati (Cholesky)
# ==============================================================================

import numpy as np

# ------------------------------------------------------------------------------
# 1. Definizione dei Parametri del Caso
# ------------------------------------------------------------------------------
params = {
    # Contratto
    'N': 100.0,              # Valore facciale
    'T': 5.0,                # Scadenza in anni
    'c': 0.015,              # Tasso cedolare annuo (1.5%)
    'freq_cedole': 2,        # Frequenza cedolare semestrale
    'I0': 100.0,             # Indice prezzi iniziale
    
    # Processo Inflazione (Ornstein-Uhlenbeck)
    'i0': 0.025,             # Tasso d'inflazione iniziale
    'kappa_i': 0.80,         # Velocità di riallineamento
    'theta_i': 0.020,        # Media di lungo periodo
    'sigma_i': 0.015,        # Volatilità inflazione
    
    # Processo Tasso Nominale (Cox-Ingersoll-Ross)
    'r0': 0.030,             # Tasso nominale iniziale
    'kappa_r': 1.20,         # Velocità di riallineamento
    'theta_r': 0.030,        # Media di lungo periodo
    'sigma_r': 0.10,         # Volatilità tasso
    
    # Dipendenza e Simulazione
    'rho': 0.25,             # Correlazione shock browniani
    'dt': 1/12,              # Passo temporale mensile
    'M': 50000,              # Numero di scenari Monte Carlo
    'seed': 12345            # Seed casuale per riproducibilità
}

# ------------------------------------------------------------------------------
# 2. Costruzione della Griglia Temporale
# ------------------------------------------------------------------------------
dt = params['dt']
T = params['T']
K = int(round(T / dt))                          # Numero di passi temporali (60)
t_grid = np.linspace(0, T, K + 1)                # Vettore dei tempi [0, dt, 2dt, ..., T]

# Date delle cedole semestrali: t = 0.5, 1.0, 1.5, ..., 5.0
n_cedole = int(params['T'] * params['freq_cedole'])
t_cedole = np.array([j * 0.5 for j in range(1, n_cedole + 1)])

# Indici dei passi temporali corrispondenti alle date cedolari
idx_cedole = [int(round(tc / dt)) for tc in t_cedole]

# ------------------------------------------------------------------------------
# 3. Generazione degli Shock Correlati tramite Decomposizione di Cholesky
# ------------------------------------------------------------------------------
np.random.seed(params['seed'])
M = params['M']

# Generazione di M x K shock indipendenti N(0,1)
Z1 = np.random.normal(0.0, 1.0, size=(M, K))
Z2 = np.random.normal(0.0, 1.0, size=(M, K))

# Decomposizione di Cholesky Bivariata:
# Z_i = Z1
# Z_r = rho * Z1 + sqrt(1 - rho^2) * Z2
rho = params['rho']
Z_i = Z1
Z_r = rho * Z1 + np.sqrt(1.0 - rho**2) * Z2

# ------------------------------------------------------------------------------
# 4. Controllo 1: Verifica Empirica della Correlazione degli Shock
# ------------------------------------------------------------------------------
rho_empirica = np.corrcoef(Z_i.flatten(), Z_r.flatten())[0, 1]
tolleranza_rho = 0.01
scostamento_rho = abs(rho_empirica - rho)

# Output informativo
print("=" * 70)
print("TAPPA 1: INIZIALIZZAZIONE E GENERAZIONE SHOCK COMPLETATA")
print("=" * 70)
print(f"Griglia temporale: {K} passi mensili da dt = {dt:.4f} anni (T = {T} anni).")
print(f"Date cedole semestrali (passi griglia): {idx_cedole}")
print(f"Dimensioni matrici shock casuali: {Z_i.shape} (Scenari x Passi)")
print("-" * 70)
print(f"Correlazione teorica assegnata (rho): {rho:.4f}")
print(f"Correlazione empirica simulata:       {rho_empirica:.4f}")
print(f"Scostamento assoluto:                 {scostamento_rho:.6f}")
print("=" * 70)

# Test d'arresto di robustezza (Controllo 1)
assert scostamento_rho < tolleranza_rho, (
    f"ERRORE DI CONTROLLO 1: La correlazione empirica {rho_empirica:.4f} "
    f"differisce da quella teorica {rho:.4f} oltre la tolleranza {tolleranza_rho}."
)
print("CONTROLLO 1 SUPERATO: La correlazione empirica è coerente con il parametro rho.")
```

# Tappa 2 — Simulazione delle Traiettorie dell'Inflazione (OU) e del Tasso Nominale (CIR)

## 1. Copia fedele della richiesta dello studente
> Costruisci la Tappa 2.  
> A questo punto bisogna usare gli shock generati prima per simulare le traiettorie dell’inflazione e del tasso nominale con i due modelli assegnati. Mi sembra importante ottenere le matrici delle traiettorie e controllare se nella simulazione del CIR compaiono valori negativi del tasso.

---

## 2. Sintesi logica degli output previsti
In questa tappa vengono simulate le traiettorie temporali dei due fattori di rischio stocastici per l'intera griglia temporale $t_k \in [0, T]$ su $M = 50.000$ scenari:

1. **Simulazione Tasso di Inflazione $i_t$ (Ornstein-Uhlenbeck):** Applicazione dello schema di discretizzazione di Euler-Maruyama al processo di Ornstein-Uhlenbeck:
   $$i_{t+\Delta t} = i_t + \kappa_i (\theta_i - i_t) \Delta t + \sigma_i \sqrt{\Delta t} Z_t^{(i)}$$

2. **Simulazione Tasso Nominale $r_t$ (Cox-Ingersoll-Ross):** Discretizzazione del processo CIR per il tasso di interesse short-rate, implementando il troncamento sotto radice $\sqrt{\max(r_t, 0)}$ per garantire la stabilità numerica del termine di diffusione:
   $$r_{t+\Delta t} = r_t + \kappa_r (\theta_r - r_t) \Delta t + \sigma_r \sqrt{\max(r_t, 0)} \sqrt{\Delta t} Z_t^{(r)}$$

3. **Matrici di Output:** Produzione di due matrici NumPy di dimensione $(M, K+1)$ (`i_sim` ed `r_sim`), dove la prima colonna corrisponde alle condizioni iniziali $i_0 = 2{,}5\%$ e $r_0 = 3{,}0\%$.

4. **Controllo di Robustezza (Controllo 4):** Verifica e rendicontazione della presenza di eventuali valori negativi generati provvisoriamente durante il passo discretizzato del processo CIR ($r_t < 0$), e verifica del rispetto della condizione di Feller ($2\kappa_r \theta_r \geq \sigma_r^2$).

```python
# ==============================================================================
# TAPPA 2: Simulazione delle Traiettorie dell'Inflazione e del Tasso Nominale
# ==============================================================================

import numpy as np
import pandas as pd

# ------------------------------------------------------------------------------
# 1. Recupero Parametri e Inizializzazione Matrici
# ------------------------------------------------------------------------------
M = params['M']
K = int(round(params['T'] / params['dt']))
dt = params['dt']
sqrt_dt = np.sqrt(dt)

# Matrici per memorizzare le traiettorie simulate: dimensione (M, K + 1)
# La colonna 0 contiene i valori iniziali a t = 0
i_sim = np.zeros((M, K + 1))
r_sim = np.zeros((M, K + 1))

# Condizioni iniziali
i_sim[:, 0] = params['i0']
r_sim[:, 0] = params['r0']

# ------------------------------------------------------------------------------
# 2. Simulazione Ricorsiva (Euler-Maruyama)
# ------------------------------------------------------------------------------
# Parametri Inflazione (OU)
kappa_i = params['kappa_i']
theta_i = params['theta_i']
sigma_i = params['sigma_i']

# Parametri Tasso Nominale (CIR)
kappa_r = params['kappa_r']
theta_r = params['theta_r']
sigma_r = params['sigma_r']

# Ciclo sui passi temporali (t_k -> t_{k+1})
for k in range(K):
    # --- Inflazione i_t (Ornstein-Uhlenbeck) ---
    i_curr = i_sim[:, k]
    di = kappa_i * (theta_i - i_curr) * dt + sigma_i * sqrt_dt * Z_i[:, k]
    i_sim[:, k + 1] = i_curr + di
    
    # --- Tasso Nominale r_t (Cox-Ingersoll-Ross) ---
    r_curr = r_sim[:, k]
    # Troncamento per il termine sotto radice (gestione discreta CIR)
    r_pos = np.maximum(r_curr, 0.0)
    dr = kappa_r * (theta_r - r_curr) * dt + sigma_r * np.sqrt(r_pos) * sqrt_dt * Z_r[:, k]
    r_sim[:, k + 1] = r_curr + dr

# ------------------------------------------------------------------------------
# 3. Verifica Condizione di Feller e Controllo 4 (Anomalie CIR)
# ------------------------------------------------------------------------------
feller_lhs = 2.0 * kappa_r * theta_r
feller_rhs = sigma_r**2
feller_soddisfatta = feller_lhs >= feller_rhs

# Conteggio valori negativi generati nella discretizzazione CIR
conteggio_r_negativi = np.sum(r_sim < 0)
totale_punti_griglia = M * (K + 1)
pct_r_negativi = (conteggio_r_negativi / totale_punti_griglia) * 100

# ------------------------------------------------------------------------------
# 4. Formattazione Tabellare degli Output e Verifiche
# ------------------------------------------------------------------------------
sintesi_tappa2 = pd.DataFrame({
    'Metrica / Controllo': [
        'Dimensioni Matrice Inflazione (i_sim)',
        'Dimensioni Matrice Tasso Nominale (r_sim)',
        'Condizione di Feller (2 * kappa_r * theta_r >= sigma_r^2)',
        'Feller LHS (2 * kappa_r * theta_r)',
        'Feller RHS (sigma_r^2)',
        'Controllo 4: Valori r_t < 0 rilevati',
        'Controllo 4: Percentuale punti griglia con r_t < 0',
        'Integrità Dati (Assenza NaN / Inf)'
    ],
    'Valore / Esito': [
        f"{i_sim.shape[0]} scenari x {i_sim.shape[1]} passi",
        f"{r_sim.shape[0]} scenari x {r_sim.shape[1]} passi",
        "Soddisfatta" if feller_soddisfatta else "Non Soddisfatta",
        f"{feller_lhs:.4f}",
        f"{feller_rhs:.4f}",
        f"{conteggio_r_negativi:,}",
        f"{pct_r_negativi:.4f}%",
        "Superato (0 NaN / Inf)" if (not np.isnan(i_sim).any() and not np.isnan(r_sim).any()) else "Fallito"
    ]
})

print("=" * 70)
print("TAPPA 2: SIMULAZIONE TRAIETTORIE TASSI COMPLETATA")
print("=" * 70)
print(sintesi_tappa2.to_string(index=False))
print("=" * 70)

# Assert di integrità e continuità numerica
assert not np.isnan(i_sim).any(), "ERRORE: Rilevati valori NaN nelle traiettorie dell'inflazione."
assert not np.isnan(r_sim).any(), "ERRORE: Rilevati valori NaN nelle traiettorie del tasso nominale."
assert not np.isinf(i_sim).any(), "ERRORE: Rilevati valori Inf nelle traiettorie dell'inflazione."
assert not np.isinf(r_sim).any(), "ERRORE: Rilevati valori Inf nelle traiettorie del tasso nominale."

print("\nCONTROLLO INTEGRITÀ SUPERATO: Matrici i_sim e r_sim generate correttamente.")
```

# Tappa 3 — Costruzione del Coefficiente di Indicizzazione $J_t$ e dei Fattori di Sconto $D(0,t)$

> Mi sembra utile che l'output mostri almeno che le matrici ottenute abbiano le dimensioni corrette, che i valori di (J_t) siano sempre positivi e che non ci siano valori mancanti o infiniti.

---

## 2. Sintesi logica degli output previsti
In questa tappa vengono calcolati i due processi fondamentali per la valutazione del titolo garantito dall'inflazione:

1. **Coefficiente di Indicizzazione Capitale/Prezzi $J_t$:**
   Costruito integrando la traiettoria dell'inflazione stocastica $i_u$ mediante approssimazione trapezoidale o rettangolare (somma sulla griglia temporale):
   $$J_{t_k} = \exp\left( \sum_{m=0}^{k-1} i_{t_m} \Delta t \right), \quad \text{con } J_0 = 1.0$$
   Poiché $J_{t_k}$ rappresenta il rapporto tra l'indice dei prezzi al tempo $t_k$ e quello iniziale $I_0$ ($J_{t_k} = I_{t_k}/I_0$), la funzione esponenziale garantisce $J_{t_k} > 0$ strictly positive in ogni scenario.

2. **Fattore di Sconto Cumulato $D(0, t_k)$:**
   Calcolato come sconto stocastico ottenuto dall'accumulazione del tasso di interesse di breve termine nominale $r_u$:
   $$D(0, t_k) = \exp\left( -\sum_{m=0}^{k-1} r_{t_m} \Delta t \right), \quad \text{con } D(0, 0) = 1.0$$

3. **Verifiche di Integrità e Controlli Numerici:**
   - Verifica delle dimensioni delle matrici $J$ e $D$ ($M \times (K+1)$).
   - Controllo sulla stretta positività del coefficiente di indicizzazione ($J_{t_k} > 0$).
   - Controllo di continuità numerica e assenza di anomalie (`NaN` / `Inf`).
   - Tabellazione dei risultati in un `pandas.DataFrame` formattato in colonne destinate a metrica e valore/esito.

```python
# ==============================================================================
# TAPPA 3: Calcolo del Coefficiente di Indicizzazione J_t e Fattori di Sconto D(0,t)
# ==============================================================================

import numpy as np
import pandas as pd

# ------------------------------------------------------------------------------
# 1. Calcolo del Coefficiente di Indicizzazione J_t e dei Fattori di Sconto D(0,t)
# ------------------------------------------------------------------------------
dt = params['dt']
M = params['M']
K = int(round(params['T'] / dt))

# Somme cumulate sulla griglia temporale (somma sui passi k = 0, ..., K-1)
# np.cumsum sui passi temporali (axis=1) preceduto da un pad di zeri a t=0
i_integral = np.zeros((M, K + 1))
r_integral = np.zeros((M, K + 1))

i_integral[:, 1:] = np.cumsum(i_sim[:, :-1] * dt, axis=1)
r_integral[:, 1:] = np.cumsum(r_sim[:, :-1] * dt, axis=1)

# Calcolo del coefficiente di indicizzazione J_t e del fattore di sconto D(0,t)
J_sim = np.exp(i_integral)
D_sim = np.exp(-r_integral)

# ------------------------------------------------------------------------------
# 2. Controlli Numerici e Valutazione delle Proprietà
# ------------------------------------------------------------------------------
# Controllo positività di J_t
min_J = np.min(J_sim)
J_positivo = min_J > 0.0

# Controlli su NaN e Inf
no_nan_J = not np.isnan(J_sim).any()
no_inf_J = not np.isinf(J_sim).any()
no_nan_D = not np.isnan(D_sim).any()
no_inf_D = not np.isinf(D_sim).any()

# Statistiche descrittive sintetiche a scadenza T = 5 anni (colonna K)
mean_J_T = np.mean(J_sim[:, -1])
mean_D_T = np.mean(D_sim[:, -1])

# ------------------------------------------------------------------------------
# 3. Presentazione Tabellare degli Output e dei Controlli
# ------------------------------------------------------------------------------
sintesi_tappa3 = pd.DataFrame({
    'Metrica / Controllo': [
        'Dimensioni Matrice Coefficiente Indicizzazione (J_sim)',
        'Dimensioni Matrice Fattori di Sconto (D_sim)',
        'Valore Minimo assoluto di J_t',
        'Controllo Positività J_t > 0',
        'Valore Medio J_T a Scadenza (T = 5 anni)',
        'Valore Medio D(0,T) a Scadenza (T = 5 anni)',
        'Integrità J_sim (Assenza NaN / Inf)',
        'Integrità D_sim (Assenza NaN / Inf)'
    ],
    'Valore / Esito': [
        f"{J_sim.shape[0]} scenari x {J_sim.shape[1]} passi",
        f"{D_sim.shape[0]} scenari x {D_sim.shape[1]} passi",
        f"{min_J:.6f}",
        "Superato (J_t > 0 sempre)" if J_positivo else "Fallito",
        f"{mean_J_T:.6f}",
        f"{mean_D_T:.6f}",
        "Superato (0 NaN / Inf)" if (no_nan_J and no_inf_J) else "Fallito",
        "Superato (0 NaN / Inf)" if (no_nan_D and no_inf_D) else "Fallito"
    ]
})

print("=" * 70)
print("TAPPA 3: CALCOLO INDICIZZAZIONE J_t E SCONTI D(0,t) COMPLETATO")
print("=" * 70)
print(sintesi_tappa3.to_string(index=False))
print("=" * 70)

# ------------------------------------------------------------------------------
# 4. Assert di Arresto di Robustezza
# ------------------------------------------------------------------------------
assert J_positivo, "ERRORE CRITICO: Trovati valori di J_t <= 0 nella matrice simulata!"
assert no_nan_J and no_inf_J, "ERRORE: Rilevati NaN o Inf nella matrice J_sim!"
assert no_nan_D and no_inf_D, "ERRORE: Rilevati NaN o Inf nella matrice D_sim!"

print("\nCONTROLLO INTEGRITÀ SUPERATO: Matrici J_sim e D_sim valide e matematicamente coerenti.")
```

# Tappa 4 — Calcolo dei Flussi di Cassa Indicizzati, del Rimborso Protetto e del Valore Attuale del Bond per Scenario

## 1. Copia fedele della richiesta dello studente

> Vorrei che l'output mostrasse in modo chiaro almeno il controllo sul fatto che il rimborso non scenda mai sotto il valore nominale e alcune informazioni sintetiche sui valori attuali ottenuti.

## 2. Sintesi logica degli output previsti
In questa tappa vengono integrati i risultati della Tappa 3 ($J_t$ e $D(0,t)$) per determinare la struttura dei flussi di cassa (cash flow) dell'Inflation-Linked Bond e il relativo valore attuale in ciascuno scenario Monte Carlo:

1. **Cedole Indicizzate ($C_t$):**
   Calcolate nei momenti di stacco cedolare $t \in \{1, 2, 3, 4, 5\}$ applicando il tasso cedolare reale fisso $c$ al Capitale Nominale Rivalutato tramite $J_t$:
   $$C_t = N \cdot c \cdot J_t$$

2. **Rimborso del Capitale a Scadenza Protetto ($R_T$):**
   In presenza di clausola di protezione dal rischio deflazionistico (floor sul capitale), il valore di rimborso al tempo $T$ garantisce la restituzione di almeno il valore nominale $N$:
   $$R_T = N \cdot \max(1.0, J_T)$$

3. **Valore Attuale del Bond per Scenario ($PV_{\omega}$):**
   Per ciascuno scenario $\omega = 1, \dots, M$, il valore attuale all'epoca $t=0$ è dato dalla somma dei flussi di cassa scomputati mediante i rispettivi fattori di sconto stocastici $D(0, t)$:
   $$PV_{\omega} = \sum_{t=1}^{T-1} C_t(\omega) \cdot D(0, t, \omega) + \Big(C_T(\omega) + R_T(\omega)\Big) \cdot D(0, T, \omega)$$

4. **Controlli di Integrità e Sintesi Statistica:**
   - Verifica della clausola di protezione: $R_T(\omega) \ge N$ per ogni scenario $\omega$.
   - Statistiche sintetiche sui valori attuali $PV_{\omega}$ (media, deviazione standard, minimo e massimo).
   - Verifica di assenza di valori `NaN` o `Inf`.
   - Presentazione tabellare dei controlli e dei risultati tramite `pandas.DataFrame`.

```python
# ==============================================================================
# TAPPA 4: Calcolo Cedole Indicizzate, Rimborso Protetto e Valore Attuale (PV)
# ==============================================================================

import numpy as np
import pandas as pd

# ------------------------------------------------------------------------------
# 1. Estrazione parametri e griglia temporale delle scadenze cedolari
# ------------------------------------------------------------------------------
N = params['N']             # Valore nominale del titolo
c = params['c']             # Tasso cedolare reale annuo
dt = params['dt']
T_years = int(params['T'])  # Scadenza in anni (e.g., 5 anni)

# Indici dei passi temporali corrispondenti alle scadenze annuali t = 1, 2, ..., T
steps_per_year = int(round(1.0 / dt))
coupon_step_indices = [t * steps_per_year for t in range(1, T_years + 1)]

# ------------------------------------------------------------------------------
# 2. Calcolo dei Flussi di Cassa per ogni scenario
# ------------------------------------------------------------------------------
M = J_sim.shape[0]

# Matrice delle cedole indicizzate staccate negli anni t = 1, ..., T: shape (M, T)
coupons_sim = np.zeros((M, T_years))
for idx, step in enumerate(coupon_step_indices):
    coupons_sim[:, idx] = N * c * J_sim[:, step]

# Rimborso del Capitale a Scadenza con Protezione Deflazionistica (Floor a N)
# Step finale: coupon_step_indices[-1]
J_T = J_sim[:, coupon_step_indices[-1]]
redemption_T = N * np.maximum(1.0, J_T)

# ------------------------------------------------------------------------------
# 3. Calcolo del Valore Attuale (PV) per Scenario
# ------------------------------------------------------------------------------
# Estrazione dei fattori di sconto D(0, t) alle date di stacco
discount_factors_coupon = np.zeros((M, T_years))
for idx, step in enumerate(coupon_step_indices):
    discount_factors_coupon[:, idx] = D_sim[:, step]

# Attualizzazione delle cedole t = 1, ..., T
pv_coupons = np.sum(coupons_sim * discount_factors_coupon, axis=1)

# Attualizzazione del rimborso del capitale a T
pv_redemption = redemption_T * discount_factors_coupon[:, -1]

# Valore Attuale Totale del Bond per ciascuno scenario
PV_sim = pv_coupons + pv_redemption

# ------------------------------------------------------------------------------
# 4. Controlli Numerici e Sintesi Statistica
# ------------------------------------------------------------------------------
# Controllo che il rimborso non scenda mai sotto il valore nominale N
min_redemption = np.min(redemption_T)
floor_rispettato = min_redemption >= N - 1e-12

# Controlli su NaN / Inf
no_nan_pv = not np.isnan(PV_sim).any()
no_inf_pv = not np.isinf(PV_sim).any()

# Statistiche sui Valori Attuali (PV)
mean_pv = np.mean(PV_sim)
std_pv = np.std(PV_sim, ddof=1)
min_pv = np.min(PV_sim)
max_pv = np.max(PV_sim)

# ------------------------------------------------------------------------------
# 5. Presentazione Tabellare degli Output e dei Controlli
# ------------------------------------------------------------------------------
sintesi_tappa4 = pd.DataFrame({
    'Metrica / Controllo': [
        'Valore Nominale di Riferimento (N)',
        'Rimborso Minimo Registrato a Scadenza (R_T min)',
        'Controllo Floor Capitale (R_T >= N)',
        'Valore Attuale Medio (PV medio)',
        'Deviazione Standard del Valore Attuale (PV std)',
        'Valore Attuale Minimo (PV min)',
        'Valore Attuale Massimo (PV max)',
        'Integrità Valori Attuali (Assenza NaN / Inf)'
    ],
    'Valore / Esito': [
        f"{N:,.2f} €",
        f"{min_redemption:,.2f} €",
        "Superato (Rimborso >= N garantito)" if floor_rispettato else "Fallito",
        f"{mean_pv:,.4f} €",
        f"{std_pv:,.4f} €",
        f"{min_pv:,.4f} €",
        f"{max_pv:,.4f} €",
        "Superato (0 NaN / Inf)" if (no_nan_pv and no_inf_pv) else "Fallito"
    ]
})

print("=" * 70)
print("TAPPA 4: CALCOLO FLUSSI DI CASSA E VALORI ATTUALI COMPLETATO")
print("=" * 70)
print(sintesi_tappa4.to_string(index=False))
print("=" * 70)

# ------------------------------------------------------------------------------
# 6. Assert di Arresto di Robustezza
# ------------------------------------------------------------------------------
assert floor_rispettato, "ERRORE CRITICO: Trovati valori di rimborso inferiori al valore nominale N!"
assert no_nan_pv and no_inf_pv, "ERRORE: Rilevati NaN o Inf nel calcolo del Valore Attuale!"

print("\nCONTROLLO INTEGRITÀ SUPERATO: Flussi di cassa e valori attuali calcolati correttamente.")
```

In questa tappa vengono calcolati ed esposti gli output analitici e finanziari finali del modello Monte Carlo:
1. **Prezzo Monte Carlo del Titolo ($\hat{P}_0$)**: Valore attuale medio complessivo (somma dei VA medi di cedole e rimborso a scadenza).
2. **Errore Standard ($SE$) e Intervallo di Confidenza al 95% ($CI_{95\%}$)**: Valutazione della precisione della stima stocastica.

5. **Analisi di Convergenza al Crescere di $M$**: Verifica dell'effetto della dimensione campionaria ($M$) sulla riduzione dell'errore standard ($SE \propto 1/\sqrt{M}$).
6. **Controlli di Integrità Finanziaria e Numerica**:
   - Controllo di coerenza contabile: $\hat{P}_0 = \mathbb{E}[PV_{\text{cedole}}] + \mathbb{E}[PV_{\text{rimborso}}]$.

   - Test di arresto rigorosi via `assert`.

```python
import numpy as np
import pandas as pd

# ==============================================================================
# TAPPA 5: STIMA DEL PREZZO, ERRORE STANDARD, STATISTICHE E CONVERGENZA
# ==============================================================================

# ------------------------------------------------------------------------------
# 1. STIMA DEL PREZZO MONTE CARLO, SE E INTERVALLO DI CONFIDENZA (95%)
# ------------------------------------------------------------------------------
# Il valore attuale totale per ciascun scenario m è PV_total_sim
price_mc = np.mean(PV_total_sim)
std_pv = np.std(PV_total_sim, ddof=1)
se_mc = std_pv / np.sqrt(M)

# Intervallo di confidenza al 95% (z_alpha/2 = 1.96)
z_95 = 1.960
ci_lower = price_mc - z_95 * se_mc
ci_upper = price_mc + z_95 * se_mc

# Decomposizione del prezzo medio in Quota Cedole e Quota Rimborso
pv_coupons_mean = np.mean(PV_coupons_sim)
pv_redemption_mean = np.mean(PV_redemption_sim)

# ------------------------------------------------------------------------------
# 2. PROBABILITÀ DI EVENTI CRITICI E STATISTICHE DESCRITTIVE DELLE VARIABILI
# ------------------------------------------------------------------------------
# Probabilità che J_T < 1 (Indice al sotto della parità)
prob_J_lt_1 = np.mean(J_T_sim < 1.0)

def compute_stats(data):
    return {
        'Media': np.mean(data),
        'Dev. Std.': np.std(data, ddof=1),
        'Minimo': np.min(data),
        'Mediana (Q2)': np.median(data),
        'Massimo': np.max(data)
    }

df_stats_vars = pd.DataFrame({
    'Inflazione Accumulata (i_T)': compute_stats(i_T_sim),
    'Tasso Reale Accumulato (r_T)': compute_stats(r_T_sim),
    'Indice Riferimento (J_T)': compute_stats(J_T_sim)
}).T

# ------------------------------------------------------------------------------
# 3. ANALISI DI CONVERGENZA AL CRESCERE DI M
# ------------------------------------------------------------------------------
sample_sizes = [1000, 5000, 10000, 25000, M]
convergence_results = []

for m_sub in sample_sizes:
    pv_sub = PV_total_sim[:m_sub]
    p_mean_sub = np.mean(pv_sub)
    se_sub = np.std(pv_sub, ddof=1) / np.sqrt(m_sub)
    ci_l_sub = p_mean_sub - z_95 * se_sub
    ci_u_sub = p_mean_sub + z_95 * se_sub
    
    convergence_results.append({
        'N. Simulazioni (M)': f"{m_sub:,}",
        'Prezzo Stimato (€)': f"{p_mean_sub:,.4f}",
        'Errore Standard SE (€)': f"{se_sub:,.6f}",
        'IC 95% Inferiore (€)': f"{ci_l_sub:,.4f}",
        'IC 95% Superiore (€)': f"{ci_u_sub:,.4f}"
    })

df_convergence = pd.DataFrame(convergence_results)

# ------------------------------------------------------------------------------
# 4. CONTROLLI NUMERICI E TEST DI ARRESTO (ASSERT)
# ------------------------------------------------------------------------------
# Controllo 1: Coerenza Additiva del Prezzo (Somma Componenti == Prezzo Totale)
diff_decomposizione = abs(price_mc - (pv_coupons_mean + pv_redemption_mean))
assert diff_decomposizione < 1e-10, f"ERRORE: Discrepanza nella scomposizione del prezzo: {diff_decomposizione}"

# Controllo 2: Riduzione Errore Standard con M (SE M=1000 > SE M=50000)
se_1k = np.std(PV_total_sim[:1000], ddof=1) / np.sqrt(1000)
assert se_mc < se_1k, "ERRORE: L'errore standard non dimunuisce all'aumentare delle simulazioni!"

# Controllo 3: Coerenza Intervallo di Confidenza
assert ci_lower < price_mc < ci_upper, "ERRORE: Intervallo di confidenza incoerente!"

# ------------------------------------------------------------------------------
# 5. PRESENTAZIONE TABELLARE DEGLI OUTPUT
# ------------------------------------------------------------------------------
sintesi_prezzo = pd.DataFrame({
    'Metrica / Valutazione': [
        'Prezzo Monte Carlo stimato (P_0)',
        'Quota VA Medio Cedole',
        'Quota VA Medio Rimborso Capitale',
        'Errore Standard (SE)',
        'IC 95% Limite Inferiore',
        'IC 95% Limite Superiore',
        'Probabilità P(J_T < 1.0)',
        'Test Additività Componenti',
        'Test Monotonia SE vs M'
    ],
    'Valore / Esito': [
        f"{price_mc:,.4f} €",
        f"{pv_coupons_mean:,.4f} €",
        f"{pv_redemption_mean:,.4f} €",
        f"{se_mc:,.6f} €",
        f"{ci_lower:,.4f} €",
        f"{ci_upper:,.4f} €",
        f"{prob_J_lt_1:.2%}",
        "Superato (Differenza < 1e-10)",
        "Superato (SE diminuisce con M)"
    ]
})

print("=" * 80)
print("TAPPA 5: VALUTAZIONE FINALE DEL TITOLO ED ERRORE STOCASTICO")
print("=" * 80)
print("\n--- SINTESI STIMA PREZZO E INTEGRITÀ MODELLO ---")
print(sintesi_prezzo.to_string(index=False))

print("\n--- STATISTICHE DESCRITTIVE DELLE VARIABILI SIMULATE ---")
print(df_stats_vars.round(6).to_string())

print("\n--- ANALISI DI CONVERGENZA AL CRESCERE DEL NUMERO DI SIMULAZIONI (M) ---")
print(df_convergence.to_string(index=False))
print("=" * 80)
```
