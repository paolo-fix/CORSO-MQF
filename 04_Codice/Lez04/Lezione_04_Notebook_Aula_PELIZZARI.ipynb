# Perdita obbligazionaria condizionata a regimi di tasso

## Sintesi del caso

Il caso analizza un portafoglio obbligazionario a tasso fisso con valore iniziale $V_0 = 10.000.000$ euro e duration modificata $D = 6$ anni, esposto a shock dei rendimenti di mercato a un periodo $\Delta y$. 

La variabile finale di interesse è la perdita del portafoglio $L$, definita tramite approssimazione lineare come:

$$L = D V_0 \Delta y$$

L'informazione disponibile è rappresentata dalla partizione dello spazio degli esiti $\Omega$ nei tre regimi $A_1$ (Disinflazione ordinata), $A_2$ (Inflazione persistente) e $A_3$ (Repricing severo), che generano la sigma-algebra informativa:

$$\mathcal{G} = \sigma(A_1, A_2, A_3)$$

Condizionatamente a ciascun regime $A_g$, lo shock di rendimento segue una distribuzione normale:

$$\Delta y \mid A_g \sim \mathcal{N}(\mu_g, \sigma_g^2), \qquad g = 1, 2, 3$$

La domanda quantitativa fondamentale è: **Qual è la distribuzione della perdita del portafoglio e come cambia la perdita attesa condizionando sui diversi regimi di tasso?**

---

## Quantità da stimare o calcolare

Si distinguono nettamente le seguenti quantità matematiche ed empiriche:

- **Quantità non condizionate (globali):**
  - Valore atteso della perdita: $\mathbb{E}[L]$
  - Probabilità di superamento della soglia di perdita $\ell = 500.000$ euro: $\mathbb{P}(L > \ell)$
  - Quantili della perdita ad un determinato livello di confidenza $\alpha$: $q_\alpha(L)$

- **Quantità condizionate ai singoli regimi:**
  - Per ciascun regime $g \in \{1, 2, 3\}$, la perdita attesa condizionata: $\mathbb{E}[L \mid A_g]$

- **Variabile casuale valore atteso condizionato:**
  - La variabile casuale $\mathcal{G}$-misurabile:

$$\mathbb{E}[L \mid \mathcal{G}] = \sum_{g=1}^{3} \mathbb{E}[L \mid A_g] \mathbf{1}_{A_g}$$

- **Verifica del valore atteso totale:**
  - Rispetto della legge delle aspettative iterate (Legge del Valore Atteso Totale):

$$\mathbb{E}[L] = \sum_{g=1}^{3} \mathbb{P}(A_g) \mathbb{E}[L \mid A_g]$$

```python
# ==============================================================================
# TAPPA 1: Inizializzazione e definizione dei parametri del modello
# ==============================================================================

import numpy as np
import pandas as pd

# ------------------------------------------------------------------------------
# 1. Definizione dei parametri finanziari e computazionali
# ------------------------------------------------------------------------------
V0 = 10_000_000  # Valore iniziale del portafoglio (€)
D = 6            # Duration modificata (anni)
ell = 500_000    # Soglia di perdita di riferimento (€)

n = 50_000       # Numero di simulazioni Monte Carlo
seed = 12345     # Seed per la riproducibilità

# ------------------------------------------------------------------------------
# 2. Definizione della partizione informativa e dei parametri dei regimi
# ------------------------------------------------------------------------------
regimi_nomi = ['A1: Disinflazione ordinata', 'A2: Inflazione persistente', 'A3: Repricing severo']
regimi_codici = ['A1', 'A2', 'A3']

# Probabilità a priori dei regimi P(A_g)
p_regimi = np.array([0.35, 0.45, 0.20])

# Parametri delle distribuzioni condizionate dello shock Delta y | A_g ~ N(mu_g, sigma_g^2)
mu_g = np.array([-0.0010, 0.0030, 0.0100])    # Medie condizionate (in decimali)
sigma_g = np.array([0.0040, 0.0060, 0.0100]) # Deviazioni standard condizionate (in decimali)

# ------------------------------------------------------------------------------
# 3. Costruzione e visualizzazione della Tabella 1: Tabella dei Parametri
# ------------------------------------------------------------------------------
# Sezione A: Parametri Finanziari e Computazionali
df_param_generali = pd.DataFrame({
    'Parametro': ['Valore iniziale portafoglio (V0)', 'Duration modificata (D)', 'Soglia di perdita (ell)', 'Numero simulazioni (n)', 'Seed casuale'],
    'Valore': [f"{V0:,.0f} €".replace(',', '.'), f"{D} anni", f"{ell:,.0f} €".replace(',', '.'), f"{n:,}".replace(',', '.'), str(seed)]
})

# Sezione B: Parametri dei Regimi di Tasso
df_param_regimi = pd.DataFrame({
    'Codice Regime': regimi_codici,
    'Descrizione Regime': ['Disinflazione ordinata', 'Inflazione persistente', 'Repricing severo'],
    'Probabilità P(A_g)': p_regimi,
    'Media mu_g (decimali)': mu_g,
    'Media mu_g (bps)': mu_g * 10_000,
    'Dev. Std. sigma_g (decimali)': sigma_g,
    'Dev. Std. sigma_g (bps)': sigma_g * 10_000
})

print("=" * 80)
print("TABELLA 1A: PARAMETRI FINANZIARI E COMPUTAZIONALI")
print("=" * 80)
display(df_param_generali)

print("\n" + "=" * 80)
print("TABELLA 1B: PROBABILITÀ E PARAMETRI CONDIZIONATI DEI REGIMI DI TASSO")
print("=" * 80)
display(df_param_regimi)

# ------------------------------------------------------------------------------
# 4. CONTROLLO 1: Verifica della somma delle probabilità dei regimi
# ------------------------------------------------------------------------------
somma_prob = np.sum(p_regimi)
esito_controllo_1 = np.isclose(somma_prob, 1.0)

print("\n" + "-" * 80)
print("ESITO CONTROLLO 1 (Scheda Caso §7.1):")
print(f"  - Somma delle probabilità dei regimi: {somma_prob:.6f}")
print(f"  - Condizione sum(P(A_g)) == 1.0: {'SUPERATO [OK]' if esito_controllo_1 else 'FALLITO [ERRORE]'}")
print("-" * 80)

assert esito_controllo_1, "ERRORE CRITICO: La somma delle probabilità dei regimi non è pari a 1!"
```
