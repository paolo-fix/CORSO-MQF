# Perdita obbligazionaria condizionata a regimi di tasso

## Sintesi del caso

Il caso analizza un portafoglio obbligazionario a tasso fisso con valore iniziale $V_0 = 10.000.000$ euro e duration modificata $D = 6$ anni, esposto a shock dei rendimenti di mercato a un periodo $\Delta y$. 

La variabile finale di interesse è la perdita del portafoglio $L$, definita tramite approssimazione lineare come:

$$L = D V_0 \Delta y$$

L'informazione disponibile è rappresentata dalla partizione dello spazio degli esiti $\Omega$ nei tre regimi $A_1$ (Disinflazione ordinata), $A_2$ (Inflazione persistente) e $A_3$ (Repricing severo), che generano la sigma-algebra informativa:

$$\mathcal{G} = \sigma(A_1, A_2, A_3)$$

Condizionatamente a ciascun regime $A_g$, lo shock di rendimento segue una distribuzione normale:

$$\Delta y \mid A_g \sim \mathcal{N}(\mu_g, \sigma_g^2), \qquad g = 1, 2, 3$$

La domanda quantitativa fondamentale è: **Qual è la distribuzione della perdita del portafoglio e come cambia la perdita attesa condizionando sui diversi regimi di tasso?**

---

## Quantità da stimare o calcolare

Si distinguono nettamente le seguenti quantità matematiche ed empiriche:

- **Quantità non condizionate (globali):**
  - Valore atteso della perdita: $\mathbb{E}[L]$
  - Probabilità di superamento della soglia di perdita $\ell = 500.000$ euro: $\mathbb{P}(L > \ell)$
  - Quantili della perdita ad un determinato livello di confidenza $\alpha$: $q_\alpha(L)$

- **Quantità condizionate ai singoli regimi:**
  - Per ciascun regime $g \in \{1, 2, 3\}$, la perdita attesa condizionata: $\mathbb{E}[L \mid A_g]$

- **Variabile casuale valore atteso condizionato:**
  - La variabile casuale $\mathcal{G}$-misurabile:

$$\mathbb{E}[L \mid \mathcal{G}] = \sum_{g=1}^{3} \mathbb{E}[L \mid A_g] \mathbf{1}_{A_g}$$

- **Verifica del valore atteso totale:**
  - Rispetto della legge delle aspettative iterate (Legge del Valore Atteso Totale):

$$\mathbb{E}[L] = \sum_{g=1}^{3} \mathbb{P}(A_g) \mathbb{E}[L \mid A_g]$$

## Sintesi logica della tappa

### Input
- **Dati originati dalla Tappa 1:**
  - Numero di simulazioni: $n = 50.000$.
  - Seed casuale: `12345`.
  - Vettore delle probabilità dei regimi: $\mathbf{p} = [0{,}35, 0{,}45, 0{,}20]$.
  - Vettori dei parametri condizionati degli shock: $\boldsymbol{\mu} = [-0{,}0010, 0{,}0030, 0{,}0100]$ e $\boldsymbol{\sigma} = [0{,}0040, 0{,}0060, 0{,}0100]$.
  - Codici e descrizioni dei regimi $A_1, A_2, A_3$.

### Operazioni
1. Impostazione del seed pseudo-casuale (`np.random.seed(12345)`) per garantire la perfetta riproducibilità.
2. Estrazione campionaria dei regimi informativi $g_i \in \{1, 2, 3\}$ per $i = 1, \dots, n$ secondo la distribuzione categoriale definita dalle probabilità a priori $\mathbf{p}$.
3. Generazione condizionata degli shock di rendimento $\Delta y_i \mid A_{g_i} \sim \mathcal{N}(\mu_{g_i}, \sigma_{g_i}^2)$ per ogni simulazione $i$.
4. Calcolo delle frequenze empiriche dei regimi e delle medie e deviazioni standard empiriche condizionate dello shock $\Delta y \mid A_g$.
5. Costruzione e visualizzazione della **Tabella 2 (Tabella di controllo della simulazione)**.
6. Esecuzione del **Controllo 2** sulla convergenza delle statistiche campionarie ai parametri teorici.

### Output prodotti
- Vettore dei regimi estratti `regimi` (array NumPy di dimensione $n \times 1$ con valori in $\{1, 2, 3\}$).
- Vettore degli shock di rendimento `delta_y` (array NumPy di dimensione $n \times 1$).
- **Tabella 2:** Tabella di controllo della simulazione con frequenze empiriche dei regimi e confronto tra parametri teorici ed empirici di $\Delta y \mid A_g$.
- **Controllo 2:** Verifica della convergenza campionaria e della coerenza della simulazione Monte Carlo.

### Controlli eseguiti
- **Controllo 2 (Scheda Caso §7.2):** Verifica che le frequenze empiriche dei regimi siano coerenti con le probabilità teoriche $\mathbb{P}(A_g)$ e che le medie e deviazioni standard empiriche dello shock condizionato $\Delta y \mid A_g$ convergano ai rispettivi parametri $\mu_g$ e $\sigma_g$.

### Uso successivo dell'output
- **Tappa 3:** Calcolo del vettore delle perdite di portafoglio $L = D V_0 \Delta y$ a partire dal vettore `delta_y`.
- **Tappa 5:** Raggruppamento delle perdite per regime mediante il vettore `regimi` per il calcolo delle medie condizionate e la verifica della legge del valore atteso totale.
- **Tappe 6-7:** Visualizzazioni grafiche e sintesi finale dei risultati.

```python
# ==============================================================================
# TAPPA 2: Simulazione Monte Carlo dei regimi e degli shock di rendimento
# ==============================================================================

import numpy as np
import pandas as pd

# ------------------------------------------------------------------------------
# 1. Impostazione del seed casuale per la riproducibilità
# ------------------------------------------------------------------------------
np.random.seed(seed)

# ------------------------------------------------------------------------------
# 2. Estrazione dei regimi informativi (distribuzione categoriale)
# ------------------------------------------------------------------------------
# Generazione di n indici di regime in {0, 1, 2} corrispondenti a {A1, A2, A3}
indici_regimi = np.random.choice([0, 1, 2], size=n, p=p_regimi)

# Vettore dei regimi (1-based per coerenza con la notazione teorica A_1, A_2, A_3)
regimi = indici_regimi + 1

# ------------------------------------------------------------------------------
# 3. Generazione condizionata degli shock di rendimento Delta y | A_g
# ------------------------------------------------------------------------------
# Estrazione per ciascuna simulazione i dalla N(mu_{g_i}, sigma_{g_i}^2)
delta_y = np.random.normal(loc=mu_g[indici_regimi], scale=sigma_g[indici_regimi])

# ------------------------------------------------------------------------------
# 4. Calcolo delle statistiche empiriche di controllo
# ------------------------------------------------------------------------------
freq_assolute = np.bincount(indici_regimi, minlength=3)
freq_empiriche = freq_assolute / n

mu_empirici = np.array([np.mean(delta_y[regimi == g]) for g in [1, 2, 3]])
sigma_empirici = np.array([np.std(delta_y[regimi == g], ddof=1) for g in [1, 2, 3]])

# ------------------------------------------------------------------------------
# 5. Costruzione e visualizzazione della Tabella 2: Controllo della simulazione
# ------------------------------------------------------------------------------
df_tabella_2 = pd.DataFrame({
    'Regime': regimi_codici,
    'Frequenza Assoluta': freq_assolute,
    'Prob. Teorica P(A_g)': p_regimi,
    'Freq. Empirica P_hat(A_g)': freq_empiriche,
    'Scarto Freq. (%)': (freq_empiriche - p_regimi) * 100,
    'mu_g Teorica (bps)': mu_g * 10_000,
    'mu_g Empirica (bps)': mu_empirici * 10_000,
    'sigma_g Teorica (bps)': sigma_g * 10_000,
    'sigma_g Empirica (bps)': sigma_empirici * 10_000
})

print("=" * 90)
print("TABELLA 2: TABELLA DI CONTROLLO DELLA SIMULAZIONE MONTE CARLO")
print("=" * 90)
display(df_tabella_2)

# ------------------------------------------------------------------------------
# 6. CONTROLLO 2: Coerenza delle frequenze e dei parametri condizionati
# ------------------------------------------------------------------------------
# Verifichiamo che i parametri empirici non si discostino in modo anomalo dai teorici (errore campionario < 1%)
toll_freq = 0.01
toll_param = 0.0005  # 5 bps

esito_freq = np.all(np.abs(freq_empiriche - p_regimi) < toll_freq)
esito_mu = np.all(np.abs(mu_empirici - mu_g) < toll_param)
esito_sigma = np.all(np.abs(sigma_empirici - sigma_g) < toll_param)

esito_controllo_2 = esito_freq and esito_mu and esito_sigma

print("\n" + "-" * 90)
print("ESITO CONTROLLO 2 (Scheda Caso §7.2):")
print(f"  - Coerenza frequenze empiriche regimi: {'OK' if esito_freq else 'ANOMALIA'}")
print(f"  - Convergenza medie conditazionate shock: {'OK' if esito_mu else 'ANOMALIA'}")
print(f"  - Convergenza dev.std condizionate shock: {'OK' if esito_sigma else 'ANOMALIA'}")
print(f"  - Esito globale Controllo 2: {'SUPERATO [OK]' if esito_controllo_2 else 'FALLITO [ERRORE]'}")
print("-" * 90)

assert esito_controllo_2, "ERRORE CRITICO: Le frequenze o i parametri empirici differiscono eccessivamente dai valori teorici!"
```
