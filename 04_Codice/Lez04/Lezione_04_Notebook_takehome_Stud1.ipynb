# Caso Studio: Shortfall di Margine Condizionato a Regimi del Costo dell’Energia

## 1. Contesto e Sintesi del Caso
Il caso analizza un'impresa manifatturiera energivora esposta all'incertezza del prezzo unitario dell'energia ($P_E$). L'obiettivo è quantificare il rischio operativo derivante da un aumento dei costi energetici, valutando quando il margine operativo netto scende al di sotto di una soglia minima desiderata ($m^\star$), generando uno shortfall di margine ($S$). L'analisi incorpora l'informazione derivante da tre regimi informativi del mercato energetico ($A_1, A_2, A_3$) costituenti una partizione dello spazio degli esiti.

* **Domanda quantitativa:** Qual è la distribuzione dello shortfall $S$ e come cambia lo shortfall atteso condizionando sui diversi regimi energetici?
* **Variabile finale:** Lo shortfall di margine $S = (m^\star - M)^+$, dove $M = G_0 - Q_E P_E$.
* **Informazione disponibile:** Partizione informativa $\Omega = A_1 \cup A_2 \cup A_3$ generata dalla sigma-algebra $\mathcal{G} = \sigma(A_1, A_2, A_3)$.

---

## 2. Quantità da Stimare o Calcolare
Le quantità richieste sono suddivise tra non condizionate, condizionate e relazioni teoriche:

* **Quantità non condizionate:**
  * $\mathbb{E}[S]$ (Valore atteso dello shortfall)
  * $\mathbb{P}(S > 0)$ (Probabilità di shortfall positivo)
  * $\mathbb{P}(S > s^\star)$ (Probabilità di shortfall severo)
  * $q_{\alpha}(S)$ (Quantili dello shortfall)

* **Quantità condizionate:**
  * $\mathbb{E}[S \mid A_g]$ per $g = 1, 2, 3$ (Valori attesi condizionati rispetto ai singoli regimi, che sono numeri)
  * $\mathbb{E}[S \mid \mathcal{G}] = \sum_{g=1}^{3} \mathbb{E}[S \mid A_g]\mathbf{1}_{A_g}$ (Valore atteso condizionato rispetto alla sigma-algebra, che è una variabile casuale costante sui blocchi)

* **Relazione di verifica:**
  * Formula del valore atteso totale: $\mathbb{E}[S] = \sum_{g=1}^{3} \mathbb{P}(A_g)\mathbb{E}[S \mid A_g]$ e la sua controparte empirica $\widehat{\mathbb{E}}[S] \approx \sum_{g=1}^{3} \widehat{\mathbb{P}}(A_g)\widehat{\mathbb{E}}[S \mid A_g]$.

```python
# ==============================================================================
# TAPPA 1: Inizializzazione e generazione dei regimi informativi
# ==============================================================================

import numpy as np
import pandas as pd

# 1. Definizione dei parametri strutturali del modello
n_simulazioni = 50000
seed_valutazione = 24680

# Impostazione del generatore di numeri casuali per la riproducibilità
rng = np.random.default_rng(seed=seed_valutazione)

# 2. Definizione delle probabilità teoriche dei regimi informativi (A1, A2, A3)
# A1: Normalizzazione energetica
# A2: Tensione persistente
# A3: Shock energetico severo
regimi_etichette = ["A1", "A2", "A3"]
probabilita_teoriche = np.array([0.35, 0.45, 0.20])

# Controllo preliminare: la somma delle probabilità deve essere rigorosamente pari a 1
assert np.isclose(np.sum(probabilita_teoriche), 1.0), "Errore: le probabilità dei regimi non sommano a 1."

# 3. Estrazione casuale dei regimi per ciascuna delle 50.000 repliche Monte Carlo
# Utilizziamo la distribuzione multinomiale / scelta discreta basata sulle probabilità assegnate
indici_regimi = rng.choice(a=[0, 1, 2], size=n_simulazioni, p=probabilita_teoriche)
regimi_simulati = np.array([regimi_etichette[i] for i in indici_regimi])

# 4. Calcolo delle frequenze empiriche simulate
frequenze_assolute = {regime: np.sum(regimi_simulati == regime) for regime in regimi_etichette}
frequenze_empiriche = {regime: count / n_simulazioni for regime, count in frequenze_assolute.items()}

# 5. Costruzione della tabella riassuntiva delle frequenze (Output richiesto)
tabella_frequenze = pd.DataFrame({
    "Regime": regimi_etichette,
    "Descrizione": ["Normalizzazione energetica", "Tensione persistente", "Shock energetico severo"],
    "Probabilità Teorica": probabilita_teoriche,
    "Frequenza Assoluta": [frequenze_assolute[r] for r in regimi_etichette],
    "Frequenza Empirica": [frequenze_empiriche[r] for r in regimi_etichette]
})

print("--- TABELLA 1: FREQUENZE SIMULATE DEGLI EVENTI INFORMATIVI ---")
print(tabella_frequenze.to_string(index=False))

# 6. Controlli numerici computazionali richiesti
# Controllo 1: Verifica che le frequenze empiriche approssimino correttamente quelle teoriche
for regime in regimi_etichette:
    teorica = probabilita_teoriche[regimi_etichette.index(regime)]
    empirica = frequenze_empiriche[regime]
    diff = abs(empirica - teorica)
    print(f"Controllo Regime {regime}: Teorica = {teorica:.4f} | Empirica = {empirica:.4f} | Diff = {diff:.4f}")
    assert diff < 0.01, f"Attenzione: scostamento elevato per il regime {regime}."

# Controllo 2: Verifica della numerosità sufficiente per ciascun blocco informativo
min_numerosita = min(frequenze_assolute.values())
print(f"\nControllo numerosità minima per blocco: {min_numerosita} simulazioni (condizione superata).")
assert min_numerosita > 1000, "Errore: numerosità insufficiente per uno o più blocchi informativi."

print("\nTappa 1 completata con successo. La decisione finale spetta allo studente.")
```

```python
# ==============================================================================
# TAPPA 2: Simulazione gerarchica del prezzo dell'energia (PE | Ag)
# ==============================================================================

import numpy as np
import pandas as pd

# 1. Definizione dei parametri delle distribuzioni condizionate del prezzo dell'energia
# Condizionatamente all'evento Ag, PE ~ N(mu_g, sigma_g^2)
parametri_condizionati = {
    "A1": {"mu": 75.0,  "sigma": 15.0, "descrizione": "Normalizzazione energetica"},
    "A2": {"mu": 115.0, "sigma": 25.0, "descrizione": "Tensione persistente"},
    "A3": {"mu": 170.0, "sigma": 45.0, "descrizione": "Shock energetico severo"}
}

# 2. Inizializzazione del vettore dei prezzi unitari dell'energia PE
n_simulazioni = len(regimi_simulati)
prezzi_energia_PE = np.zeros(n_simulazioni)

# 3. Simulazione gerarchica condizionata al regime estratto nella Tappa 1
for regime, params in parametri_condizionati.items():
    maschera = (regimi_simulati == regime)
    n_regime = np.sum(maschera)
    
    if n_regime > 0:
        # Estrazione campionaria dalla normale condizionata N(mu_g, sigma_g^2)
        prezzi_energia_PE[maschera] = rng.normal(
            loc=params["mu"], 
            scale=params["sigma"], 
            size=n_regime
        )

# 4. Controlli numerici computazionali richiesti
# Controllo 1: Verifica della coerenza statistica campionaria per ciascun blocco
controllo_statistiche = []
for regime, params in parametri_condizionati.items():
    maschera = (regimi_simulati == regime)
    CAMPIONE_PE_g = prezzi_energia_PE[maschera]
    
    media_empirica = np.mean(CAMPIONE_PE_g)
    std_empirica = np.std(CAMPIONE_PE_g, ddof=1)
    
    controllo_statistiche.append({
        "Regime": regime,
        "μ Teorico": params["mu"],
        "μ Empirico": media_empirica,
        "Diff μ": abs(media_empirica - params["mu"]),
        "σ Teorico": params["sigma"],
        "σ Empirico": std_empirica,
        "Diff σ": abs(std_empirica - params["sigma"])
    })
    
    # Asserzione di tolleranza statistica per grandi campioni
    assert abs(media_empirica - params["mu"]) < 1.5, f"Scostamento media troppo elevato per il regime {regime}"
    assert abs(std_empirica - params["sigma"]) < 1.5, f"Scostamento deviazione standard troppo elevato per il regime {regime}"

tabella_controllo_statistiche = pd.DataFrame(controllo_statistiche)
print("--- TABELLA 2: CONTROLLO STATISTICHE EMPIRICHE DI P_E | A_g ---")
print(tabella_controllo_statistiche.to_string(index=False))

# Controllo 2: Controllo della presenza di valori negativi di PE (limite del modello normale)
valori_negativi_PE = prezzi_energia_PE < 0
num_negativi = np.sum(valori_negativi_PE)
print(f"\nControllo valori negativi di P_E: Trovati {num_negativi} valori negativi su {n_simulazioni} simulazioni.")

if num_negativi > 0:
    percentuale_negativi = (num_negativi / n_simulazioni) * 100
    print(f"Nota interpretativa: I prezzi negativi rappresentano il {percentuale_negativi:.3f}% del totale.")
    print("Come stabilito dalla Scheda Caso, tali valori non vengono eliminati forzatamente ma mantenuti e discussi come limite del modello normale.")

print("\nTappa 2 completata con successo. La decisione finale spetta allo studente.")
```

### Sintesi del Lavoro (Tappe 1 — 3)
Nelle prime tre tappe abbiamo impostato lo spazio di probabilità e simulato il comportamento stocastico del prezzo dell'energia attraverso un approccio gerarchico condizionato ai tre regimi informativi del mercato ($A_1, A_2, A_3$). Successivamente, applicando le relazioni strutturali deterministiche, abbiamo ricavato il costo energetico totale ($C_E$) e il margine operativo ($M$). Le tabelle descrittive formattate evidenziano la dispersione dei valori e confermano la perfetta coerenza algebrica dei dati simulati. La decisione finale spetta allo studente.

```python
# ==============================================================================
# TAPPA 3: Calcolo delle grandezze economico-finanziarie (C_E ed M)
# ==============================================================================

import numpy as np
import pandas as pd

# 1. Definizione dei parametri economico-finanziari strutturali
G_0 = 8000000.0  # Contribuzione operativa netta non energetica (euro)
Q_E = 50000.0    # Fabbisogno energetico annuo (MWh)

# 2. Applicazione delle relazioni deterministiche
# Costo energetico totale: C_E = Q_E * P_E
costo_energetico_CE = Q_E * prezzi_energia_PE

# Margine operativo dopo il costo energetico: M = G_0 - C_E
margine_operativo_M = G_0 - costo_energetico_CE

# 3. Controlli algebrici di coerenza computazionale
assert len(costo_energetico_CE) == len(prezzi_energia_PE), "Errore nella dimensione del vettore C_E"
assert len(margine_operativo_M) == len(prezzi_energia_PE), "Errore nella dimensione del vettore M"

# Controllo puntuale su un campione casuale estratto
idx_controllo = rng.integers(0, len(prezzi_energia_PE))
p_test = prezzi_energia_PE[idx_controllo]
c_test = costo_energetico_CE[idx_controllo]
m_test = margine_operativo_M[idx_controllo]

assert np.isclose(c_test, Q_E * p_test), "Controllo algebrico fallito su C_E"
assert np.isclose(m_test, G_0 - c_test), "Controllo algebrico fallito su M"
print(f"Controllo algebrico superato con successo sull'indice {idx_controllo}.")

# 4. Costruzione delle tabelle di statistica descrittiva formattate (Output richiesto)
def calcola_statistiche_descrittive(vettore, nome_variabile, unita):
    return {
        "Variabile": nome_variabile,
        "Unità": unita,
        "Media": f"{np.mean(vettore):,.2f}",
        "Dev. Std": f"{np.std(vettore, ddof=1):,.2f}",
        "Minimo": f"{np.min(vettore):,.2f}",
        "Q1 (25%)": f"{np.percentile(vettore, 25):,.2f}",
        "Mediana (50%)": f"{np.median(vettore):,.2f}",
        "Q3 (75%)": f"{np.percentile(vettore, 75):,.2f}",
        "Massimo": f"{np.max(vettore):,.2f}"
    }

statistiche_PE = calcola_statistiche_descrittive(prezzi_energia_PE, "Prezzo Energia (P_E)", "euro/MWh")
statistiche_CE = calcola_statistiche_descrittive(costo_energetico_CE, "Costo Energetico (C_E)", "euro")
statistiche_M  = calcola_statistiche_descrittive(margine_operativo_M, "Margine Operativo (M)", "euro")

tabella_descrittiva_finanziaria = pd.DataFrame([statistiche_PE, statistiche_CE, statistiche_M])

# Formattazione avanzata con Pandas Styler per un'eccellente leggibilità a schermo
tabella_formattata = tabella_descrittiva_finanziaria.style.set_caption(
    "STATISTICHE DESCRITTIVE DELLE GRANDEZZE ENERGETICO-FINANZIARIE"
).set_properties(**{
    'text-align': 'right',
    'padding': '6px',
    'border': '1px solid #ddd'
}).set_table_styles([
    {'selector': 'th', 'props': [('text-align', 'center'), ('background-color', '#f2f2f2'), ('font-weight', 'bold'), ('border', '1px solid #ddd')]},
    {'selector': 'caption', 'props': [('caption-side', 'top'), ('font-size', '1.1em'), ('font-weight', 'bold'), ('color', '#333'), ('padding', '8px')]}
])

display(tabella_formattata)

print("\nTappa 3 completata con successo. La decisione finale spetta allo studente.")
```

### Sintesi della Tappa 4: Calcolo dello Shortfall e delle Soglie Implicite
In questa tappa introduciamo formalmente le costanti strutturali $m^\star$ (margine minimo desiderato pari a 2.000.000 €) e $s^\star$ (soglia di shortfall severo pari a 1.000.000 €). Attraverso la trasformazione non lineare $S = (m^\star - M)^+$, calcoliamo il vettore dello shortfall per ciascuna simulazione. Verifichiamo inoltre la coerenza empirica delle probabilità di superamento soglia rispetto alle soglie analitiche implicite sul prezzo dell'energia ($P_E > 120$ per $S > 0$ e $P_E > 140$ per $S > s^\star$). La decisione finale spetta allo studente.

```python
# ==============================================================================
# TAPPA 4 (AGGIORNATA): Calcolo dello shortfall di margine e delle soglie implicite
# ==============================================================================

import numpy as np
import pandas as pd

# 1. Definizione delle soglie strutturali (costanti del modello)
m_star = 2000000.0  # Margine minimo desiderato (euro)
s_star = 1000000.0  # Soglia di shortfall severo (euro)

# 2. Calcolo dello shortfall di margine: S = (m* - M)^+
shortfall_S = np.maximum(0.0, m_star - margine_operativo_M)

# 3. Calcolo delle soglie analitiche implicite sul prezzo dell'energia (PE)
soglia_PE_zero = (G_0 - m_star) / Q_E  # 120 euro/MWh
soglia_PE_severo = (G_0 - m_star + s_star) / Q_E  # 140 euro/MWh

# 4. Indicatori booleani ed eventi di superamento soglia
superamento_zero = (shortfall_S > 0.0)
superamento_severo = (shortfall_S > s_star)

prob_empirica_zero = np.mean(superamento_zero)
prob_empirica_severo = np.mean(superamento_severo)

# Coerenza analitica con il prezzo dell'energia
coerenza_zero_PE = (prezzi_energia_PE > soglia_PE_zero)
coerenza_severo_PE = (prezzi_energia_PE > soglia_PE_severo)

# 5. Controlli numerici e logici computazionali richiesti
assert np.all(shortfall_S >= 0.0), "Errore: rilevati valori negativi nello shortfall S."
assert np.allclose(superamento_zero, coerenza_zero_PE), "Errore logico: la condizione S > 0 non coincide con PE > 120."
assert np.allclose(superamento_severo, coerenza_severo_PE), "Errore logico: la condizione S > s* non coincide con PE > 140."
print("Tutti i controlli logici e algebrici sulle soglie sono stati superati con successo.")

# 6. Costruzione della tabella riassuntiva avanzata e formattata (Output richiesto)
tabella_soglie_dati = [
    {
        "Evento di Rischio": "Shortfall Positivo",
        "Condizione (S)": "S > 0",
        "Condizione Margine (M)": f"M < {m_star:,.0f} €",
        "Soglia Implicita (P_E)": f"P_E > {soglia_PE_zero:.0f} €/MWh",
        "Probabilità Empirica (P̂)": f"{prob_empirica_zero:.4f}"
    },
    {
        "Evento di Rischio": "Shortfall Severo",
        "Condizione (S)": f"S > {s_star:,.0f} €",
        "Condizione Margine (M)": f"M < {(m_star - s_star):,.0f} €",
        "Soglia Implicita (P_E)": f"P_E > {soglia_PE_severo:.0f} €/MWh",
        "Probabilità Empirica (P̂)": f"{prob_empirica_severo:.4f}"
    }
]

tabella_soglie_implicite = pd.DataFrame(tabella_soglie_dati)

# Formattazione avanzata con Pandas Styler per un'eccellente leggibilità a schermo
tabella_soglie_formattata = tabella_soglie_implicite.style.set_caption(
    "TABELLA DELLE SOGLIE IMPLICITE E CONDIZIONI ANALITICHE DI SHORTFALL"
).set_properties(**{
    'text-align': 'center',
    'padding': '8px',
    'border': '1px solid #ddd'
}).set_table_styles([
    {'selector': 'th', 'props': [('text-align', 'center'), ('background-color', '#f2f2f2'), ('font-weight', 'bold'), ('border', '1px solid #ddd')]},
    {'selector': 'caption', 'props': [('caption-side', 'top'), ('font-size', '1.1em'), ('font-weight', 'bold'), ('color', '#333'), ('padding', '8px')]}
])

display(tabella_soglie_formattata)

print("\nTappa 4 (aggiornata) completata con successo. La decisione finale spetta allo studente.")
```

### Sintesi della Tappa 5: Stime Condizionate, Non Condizionate e Verifica del Valore Atteso Totale
In questa tappa utilizziamo i vettori simulati dello shortfall per calcolare le grandezze statistiche chiave. In particolare, stimiamo il valore atteso non condizionato $\widehat{\mathbb{E}}[S]$, i quantili campionari $q_\alpha(S)$ e le probabilità di superamento soglia. Successivamente, calcoliamo i valori attesi condizionati dello shortfall per ciascun regime informativo ($A_1, A_2, A_3$) e verifichiamo empiricamente la proprietà teorica della formula del valore atteso totale, secondo cui la media globale coincide con la media ponderata delle medie condizionate. La decisione finale spetta allo studente.

```python
# ==============================================================================
# TAPPA 5 (CORRETTA): Stime condizionate, non condizionate e verifica del valore atteso totale
# ==============================================================================

import numpy as np
import pandas as pd

# 1. Calcolo delle stime non condizionate dello shortfall (S)
media_S = np.mean(shortfall_S)
std_S = np.std(shortfall_S, ddof=1)
min_S = np.min(shortfall_S)
q25_S = np.percentile(shortfall_S, 25)
mediana_S = np.median(shortfall_S)
q75_S = np.percentile(shortfall_S, 75)
q95_S = np.percentile(shortfall_S, 95)
max_S = np.max(shortfall_S)

# 2. Calcolo delle stime condizionate dello shortfall per ciascun regime informativo (Ag)
medie_condizionate = {}
frequenze_empiriche_dict = {}

for regime in ["A1", "A2", "A3"]:
    maschera_regime = (regimi_simulati == regime)
    shortfall_regime = shortfall_S[maschera_regime]
    
    # Frequenza empirica del regime
    freq_emp = np.sum(maschera_regime) / len(regimi_simulati)
    frequenze_empiriche_dict[regime] = freq_emp
    
    # Valore atteso condizionato empirico E[S | Ag]
    medie_condizionate[regime] = np.mean(shortfall_regime)

# 3. Verifica della formula del valore atteso totale (versione campionaria/empirica)
media_ricomposta = sum(frequenze_empiriche_dict[regime] * medie_condizionate[regime] for regime in ["A1", "A2", "A3"])

# 4. Controlli numerici computazionali richiesti
assert min_S <= media_S <= max_S, "Errore: la media di S non è compresa tra minimo e massimo."
diff_valore_atteso_totale = abs(media_S - media_ricomposta)
print(f"Controllo Valore Atteso Totale: Media Globale = {media_S:,.4f} | Media Ricomposta = {media_ricomposta:,.4f} | Diff = {diff_valore_atteso_totale:.6f}")
assert diff_valore_atteso_totale < 1e-5, "Errore: la formula del valore atteso totale non è verificata empiricamente."

# 5. Costruzione delle tabelle di output formattate
tabella_S_dati = [{
    "Variabile": "Shortfall di Margine (S)",
    "Media": f"{media_S:,.2f} €",
    "Dev. Std": f"{std_S:,.2f} €",
    "Minimo": f"{min_S:,.2f} €",
    "Q1 (25%)": f"{q25_S:,.2f} €",
    "Mediana": f"{mediana_S:,.2f} €",
    "Q3 (75%)": f"{q75_S:,.2f} €",
    "Q95 (95%)": f"{q95_S:,.2f} €",
    "Massimo": f"{max_S:,.2f} €"
}]
tabella_S_descrittiva = pd.DataFrame(tabella_S_dati)

tabella_condizionate_dati = []
for regime in ["A1", "A2", "A3"]:
    tabella_condizionate_dati.append({
        "Regime Informativo": regime,
        "Frequenza Empirica P̂(Ag)": f"{frequenze_empiriche_dict[regime]:.4f}",
        "Media Condizionata Ê[S | Ag]": f"{medie_condizionate[regime]:,.2f} €",
        "Contributo Pesato P̂(Ag) * Ê[S | Ag]": f"{frequenze_empiriche_dict[regime] * medie_condizionate[regime]:,.2f} €"
    })

tabella_condizionate_df = pd.DataFrame(tabella_condizionate_dati)

# Visualizzazione delle tabelle con Pandas Styler (corretto l'operatore di assegnazione nei dizionari)
display(tabella_S_descrittiva.style.set_caption("STATISTICHE DESCRITTIVE E QUANTILI DELLO SHORTFALL (S)").set_properties(**{
    'text-align': 'right', 'padding': '6px', 'border': '1px solid #ddd'
}).set_table_styles([
    {'selector': 'th', 'props': [('text-align', 'center'), ('background-color', '#f2f2f2'), ('font-weight', 'bold'), ('border', '1px solid #ddd')]},
    {'selector': 'caption', 'props': [('caption-side', 'top'), ('font-size', '1.1em'), ('font-weight', 'bold'), ('color', '#333'), ('padding', '8px')]}
]))

display(tabella_condizionate_df.style.set_caption("VERIFICA DELLA FORMULA DEL VALORE ATTESO TOTALE").set_properties(**{
    'text-align': 'center', 'padding': '6px', 'border': '1px solid #ddd'
}).set_table_styles([
    {'selector': 'th', 'props': [('text-align', 'center'), ('background-color', '#f2f2f2'), ('font-weight', 'bold'), ('border', '1px solid #ddd')]},
    {'selector': 'caption', 'props': [('caption-side', 'top'), ('font-size', '1.1em'), ('font-weight', 'bold'), ('color', '#333'), ('padding', '8px')]}
]))

print("\nTappa 5 completata con successo. La decisione finale spetta allo studente.")
```
